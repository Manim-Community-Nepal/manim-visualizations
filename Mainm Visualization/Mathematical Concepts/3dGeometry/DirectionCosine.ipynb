# 📚 Summary Notes: Direction Cosines and Ratios

## 1. Core Concepts: DRs vs. DCs

To define the orientation of a line in 3D space, we look at its vector components relative to the origin.



### **A. Direction Ratios (DRs)**
* **Symbol:** $a, b, c$
* **Definition:** The raw components of the direction vector. They represent the relative movement along the $X, Y, Z$ axes. They are **not** unique (proportional values represent the same direction).
* **Example Vector ($L_1$):** $\vec{v} = 2\hat{i} + 2\hat{j} + 1\hat{k}$
    * **DRs:** $a=2, \quad b=2, \quad c=1$

### **B. Direction Cosines (DCs)**
* **Symbol:** $l, m, n$
* **Definition:** The components of the **Unit Vector** ($\hat{u}$). They correspond to the cosines of the actual angles ($\alpha, \beta, \gamma$) the line makes with the positive axes.
* **Formula:** Divide the DRs by the magnitude ($r$).

$$
r = \sqrt{a^2 + b^2 + c^2}
$$

$$
l = \frac{a}{r} = \cos \alpha, \quad m = \frac{b}{r} = \cos \beta, \quad n = \frac{c}{r} = \cos \gamma
$$

> **Fundamental Identity:** > $$l^2 + m^2 + n^2 = 1$$

---

## 2. Numerical Example Calculation

Using our primary example line $L_1$: **DRs** $(2, 2, 1)$.

1.  **Calculate Magnitude:**
    $$r = \sqrt{2^2 + 2^2 + 1^2} = \sqrt{4+4+1} = \sqrt{9} = 3$$

2.  **Calculate DCs:**
    $$l = \frac{2}{3}, \quad m = \frac{2}{3}, \quad n = \frac{1}{3}$$

3.  **Verify Identity:**
    $$\left(\frac{2}{3}\right)^2 + \left(\frac{2}{3}\right)^2 + \left(\frac{1}{3}\right)^2 = \frac{4}{9} + \frac{4}{9} + \frac{1}{9} = \frac{9}{9} = 1$$

---

## 3. Angle Between Two Lines

To find the angle $\theta$ between two lines, we use the **Dot Product** of their direction ratios.

$$
\cos \theta = \left| \frac{a_1 a_2 + b_1 b_2 + c_1 c_2}{\sqrt{a_1^2 + b_1^2 + c_1^2} \sqrt{a_2^2 + b_2^2 + c_2^2}} \right|
$$

### **Example Calculation:**
* **Line 1:** $(2, 2, 1)$
* **Line 2:** $(1, -2, 2)$

$$
\text{Numerator} = (2)(1) + (2)(-2) + (1)(2) = 2 - 4 + 2 = 0
$$

$$
\cos \theta = \frac{0}{3 \times 3} = 0 \implies \theta = 90^\circ
$$

---

## 4. Critical Conditions

### **A. Condition for Perpendicularity ($\theta = 90^\circ$)**
Two lines are perpendicular if their dot product is zero.

$$
a_1 a_2 + b_1 b_2 + c_1 c_2 = 0
$$

* **Visual Check:** The projection of one vector onto the other is zero.

### **B. Condition for Parallelism ($\theta = 0^\circ$)**
Two lines are parallel if their direction ratios are proportional (one is a scalar multiple of the other).


$$
\frac{a_1}{a_2} = \frac{b_1}{b_2} = \frac{c_1}{c_2}
$$

* **Hypothetical Example:** Comparing $L_1(2,2,1)$ with $L_3(4,4,2)$:
    $$\frac{2}{4} = \frac{2}{4} = \frac{1}{2} \quad (\text{Ratios are equal})$$


```python
from manim import *
import numpy as np

# --- Global Constants for Styling ---
X_COLOR = RED_C
Y_COLOR = GREEN_C
Z_COLOR = BLUE_C
LINE_1_COLOR = YELLOW
LINE_2_COLOR = ORANGE

# Styling Constants - Optimized for layout
TEXT_SCALE_HEADER = 0.75 
TEXT_SCALE_BODY = 0.55
TEXT_SCALE_MATH = 0.60
LEFT_PANEL_WIDTH = 5.0
SIDE_PANEL_X_POS = 6.0 

class DirectionCosinesAndRatios(ThreeDScene):
    def construct(self):
        # 1. Setup Layout
        self.setup_layout()

        # 2. Section: Direction Cosines (DCs)
        self.explain_direction_cosines()
        self.wait(1)

        # 3. Section: Direction Ratios (DRs) - WITH CAMERA ROTATION
        self.explain_direction_ratios()
        self.wait(1)

        # 4. Section: Angle Between Lines
        self.explain_angle_between_lines()
        self.wait(1)

        # 5. Section: Conditions (Perp & Para)
        self.explain_conditions()
        self.wait(3)

    def setup_layout(self):
        # Shift axes to the RIGHT to make room for text on LEFT
        self.axes = ThreeDAxes(
            x_range=[-1, 5, 1],
            y_range=[-1, 5, 1],
            z_range=[-1, 4, 1],
            x_length=5,
            y_length=5,
            z_length=3.5,
            axis_config={"include_tip": True, "stroke_width": 2}
        ).shift(RIGHT * 3.5)

        # Color the axes tips
        self.axes.x_axis.set_color(X_COLOR)
        self.axes.y_axis.set_color(Y_COLOR)
        self.axes.z_axis.set_color(Z_COLOR)
        
        x_label = self.axes.get_x_axis_label("X").set_color(X_COLOR)
        y_label = self.axes.get_y_axis_label("Y").set_color(Y_COLOR)
        z_label = self.axes.get_z_axis_label("Z").set_color(Z_COLOR)
        
        self.axis_labels = VGroup(x_label, y_label, z_label)
        self.set_camera_orientation(phi=75 * DEGREES, theta=-30 * DEGREES)
        self.add(self.axes, self.axis_labels)

        # Prepare Text Panel
        self.left_panel_group = VGroup()
        self.panel_anchor = UP * 3.5 + LEFT * 6.5

    def add_to_left_panel(self, mobject, buff=0.25):
        if len(self.left_panel_group) == 0:
            mobject.move_to(self.panel_anchor, aligned_edge=UL)
        else:
            mobject.next_to(self.left_panel_group[-1], DOWN, buff=buff, aligned_edge=LEFT)
        
        if mobject.width > LEFT_PANEL_WIDTH:
            scale_factor = LEFT_PANEL_WIDTH / mobject.width
            mobject.scale(scale_factor)
            
        self.left_panel_group.add(mobject)
        self.add_fixed_in_frame_mobjects(mobject)
        return mobject

    def clear_left_panel(self):
        self.remove(*self.left_panel_group)
        self.left_panel_group = VGroup()

    # --- Section 1: Direction Cosines ---
    def explain_direction_cosines(self):
        header = Text("1. Direction Cosines (DCs)", weight=BOLD).scale(TEXT_SCALE_HEADER)
        self.add_to_left_panel(header)

        # Unit Vector components
        vec_coords_unit = np.array([0.5, 0.707, 0.5])
        # Scale the vector to be longer for better visuals
        visual_scale = 2.5
        vec_coords_visual = vec_coords_unit * visual_scale

        origin = self.axes.c2p(0,0,0)
        end_point = self.axes.c2p(*vec_coords_visual)

        unit_vector_line = Arrow3D(
            start=origin,
            end=end_point,
            color=LINE_1_COLOR,
            thickness=0.02
        )

        def_angles_text = VGroup(
            Tex(r"Angles with positive axes:"),
            Tex(r"$\alpha$ with $X$-axis", color=X_COLOR),
            Tex(r"$\beta$ with $Y$-axis", color=Y_COLOR),
            Tex(r"$\gamma$ with $Z$-axis", color=Z_COLOR),
        ).scale(TEXT_SCALE_BODY).arrange(DOWN, aligned_edge=LEFT, buff=0.1)
        self.add_to_left_panel(def_angles_text)

        self.play(Create(unit_vector_line))

        # --- DRAWING CURVES (ARCS) FOR ANGLES ---
        radius = 1.5
        px = self.axes.c2p(radius, 0, 0)
        py = self.axes.c2p(0, radius, 0)
        pz = self.axes.c2p(0, 0, radius)
        
        v_scaled_for_arc = vec_coords_visual * (radius / np.linalg.norm(vec_coords_visual))
        pv = self.axes.c2p(*v_scaled_for_arc)

        curve_alpha = ArcBetweenPoints(px, pv, angle=PI/4, color=X_COLOR)
        label_alpha = MathTex(r"\alpha", color=X_COLOR).scale(0.8).move_to(curve_alpha.get_center() + RIGHT*0.2 + UP*0.1)
        
        curve_beta = ArcBetweenPoints(py, pv, angle=PI/4, color=Y_COLOR)
        label_beta = MathTex(r"\beta", color=Y_COLOR).scale(0.8).move_to(curve_beta.get_center() + UP*0.2)

        curve_gamma = ArcBetweenPoints(pz, pv, angle=PI/4, color=Z_COLOR)
        label_gamma = MathTex(r"\gamma", color=Z_COLOR).scale(0.8).move_to(curve_gamma.get_center() + LEFT*0.2)
        
        for lbl in [label_alpha, label_beta, label_gamma]:
             lbl.rotate(90*DEGREES, axis=RIGHT)

        self.play(
            Create(curve_alpha), Write(label_alpha),
            Create(curve_beta), Write(label_beta),
            Create(curve_gamma), Write(label_gamma)
        )
        
        self.move_camera(theta=-10 * DEGREES, run_time=2)

        # Text: Define Cosines
        def_cosines_text = VGroup(
            Tex(r"\textbf{Direction Cosines }($l, m, n$):"),
            Tex(r"$l = \cos \alpha$", color=X_COLOR),
            Tex(r"$m = \cos \beta$", color=Y_COLOR),
            Tex(r"$n = \cos \gamma$", color=Z_COLOR),
        ).scale(TEXT_SCALE_BODY).arrange(DOWN, aligned_edge=LEFT, buff=0.1)
        self.add_to_left_panel(def_cosines_text, buff=0.4)
        self.play(Write(def_cosines_text))

        # Identity
        identity_header = Tex(r"\textbf{Fundamental Identity:}").scale(TEXT_SCALE_BODY)
        identity_math = MathTex(r"l^2 + m^2 + n^2 = 1", color=YELLOW).scale(TEXT_SCALE_HEADER)
        
        self.add_to_left_panel(identity_header, buff=0.4)
        self.add_to_left_panel(identity_math, buff=0.15)
        self.play(Write(identity_header), Write(identity_math))

        # Cleanup Visuals
        self.vec_v1_coords = vec_coords_unit 
        self.line1_3d = unit_vector_line
        self.play(
            FadeOut(curve_alpha), FadeOut(label_alpha),
            FadeOut(curve_beta), FadeOut(label_beta),
            FadeOut(curve_gamma), FadeOut(label_gamma)
        )

    # --- Section 2: Direction Ratios (UPDATED WITH CAMERA ROTATION) ---
    def explain_direction_ratios(self):
        self.clear_left_panel()
        header = Text("2. Direction Ratios (DRs)", weight=BOLD).scale(TEXT_SCALE_HEADER)
        self.add_to_left_panel(header)

        dr_def = VGroup(
            Tex(r"Numbers $a, b, c$ proportional"),
            Tex(r"to Direction Cosines."),
            MathTex(r"\frac{l}{a} = \frac{m}{b} = \frac{n}{c} = k")
        ).scale(TEXT_SCALE_BODY).arrange(DOWN, aligned_edge=LEFT, buff=0.1)
        self.add_to_left_panel(dr_def)
        self.play(Write(dr_def))

        # Scale up vector
        scale_factor = 3.0
        dr_coords = self.vec_v1_coords * scale_factor
        
        long_vector_line = Arrow3D(
            start=self.axes.c2p(0, 0, 0),
            end=self.axes.c2p(*dr_coords),
            color=LINE_1_COLOR,
            thickness=0.02
        )
        
        self.play(Transform(self.line1_3d, long_vector_line))

        # --- VISUALIZE COMPONENTS WITH ROTATING CAMERA ---
        
        # 1. Visualize X-Axis Component
        # Rotate camera to align roughly with X-projection view
        self.move_camera(phi=60*DEGREES, theta=-10*DEGREES, run_time=1.5)
        
        p_tip = self.axes.c2p(*dr_coords)
        p_x_comp = self.axes.c2p(dr_coords[0], 0, 0)
        
        drop_line_x = DashedLine(p_tip, p_x_comp, color=GRAY)
        comp_line_x = Line(self.axes.c2p(0,0,0), p_x_comp, color=X_COLOR, stroke_width=4)
        label_a = MathTex("a", color=X_COLOR).move_to(comp_line_x.get_center() + DOWN*0.3).rotate(90*DEGREES, axis=RIGHT)
        
        self.play(Create(drop_line_x), Create(comp_line_x), Write(label_a))
        self.wait(1)
        self.play(FadeOut(drop_line_x), FadeOut(comp_line_x), FadeOut(label_a))
        
        # 2. Visualize Y-Axis Component
        # Rotate camera to align with Y-projection view
        self.move_camera(phi=60*DEGREES, theta=-100*DEGREES, run_time=1.5)
        
        p_y_comp = self.axes.c2p(0, dr_coords[1], 0)
        drop_line_y = DashedLine(p_tip, p_y_comp, color=GRAY)
        comp_line_y = Line(self.axes.c2p(0,0,0), p_y_comp, color=Y_COLOR, stroke_width=4)
        label_b = MathTex("b", color=Y_COLOR).move_to(comp_line_y.get_center() + RIGHT*0.3).rotate(90*DEGREES, axis=RIGHT)

        self.play(Create(drop_line_y), Create(comp_line_y), Write(label_b))
        self.wait(1)
        self.play(FadeOut(drop_line_y), FadeOut(comp_line_y), FadeOut(label_b))

        # 3. Visualize Z-Axis Component
        # Rotate camera to a low angle ("ground view") to see height
        self.move_camera(phi=85*DEGREES, theta=-45*DEGREES, run_time=1.5)
        
        p_z_comp = self.axes.c2p(0, 0, dr_coords[2])
        drop_line_z = DashedLine(p_tip, p_z_comp, color=GRAY)
        comp_line_z = Line(self.axes.c2p(0,0,0), p_z_comp, color=Z_COLOR, stroke_width=4)
        label_c = MathTex("c", color=Z_COLOR).move_to(comp_line_z.get_center() + LEFT*0.3).rotate(90*DEGREES, axis=RIGHT)

        self.play(Create(drop_line_z), Create(comp_line_z), Write(label_c))
        self.wait(1)
        self.play(FadeOut(drop_line_z), FadeOut(comp_line_z), FadeOut(label_c))

        # Reset Camera for Text Formula
        self.move_camera(phi=75 * DEGREES, theta=-30 * DEGREES, run_time=1.5)

        # Conversion Formula Text
        conv_header = Tex(r"\textbf{Converting DRs to DCs:}").scale(TEXT_SCALE_BODY)
        self.add_to_left_panel(conv_header, buff=0.4)
        
        magnitude_tex = MathTex(r"Magnitude = \sqrt{a^2+b^2+c^2}").scale(TEXT_SCALE_MATH)
        self.add_to_left_panel(magnitude_tex)

        eq1 = MathTex(r"l = \pm \frac{a}{\sqrt{a^2+b^2+c^2}}")
        eq2 = MathTex(r"m = \pm \frac{b}{\sqrt{a^2+b^2+c^2}}")
        eq3 = MathTex(r"n = \pm \frac{c}{\sqrt{a^2+b^2+c^2}}")
        eq1.set_color_by_tex("a", X_COLOR)
        eq2.set_color_by_tex("b", Y_COLOR)
        eq3.set_color_by_tex("c", Z_COLOR)

        conv_formula = VGroup(eq1, eq2, eq3).scale(TEXT_SCALE_MATH - 0.1).arrange(DOWN, aligned_edge=LEFT, buff=0.15)

        self.add_to_left_panel(conv_formula)
        self.play(Write(conv_header), Write(magnitude_tex), Write(conv_formula))
        self.wait(2)
        
        # Cleanup
        new_v1_drs = np.array([2, 2, 1])
        self.v1_drs = new_v1_drs
        self.v1_mag = np.linalg.norm(new_v1_drs)
        self.v1_dcs = new_v1_drs / self.v1_mag

        new_line1 = Arrow3D(
            start=self.axes.c2p(0, 0, 0),
            end=self.axes.c2p(*new_v1_drs),
            color=LINE_1_COLOR,
            thickness=0.02
        )
        self.play(Transform(self.line1_3d, new_line1))

    # --- Section 3: Angle Between Lines ---
    def explain_angle_between_lines(self):
        self.clear_left_panel()
        header = Text("3. Angle Between Lines", weight=BOLD).scale(TEXT_SCALE_HEADER)
        self.add_to_left_panel(header)

        self.v2_drs = np.array([1, -2, 2])
        self.v2_mag = np.linalg.norm(self.v2_drs)
        self.v2_dcs = self.v2_drs / self.v2_mag

        self.line2_3d = Arrow3D(
            start=self.axes.c2p(0, 0, 0),
            end=self.axes.c2p(*self.v2_drs),
            color=LINE_2_COLOR,
            thickness=0.02
        )

        l1_data = VGroup(
            Tex(r"Line 1 (Yellow):", color=LINE_1_COLOR),
            MathTex(f"DRs = ({self.v1_drs[0]}, {self.v1_drs[1]}, {self.v1_drs[2]})").scale(TEXT_SCALE_MATH)
        ).arrange(DOWN, aligned_edge=LEFT)
        self.add_to_left_panel(l1_data)

        self.play(Create(self.line2_3d))
        l2_data = VGroup(
            Tex(r"Line 2 (Orange):", color=LINE_2_COLOR),
            MathTex(f"DRs = ({self.v2_drs[0]}, {self.v2_drs[1]}, {self.v2_drs[2]})").scale(TEXT_SCALE_MATH)
        ).arrange(DOWN, aligned_edge=LEFT)
        self.add_to_left_panel(l2_data, buff=0.25)
        self.play(Write(l1_data), Write(l2_data))

        p1_scaled = self.axes.c2p(*(self.v1_dcs * 1.5))
        p2_scaled = self.axes.c2p(*(self.v2_dcs * 1.5))
        self.angle_arc = CurvedArrow(p1_scaled, p2_scaled, angle=PI/4, color=WHITE, radius=2)
        self.angle_label = MathTex(r"\theta").move_to(self.angle_arc.get_center() + OUT*0.2).scale(0.7)
        self.angle_label.rotate(90 * DEGREES, axis=RIGHT)

        self.play(Create(self.angle_arc), Write(self.angle_label))

        formula_header = Tex(r"\textbf{Angle Formula:}").scale(TEXT_SCALE_BODY)
        self.add_to_left_panel(formula_header, buff=0.4)

        formula_tex = MathTex(
            r"\cos \theta = \frac{|a_1 a_2 + b_1 b_2 + c_1 c_2|}{\sqrt{a_1^2+\dots}\sqrt{a_2^2+\dots}}"
        ).scale(TEXT_SCALE_MATH)
        self.add_to_left_panel(formula_tex)
        self.play(Write(formula_tex))

        calc_tex = MathTex(
            r"\cos \theta = \frac{|(2)(1) + (2)(-2) + (1)(2)|}{3 \cdot 3} = 0",
        ).scale(TEXT_SCALE_MATH)
        
        self.add_to_left_panel(calc_tex, buff=0.2)
        self.play(Write(calc_tex))
        self.wait(2)

    # --- Section 4: Conditions ---
    def explain_conditions(self):
        self.clear_left_panel()
        header = Text("4. Conditions", weight=BOLD).scale(TEXT_SCALE_HEADER)
        self.add_to_left_panel(header)

        perp_text_group = VGroup()

        perp_header = Tex(r"\textbf{A. Perpendicularity }($\theta=90^\circ$)", color=RED).scale(TEXT_SCALE_BODY)
        self.add_to_left_panel(perp_header, buff=0.25)
        perp_text_group.add(perp_header)
        
        perp_cond = MathTex(r"a_1 a_2 + b_1 b_2 + c_1 c_2 = 0").scale(TEXT_SCALE_MATH)
        self.add_to_left_panel(perp_cond)
        perp_text_group.add(perp_cond)

        perp_example = VGroup(
            Tex("Example (Dot product = 0):"),
            MathTex(r"2(1) + 2(-2) + 1(2) = 0", color=RED)
        ).scale(TEXT_SCALE_MATH).arrange(DOWN, aligned_edge=LEFT)
        self.add_to_left_panel(perp_example, buff=0.15)
        perp_text_group.add(perp_example)
        
        self.play(FadeOut(self.angle_arc), FadeOut(self.angle_label))
        
        # --- PERPENDICULAR SYMBOL ---
        v1 = np.array([2, 2, 1])
        v2 = np.array([1, -2, 2])
        v1_u = v1 / np.linalg.norm(v1)
        v2_u = v2 / np.linalg.norm(v2)
        size = 0.5
        
        p1 = self.axes.c2p(*(v1_u * size))
        p2 = self.axes.c2p(*(v2_u * size))
        corner_vec = (v1_u * size) + (v2_u * size)
        p_corner = self.axes.c2p(*corner_vec)
        
        symbol_leg1 = Line(p1, p_corner, color=RED, stroke_width=4)
        symbol_leg2 = Line(p2, p_corner, color=RED, stroke_width=4)
        perp_symbol = VGroup(symbol_leg1, symbol_leg2)

        self.play(Write(perp_text_group), Create(perp_symbol))
        self.wait(3)

        # 4B. TRANSITION TO PARALLELISM
        self.play(
            FadeOut(perp_text_group),
            FadeOut(perp_symbol)
        )

        new_v2_drs = np.array([4, 4, 2])
        parallel_line_2 = Arrow3D(
            start=self.axes.c2p(0, 0, 0),
            end=self.axes.c2p(*new_v2_drs),
            color=LINE_2_COLOR,
            thickness=0.03 
        )

        self.play(Transform(self.line2_3d, parallel_line_2))
        self.move_camera(theta=-30 * DEGREES)
        
        para_header = Tex(r"\textbf{B. Parallelism }($0^\circ / 180^\circ$)", color=BLUE).scale(TEXT_SCALE_BODY)
        self.clear_left_panel()
        self.add_to_left_panel(header)
        
        self.add_to_left_panel(para_header, buff=0.4)
        
        para_cond = MathTex(
            r"\frac{a_1}{a_2} = \frac{b_1}{b_2} = \frac{c_1}{c_2}"
        ).scale(TEXT_SCALE_MATH)
        self.add_to_left_panel(para_cond)

        para_example = VGroup(
            Tex("New Example:"),
            MathTex(r"L_1: (2, 2, 1)", color=LINE_1_COLOR),
            MathTex(r"L_2: (4, 4, 2)", color=LINE_2_COLOR),
            MathTex(r"\frac{2}{4} = \frac{2}{4} = \frac{1}{2}", color=BLUE)
        ).scale(TEXT_SCALE_MATH).arrange(DOWN, aligned_edge=LEFT)
        self.add_to_left_panel(para_example, buff=0.15)

        self.play(Write(para_header), Write(para_cond))
        self.play(Write(para_example))
        
        self.move_camera(phi=80*DEGREES, theta=10*DEGREES, run_time=4)
        self.wait()


%manim -qk -v warning DirectionCosinesAndRatios
```
