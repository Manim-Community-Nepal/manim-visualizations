# Visualizing Plane Equations: Summary Notes

This notebook summarizes the four primary forms of plane equations in 3D space, along with the specific numerical examples visualized in the Manim animation.

## 1. General Form
The most common representation of a plane.

$$ax + by + cz + d = 0$$

### Key Components
* **Normal Vector:** $\vec{n} = (a, b, c)$ is perpendicular to the surface.
* **Position:** The scalar $d$ determines the plane's distance/position relative to the origin.

### Animation Examples
| Description | Equation | Normal $\vec{n}$ |
| :--- | :--- | :--- |
| **Standard Plane** | $x + y + z - 3 = 0$ | $(1, 1, 1)$ |
| **Vertical (Missing y)** | $x + z - 2 = 0$ | $(1, 0, 1)$ |
| **Horizontal Plane** | $z - 2 = 0$ | $(0, 0, 1)$ |
| **Through Z-axis** | $x - 2y = 0$ | $(1, -2, 0)$ |

---

## 2. Normal Form
Defines a plane by its orientation (direction cosines) and its perpendicular distance from the origin.

$$lx + my + nz = p$$

### Key Components
* **Distance ($p$):** The perpendicular distance from the origin to the plane.
* **Direction Cosines ($l, m, n$):** Components of the unit normal vector, where $l^2 + m^2 + n^2 = 1$.

### Animation Examples
1.  **Balanced Plane:**
    $$\frac{1}{\sqrt{3}}x + \frac{1}{\sqrt{3}}y + \frac{1}{\sqrt{3}}z = 2$$
    * Distance $p = 2$.

2.  **Flat at $z=3$:**
    $$0x + 0y + 1z = 3$$
    * Distance $p = 3$. Normal points straight up.

3.  **Vertical at $x=2$:**
    $$1x + 0y + 0z = 2$$
    * Distance $p = 2$. Normal points along x-axis.

4.  **Angled Vertical:**
    $$\frac{1}{\sqrt{2}}x + \frac{1}{\sqrt{2}}y = 2$$
    * Distance $p = 2$. Tilted $45^\circ$ between x and y axes.

---

## 3. Intercept Form
Defines the plane by the points where it intersects the coordinate axes.

$$\frac{x}{a} + \frac{y}{b} + \frac{z}{c} = 1$$

### Key Components
* $a$: **x-intercept**
* $b$: **y-intercept**
* $c$: **z-intercept**

### Animation Examples
| Description | Equation | Intercepts $(a, b, c)$ |
| :--- | :--- | :--- |
| **Standard** | $\frac{x}{2} + \frac{y}{2} + \frac{z}{2} = 1$ | $(2, 2, 2)$ |
| **Mixed Signs** | $\frac{x}{4} + \frac{y}{-2} + \frac{z}{3} = 1$ | $(4, -2, 3)$ |
| **Elongated Y** | $\frac{x}{1} + \frac{y}{4} + \frac{z}{1} = 1$ | $(1, 4, 1)$ |
| **Negative X, Y** | $\frac{x}{-2} + \frac{y}{-2} + \frac{z}{2} = 1$ | $(-2, -2, 2)$ |

---

## 4. Point-Normal Form
Defines a plane using a specific fixed point and a normal vector.

$$a(x - x_1) + b(y - y_1) + c(z - z_1) = 0$$

### Key Components
* **Fixed Point:** $P(x_1, y_1, z_1)$
* **Normal Vector:** $\vec{n} = (a, b, c)$

### Animation Examples
1.  **Centered:**
    $$1(x-1) + 1(y-1) + 1(z-1) = 0$$
    * Point: $(1, 1, 1)$, Normal: $(1, 1, 1)$

2.  **XY Plane via Origin:**
    $$0(x-0) + 0(y-0) + 1(z-0) = 0$$
    * Point: $(0, 0, 0)$, Normal: $(0, 0, 1)$

3.  **YZ Plane Shifted:**
    $$1(x-2) + 0(y-0) + 0(z-0) = 0$$
    * Point: $(2, 0, 0)$, Normal: $(1, 0, 0)$

4.  **Tilted:**
    $$1(x) + 2(y) + 1(z-2) = 0$$
    * Point: $(0, 0, 2)$, Normal: $(1, 2, 1)$

```python
from manim import *
import numpy as np

class PlaneEquations(ThreeDScene):
    def construct(self):
        # ---------------------------------------------------------
        # 0. CONFIGURATION & LAYOUT
        # ---------------------------------------------------------
        self.camera.background_color = "#0B0F19"
        
        # Layout: Text panel on the LEFT, 3D scene on the RIGHT.
        SCENE_OFFSET = RIGHT * 3.5
        
        # Fixed Title
        title = Title("Visualizing Plane Equations", font_size=36, color=BLUE_C).to_edge(UP)
        self.add_fixed_in_frame_mobjects(title)

        # 3D Axes
        axes = ThreeDAxes(
            x_range=[-5, 5, 1],
            y_range=[-5, 5, 1],
            z_range=[-5, 5, 1],
            x_length=6.5, y_length=6.5, z_length=6.5,
            axis_config={"include_tip": True, "tip_shape": StealthTip}
        ).shift(SCENE_OFFSET)
        
        x_lbl = axes.get_x_axis_label("x")
        y_lbl = axes.get_y_axis_label("y")
        z_lbl = axes.get_z_axis_label("z")
        
        # Initial Standard Isometric View
        self.set_camera_orientation(phi=70 * DEGREES, theta=30 * DEGREES)
        
        self.play(Create(axes), Write(x_lbl), Write(y_lbl), Write(z_lbl))
        
        # ---------------------------------------------------------
        # 1. GENERAL FORM: ax + by + cz + d = 0
        # ---------------------------------------------------------
        self.next_section("General Form")
        
        # --- Left Text Panel ---
        t1_header = Text("1. General Form", font_size=24, color=YELLOW).to_edge(LEFT).shift(UP*1.5)
        t1_eq = MathTex("{{a}}x + {{b}}y + {{c}}z + {{d}} = 0", font_size=32).next_to(t1_header, DOWN, aligned_edge=LEFT)
        t1_eq.set_color_by_tex_to_color_map({"a": RED, "b": GREEN, "c": BLUE, "d": ORANGE})
        
        t1_list = VGroup(
            Tex(r"$\bullet$ Normal $\vec{n}$ components:", font_size=22),
            Tex(r"$\bullet$ $a$ (x), $b$ (y), $c$ (z)", font_size=22),
            Tex(r"$\bullet$ $d$ determines position", font_size=22)
        ).arrange(DOWN, aligned_edge=LEFT).next_to(t1_eq, DOWN, buff=0.3, aligned_edge=LEFT)
        
        self.add_fixed_in_frame_mobjects(t1_header, t1_eq, t1_list)
        self.play(Write(t1_header), Write(t1_eq), FadeIn(t1_list))
        
        # --- 3D Animation: Decomposing the Normal ---
        val_a, val_b, val_c = 2.8, 2.2, 1.8
        LABEL_SIZE = 32
        ARROW_THICKNESS = 0.05

        # 1. Show components
        self.move_camera(phi=80*DEGREES, theta=-90*DEGREES, run_time=1.5)
        vec_a = Arrow3D(axes.c2p(0,0,0), axes.c2p(val_a, 0, 0), color=RED, thickness=ARROW_THICKNESS)
        lbl_a = Text("a", color=RED, font_size=LABEL_SIZE).move_to(axes.c2p(val_a/2, 0, 0.6)).rotate(90*DEGREES, RIGHT)
        self.play(Create(vec_a), Write(lbl_a))
        
        self.move_camera(phi=60*DEGREES, theta=0*DEGREES, run_time=1.5)
        vec_b = Arrow3D(axes.c2p(val_a,0,0), axes.c2p(val_a, val_b, 0), color=GREEN, thickness=ARROW_THICKNESS)
        lbl_b = Text("b", color=GREEN, font_size=LABEL_SIZE).move_to(axes.c2p(val_a+0.6, val_b/2, 0))
        self.play(Create(vec_b), Write(lbl_b))
        
        self.move_camera(phi=85*DEGREES, theta=10*DEGREES, run_time=1.5)
        vec_c = Arrow3D(axes.c2p(val_a,val_b,0), axes.c2p(val_a, val_b, val_c), color=BLUE, thickness=ARROW_THICKNESS)
        lbl_c = Text("c", color=BLUE, font_size=LABEL_SIZE).move_to(axes.c2p(val_a, val_b+0.6, val_c/2)).rotate(90*DEGREES, UP)
        self.play(Create(vec_c), Write(lbl_c))
        
        # 4. Resultant Normal and Plane
        self.move_camera(phi=70*DEGREES, theta=30*DEGREES, run_time=1.5)
        n_vec_end = axes.c2p(val_a, val_b, val_c)
        arrow_n = Arrow3D(axes.c2p(0,0,0), n_vec_end, color=WHITE, thickness=0.07)
        lbl_n = MathTex(r"\vec{n}", color=WHITE, font_size=36).next_to(arrow_n, UP)

        self.play(Create(arrow_n), Write(lbl_n))
        self.play(FadeOut(vec_a), FadeOut(lbl_a), FadeOut(vec_b), FadeOut(lbl_b), FadeOut(vec_c), FadeOut(lbl_c))
        
        d_val = -(val_a**2 + val_b**2 + val_c**2)
        plane_1 = Surface(
            lambda u, v: axes.c2p(u, v, (-d_val - val_a*u - val_b*v)/val_c),
            u_range=[1, 3.5], v_range=[1, 3.5],
            fill_opacity=0.4, checkerboard_colors=[RED_D, RED_E], resolution=(12,12)
        )
        self.play(Create(plane_1), run_time=2)
        
        # --- VISUALIZE d ---
        azimuth = np.arctan2(val_b, val_a)
        side_view_theta = azimuth - PI/2
        self.move_camera(phi=90*DEGREES, theta=side_view_theta, run_time=2)

        n_vec_np = np.array([val_a, val_b, val_c])
        z_axis = np.array([0, 0, 1])
        offset_dir = np.cross(n_vec_np, z_axis)
        offset_dir = offset_dir / np.linalg.norm(offset_dir) * 1.0 
        
        p_start = np.array([0,0,0]) + offset_dir
        p_end = np.array([val_a, val_b, val_c]) + offset_dir
        
        d_line = DoubleArrow(
            axes.c2p(*p_start), 
            axes.c2p(*p_end), 
            buff=0, 
            color=ORANGE, 
            tip_length=0.2,
            stroke_width=2
        )
        
        lbl_d_vis = MathTex("d", color=ORANGE, font_size=42).next_to(d_line, UP, buff=0.2)
        lbl_d_vis.rotate(90*DEGREES, axis=RIGHT)
        lbl_d_vis.rotate(side_view_theta + 90*DEGREES, axis=OUT)

        self.play(Create(d_line), Write(lbl_d_vis))
        self.wait(2)

        # Cleanup
        self.play(
            FadeOut(t1_header), FadeOut(t1_eq), FadeOut(t1_list), 
            FadeOut(plane_1), FadeOut(arrow_n), FadeOut(lbl_n),
            FadeOut(d_line), FadeOut(lbl_d_vis)
        )

        # =========================================================
        # >>> GENERAL FORM EXAMPLES
        # =========================================================
        self.run_general_examples(axes, SCENE_OFFSET)
        
        # ---------------------------------------------------------
        # 2. NORMAL FORM: lx + my + nz = p
        # ---------------------------------------------------------
        self.next_section("Normal Form")

        t2_header = Text("2. Normal Form", font_size=24, color=YELLOW).to_edge(LEFT).shift(UP*1.5)
        t2_eq = MathTex("{{l}}x + {{m}}y + {{n}}z = {{p}}", font_size=32).next_to(t2_header, DOWN, aligned_edge=LEFT)
        t2_eq.set_color_by_tex_to_color_map({"l": RED, "m": GREEN, "n": BLUE, "p": PURPLE})
        
        t2_list = VGroup(
            Tex(r"$\bullet$ $p$: perpendicular distance", font_size=22),
            Tex(r"$\bullet$ $l,m,n$: direction cosines", font_size=22),
        ).arrange(DOWN, aligned_edge=LEFT).next_to(t2_eq, DOWN, buff=0.3, aligned_edge=LEFT)

        self.add_fixed_in_frame_mobjects(t2_header, t2_eq, t2_list)
        self.play(Write(t2_header), Write(t2_eq), FadeIn(t2_list))

        # 3D Animation
        u_vec = np.array([1, 2, 2]) / 3.0
        p_dist = 4.5
        LABEL_SIZE = 28
        
        # --- MODIFICATION: Scale up the vector for better visibility ---
        scale_factor = 2.5 
        scaled_u = u_vec * scale_factor

        # 1. Show p (Distance)
        self.set_camera_orientation(phi=70*DEGREES, theta=30*DEGREES)
        p_end = axes.c2p(*(u_vec * p_dist))
        line_p = Line3D(axes.c2p(0,0,0), p_end, color=PURPLE, thickness=0.07)
        lbl_p = Text("p", color=PURPLE, font_size=36).move_to(line_p.get_center() + UP*0.6)
        self.play(Create(line_p), Write(lbl_p))
        
        # 2. Show Projections (l, m, n) - Using SCALED vector
        scaled_tip = axes.c2p(*scaled_u)
        arrow_u = Arrow3D(axes.c2p(0,0,0), scaled_tip, color=YELLOW, thickness=0.05)
        self.play(Create(arrow_u))

        # Show 'l'
        self.move_camera(phi=0*DEGREES, theta=-90*DEGREES, run_time=1.5)
        proj_x = axes.c2p(scaled_u[0], 0, 0) # Use scaled component
        dash_x = DashedLine(scaled_tip, proj_x, color=RED_E, stroke_width=4, dash_length=0.2)
        line_l = Line(axes.c2p(0,0,0), proj_x, color=RED, stroke_width=6)
        lbl_l = Text("l", color=RED, font_size=LABEL_SIZE).next_to(line_l, DOWN)
        self.play(Create(dash_x), Create(line_l), Write(lbl_l))
        
        # Show 'm'
        self.move_camera(phi=0*DEGREES, theta=0*DEGREES, run_time=1.5)
        proj_y = axes.c2p(0, scaled_u[1], 0) # Use scaled component
        dash_y = DashedLine(scaled_tip, proj_y, color=GREEN_E, stroke_width=4, dash_length=0.2)
        line_m = Line(axes.c2p(0,0,0), proj_y, color=GREEN, stroke_width=6)
        lbl_m = Text("m", color=GREEN, font_size=LABEL_SIZE).next_to(line_m, LEFT)
        self.play(Create(dash_y), Create(line_m), Write(lbl_m))

        # Show 'n'
        self.move_camera(phi=90*DEGREES, theta=0*DEGREES, run_time=1.5)
        proj_z = axes.c2p(0, 0, scaled_u[2]) # Use scaled component
        dash_z = DashedLine(scaled_tip, proj_z, color=BLUE_E, stroke_width=4, dash_length=0.2)
        line_n = Line(axes.c2p(0,0,0), proj_z, color=BLUE, stroke_width=6)
        
        lbl_n = Text("n", color=BLUE, font_size=LABEL_SIZE)
        lbl_n.rotate(90*DEGREES, axis=RIGHT)
        lbl_n.rotate(90*DEGREES, axis=OUT)
        lbl_n.next_to(line_n, UP) 

        self.play(Create(dash_z), Create(line_n), Write(lbl_n))
        
        # Reset View & Plane
        self.move_camera(phi=70*DEGREES, theta=30*DEGREES, run_time=1.5)
        self.play(
            FadeOut(dash_x), FadeOut(line_l), FadeOut(lbl_l),
            FadeOut(dash_y), FadeOut(line_m), FadeOut(lbl_m),
            FadeOut(dash_z), FadeOut(line_n), FadeOut(lbl_n),
            FadeOut(arrow_u)
        )
        
        l, m, n = u_vec
        plane_2 = Surface(
            lambda u, v: axes.c2p(u, v, (p_dist - l*u - m*v)/n),
            u_range=[0, 3.5], v_range=[0, 3.5],
            fill_opacity=0.5, checkerboard_colors=[PURPLE_D, PURPLE_E], resolution=(12,12)
        )
        self.play(Create(plane_2))
        self.wait(1)

        # Cleanup
        self.play(
            FadeOut(t2_header), FadeOut(t2_eq), FadeOut(t2_list), 
            FadeOut(plane_2), FadeOut(line_p), FadeOut(lbl_p)
        )

        # =========================================================
        # >>> NORMAL FORM EXAMPLES
        # =========================================================
        self.run_normal_examples(axes)

        # ---------------------------------------------------------
        # 3. INTERCEPT FORM: x/a + y/b + z/c = 1
        # ---------------------------------------------------------
        self.next_section("Intercept Form")

        t3_header = Text("3. Intercept Form", font_size=24, color=YELLOW).to_edge(LEFT).shift(UP*1.5)
        t3_eq = MathTex(r"\frac{x}{a} + \frac{y}{b} + \frac{z}{c} = 1", font_size=32).next_to(t3_header, DOWN, aligned_edge=LEFT)
        t3_eq.set_color_by_tex_to_color_map({"a": RED, "b": GREEN, "c": BLUE})
        
        t3_list = VGroup(
            Tex(r"$\bullet$ Intercepts on axes", font_size=22),
        ).arrange(DOWN, aligned_edge=LEFT).next_to(t3_eq, DOWN, buff=0.3, aligned_edge=LEFT)
        
        self.add_fixed_in_frame_mobjects(t3_header, t3_eq, t3_list)
        self.play(Write(t3_header), Write(t3_eq), FadeIn(t3_list))

        val_a, val_b, val_c = 3.8, 3.2, 2.8
        LABEL_SIZE = 32
        DOT_RADIUS = 0.15

        # Show 'a'
        self.move_camera(phi=80*DEGREES, theta=-90*DEGREES, run_time=1)
        dot_a = Dot3D(axes.c2p(val_a,0,0), color=RED, radius=DOT_RADIUS)
        lbl_a = Text("a", color=RED, font_size=LABEL_SIZE).next_to(dot_a, UP).rotate(90*DEGREES, RIGHT)
        self.play(GrowFromCenter(dot_a), Write(lbl_a))

        # Show 'b'
        self.move_camera(phi=80*DEGREES, theta=0*DEGREES, run_time=1)
        dot_b = Dot3D(axes.c2p(0,val_b,0), color=GREEN, radius=DOT_RADIUS)
        lbl_b = Text("b", color=GREEN, font_size=LABEL_SIZE).next_to(dot_b, RIGHT)
        self.play(GrowFromCenter(dot_b), Write(lbl_b))

        # Show 'c'
        self.move_camera(phi=85*DEGREES, theta=45*DEGREES, run_time=1)
        dot_c = Dot3D(axes.c2p(0,0,val_c), color=BLUE, radius=DOT_RADIUS)
        lbl_c = Text("c", color=BLUE, font_size=LABEL_SIZE).next_to(dot_c, RIGHT).rotate(90*DEGREES, UP)
        self.play(GrowFromCenter(dot_c), Write(lbl_c))
        
        # Plane
        self.move_camera(phi=70*DEGREES, theta=30*DEGREES, run_time=1.5)
        tri_plane = Polygon(
            axes.c2p(val_a,0,0), axes.c2p(0,val_b,0), axes.c2p(0,0,val_c),
            fill_color=TEAL, fill_opacity=0.5, stroke_color=WHITE, stroke_width=2
        )
        self.play(Create(tri_plane))
        self.wait(1)
        
        # Cleanup
        self.play(
            FadeOut(t3_header), FadeOut(t3_eq), FadeOut(t3_list),
            FadeOut(tri_plane), FadeOut(dot_a), FadeOut(lbl_a),
            FadeOut(dot_b), FadeOut(lbl_b), FadeOut(dot_c), FadeOut(lbl_c)
        )

        # =========================================================
        # >>> INTERCEPT FORM EXAMPLES
        # =========================================================
        self.run_intercept_examples(axes)

        # ---------------------------------------------------------
        # 4. POINT-NORMAL FORM: a(x-x1)...
        # ---------------------------------------------------------
        self.next_section("Point-Normal Form")

        t4_header = Text("4. Point-Normal Form", font_size=24, color=YELLOW).to_edge(LEFT).shift(UP*1.5)
        t4_eq = MathTex(r"a(x-x_1) + b(y-y_1) + c(z-z_1) = 0", font_size=28).next_to(t4_header, DOWN, aligned_edge=LEFT)
        t4_eq.set_color_by_tex_to_color_map({"a": RED, "b": GREEN, "c": BLUE, "x_1": ORANGE, "y_1": ORANGE, "z_1": ORANGE})
        
        t4_list = VGroup(
            Tex(r"$\bullet$ Fixed Point $P(x_1, y_1, z_1)$", font_size=22),
            Tex(r"$\bullet$ Normal $\vec{n}(a,b,c)$ at P", font_size=22),
        ).arrange(DOWN, aligned_edge=LEFT).next_to(t4_eq, DOWN, buff=0.3, aligned_edge=LEFT)
        
        self.add_fixed_in_frame_mobjects(t4_header, t4_eq, t4_list)
        self.play(Write(t4_header), Write(t4_eq), FadeIn(t4_list))

        px, py, pz = 2.2, 2.8, 2.2
        LABEL_SIZE = 24
        TRACE_WIDTH = 4
        TRACE_OPACITY = 0.7
        
        # 1. Walk to P
        self.move_camera(phi=0*DEGREES, theta=-90*DEGREES, run_time=1)
        trace_x = DashedLine(axes.c2p(0,0,0), axes.c2p(px,0,0), color=ORANGE, stroke_width=TRACE_WIDTH).set_opacity(TRACE_OPACITY)
        lbl_x1 = Text("x1", color=ORANGE, font_size=LABEL_SIZE).next_to(trace_x, DOWN)
        self.play(Create(trace_x), Write(lbl_x1))
        
        trace_y = DashedLine(axes.c2p(px,0,0), axes.c2p(px,py,0), color=ORANGE, stroke_width=TRACE_WIDTH).set_opacity(TRACE_OPACITY)
        lbl_y1 = Text("y1", color=ORANGE, font_size=LABEL_SIZE).next_to(trace_y, RIGHT)
        self.play(Create(trace_y), Write(lbl_y1))
        
        self.move_camera(phi=90*DEGREES, theta=0*DEGREES, run_time=1)
        trace_z = DashedLine(axes.c2p(px,py,0), axes.c2p(px,py,pz), color=ORANGE, stroke_width=TRACE_WIDTH).set_opacity(TRACE_OPACITY)
        lbl_z1 = Text("z1", color=ORANGE, font_size=LABEL_SIZE).next_to(trace_z, RIGHT)
        p_dot = Dot3D(axes.c2p(px,py,pz), color=ORANGE, radius=0.15)
        lbl_p = Text("P", color=ORANGE, font_size=36).next_to(p_dot, UP+RIGHT)
        self.play(Create(trace_z), Write(lbl_z1), GrowFromCenter(p_dot), Write(lbl_p))
        
        # 2. Show Normal at P
        self.move_camera(phi=70*DEGREES, theta=45*DEGREES, run_time=1.5)
        self.play(FadeOut(trace_x), FadeOut(lbl_x1), FadeOut(trace_y), FadeOut(lbl_y1), FadeOut(trace_z), FadeOut(lbl_z1))
        
        norm_dir = np.array([1, 0.5, 1])
        n_end = axes.c2p(*(np.array([px,py,pz]) + norm_dir))
        arrow_np = Arrow3D(axes.c2p(px,py,pz), n_end, color=WHITE, thickness=0.07)
        lbl_np = MathTex(r"\vec{n}", color=WHITE, font_size=36).next_to(arrow_np, UP)
        self.play(Create(arrow_np), Write(lbl_np))
        
        # Plane
        plane_4 = Surface(
            lambda u, v: axes.c2p(u, v, pz - (u-px) - 0.5*(v-py)),
            u_range=[px-1.5, px+1.5], v_range=[py-1.5, py+1.5],
            fill_opacity=0.5, checkerboard_colors=[GREY, GREY_E], resolution=(12,12)
        )
        self.play(Create(plane_4))
        self.wait(2)
        
        # Cleanup
        self.play(FadeOut(Group(*self.mobjects)))
        
        # =========================================================
        # >>> POINT-NORMAL FORM EXAMPLES
        # =========================================================
        # (Since this is the end of the scene, we don't need a cleanup phase afterwards)
        # Note: I am calling this on a fresh screen, so I will rebuild axes inside the helper.
        self.run_point_normal_examples()


    # =========================================================================
    #   HELPER METHODS FOR EXAMPLES
    # =========================================================================

    def create_plane_surface(self, axes, a, b, c, d_const, u_range, v_range, color=BLUE_E):
        """Robust surface creator handling Vertical (c=0) and Horizontal cases."""
        # Equation: ax + by + cz = d_const (where d_const is -d in general form or p in normal)
        
        if abs(c) > 0.1:
            # Standard z = f(x,y)
            return Surface(
                lambda u, v: axes.c2p(u, v, (d_const - a*u - b*v)/c),
                u_range=u_range, v_range=v_range,
                fill_opacity=0.5, checkerboard_colors=[color, color], resolution=(10,10)
            )
        elif abs(b) > 0.1:
            # Vertical Plane y = f(x,z) -> parameterize with u=x, v=z
            return Surface(
                lambda u, v: axes.c2p(u, (d_const - a*u)/b, v),
                u_range=u_range, v_range=v_range,
                fill_opacity=0.5, checkerboard_colors=[color, color], resolution=(10,10)
            )
        else:
            # Vertical Plane x = const -> parameterize with u=y, v=z
            return Surface(
                lambda u, v: axes.c2p(d_const/a, u, v),
                u_range=u_range, v_range=v_range,
                fill_opacity=0.5, checkerboard_colors=[color, color], resolution=(10,10)
            )

    def run_general_examples(self, axes, offset):
        examples = [
            {"a": 1, "b": 1, "c": 1, "d": -3, "eq": "x + y + z - 3 = 0", "desc": "Standard Plane"},
            {"a": 1, "b": 0, "c": 1, "d": -2, "eq": "x + z - 2 = 0", "desc": "Vertical (Missing y)"},
            {"a": 0, "b": 0, "c": 1, "d": -2, "eq": "z - 2 = 0", "desc": "Horizontal Plane"},
            {"a": 1, "b": -2, "c": 0, "d": 0, "eq": "x - 2y = 0", "desc": "Through Z-axis (c=d=0)"},
        ]

        header = Text("General Form Examples", font_size=24, color=YELLOW).to_edge(LEFT).shift(UP*1.5)
        self.add_fixed_in_frame_mobjects(header)
        self.play(Write(header))

        for i, ex in enumerate(examples):
            # 1. Update Text
            eq_text = MathTex(ex["eq"], font_size=28).next_to(header, DOWN, aligned_edge=LEFT)
            desc_text = Text(ex["desc"], font_size=20, color=GRAY).next_to(eq_text, DOWN, aligned_edge=LEFT)
            
            n_vals = Tex(f"Normal: ({ex['a']}, {ex['b']}, {ex['c']})", font_size=24, color=WHITE).next_to(desc_text, DOWN, aligned_edge=LEFT)
            
            self.add_fixed_in_frame_mobjects(eq_text, desc_text, n_vals)
            self.play(FadeIn(eq_text), FadeIn(desc_text), FadeIn(n_vals))

            # 2. 3D Elements
            # Draw Normal Vector
            n_vec = np.array([ex['a'], ex['b'], ex['c']])
            n_len = np.linalg.norm(n_vec)
            if n_len > 0:
                n_start = axes.c2p(0,0,0)
                n_end = axes.c2p(ex['a'], ex['b'], ex['c'])
                arrow = Arrow3D(n_start, n_end, color=RED)
                lbl = MathTex(r"\vec{n}", color=RED).move_to(axes.c2p(ex['a']*1.1, ex['b']*1.1, ex['c']*1.1))
                self.play(Create(arrow), Write(lbl))
            
            # Draw Plane
            # We solve ax+by+cz = -d
            surf = self.create_plane_surface(
                axes, ex['a'], ex['b'], ex['c'], -ex['d'], 
                u_range=[-3,3], v_range=[-3,3], color=BLUE_C
            )
            
            # Camera adjustment
            if ex['c'] == 0: # Vertical
                self.move_camera(phi=60*DEGREES, theta=10*DEGREES, run_time=1)
            elif ex['a'] == 0 and ex['b'] == 0: # Horizontal
                self.move_camera(phi=80*DEGREES, theta=45*DEGREES, run_time=1)
            else:
                self.move_camera(phi=70*DEGREES, theta=45*DEGREES, run_time=1)

            self.play(Create(surf))
            self.wait(1.5)

            # Cleanup
            self.play(FadeOut(eq_text), FadeOut(desc_text), FadeOut(n_vals), FadeOut(surf), FadeOut(arrow), FadeOut(lbl))
        
        self.play(FadeOut(header))

    def run_normal_examples(self, axes):
        # l^2 + m^2 + n^2 must be 1. p is dist.
        sq3 = 1/np.sqrt(3)
        sq2 = 1/np.sqrt(2)
        
        examples = [
            {"l": sq3, "m": sq3, "n": sq3, "p": 2, "eq": r"\frac{1}{\sqrt{3}}x + \frac{1}{\sqrt{3}}y + \frac{1}{\sqrt{3}}z = 2", "desc": "Balanced Plane"},
            {"l": 0, "m": 0, "n": 1, "p": 3, "eq": "0x + 0y + 1z = 3", "desc": "Flat at z=3"},
            {"l": 1, "m": 0, "n": 0, "p": 2, "eq": "1x + 0y + 0z = 2", "desc": "Vertical at x=2"},
            {"l": sq2, "m": sq2, "n": 0, "p": 2, "eq": r"\frac{1}{\sqrt{2}}x + \frac{1}{\sqrt{2}}y = 2", "desc": "Angled Vertical"},
        ]

        header = Text("Normal Form Examples", font_size=24, color=YELLOW).to_edge(LEFT).shift(UP*1.5)
        self.add_fixed_in_frame_mobjects(header)
        self.play(Write(header))

        for ex in examples:
            eq_text = MathTex(ex["eq"], font_size=24).next_to(header, DOWN, aligned_edge=LEFT)
            desc_text = Text(ex["desc"], font_size=20, color=GRAY).next_to(eq_text, DOWN, aligned_edge=LEFT)
            p_text = Tex(f"Dist p = {ex['p']}", font_size=24, color=PURPLE).next_to(desc_text, DOWN, aligned_edge=LEFT)
            
            self.add_fixed_in_frame_mobjects(eq_text, desc_text, p_text)
            self.play(FadeIn(eq_text), FadeIn(desc_text), FadeIn(p_text))

            # Draw Distance Line (Normal vector scaled to p)
            n_vec = np.array([ex['l'], ex['m'], ex['n']])
            p_point = n_vec * ex['p']
            
            line_p = Line3D(axes.c2p(0,0,0), axes.c2p(*p_point), color=PURPLE, thickness=0.05)
            lbl_p = MathTex("p", color=PURPLE).move_to(axes.c2p(*(p_point/2)) + UP*0.2)
            
            self.play(Create(line_p), Write(lbl_p))

            # Draw Plane
            surf = self.create_plane_surface(
                axes, ex['l'], ex['m'], ex['n'], ex['p'],
                u_range=[-3,3], v_range=[-3,3], color=TEAL_E
            )
            
            # Camera
            if ex['n'] == 1: # Horizontal
                self.move_camera(phi=85*DEGREES, theta=10*DEGREES)
            elif ex['n'] == 0: # Vertical
                self.move_camera(phi=60*DEGREES, theta=45*DEGREES)
            else:
                self.move_camera(phi=70*DEGREES, theta=30*DEGREES)

            self.play(Create(surf))
            self.wait(1.5)
            self.play(FadeOut(eq_text), FadeOut(desc_text), FadeOut(p_text), FadeOut(surf), FadeOut(line_p), FadeOut(lbl_p))

        self.play(FadeOut(header))

    def run_intercept_examples(self, axes):
        examples = [
            {"a": 2, "b": 2, "c": 2, "eq": r"\frac{x}{2} + \frac{y}{2} + \frac{z}{2} = 1"},
            {"a": 4, "b": -2, "c": 3, "eq": r"\frac{x}{4} + \frac{y}{-2} + \frac{z}{3} = 1"},
            {"a": 1, "b": 4, "c": 1, "eq": r"\frac{x}{1} + \frac{y}{4} + \frac{z}{1} = 1"},
            {"a": -2, "b": -2, "c": 2, "eq": r"\frac{x}{-2} + \frac{y}{-2} + \frac{z}{2} = 1"},
        ]

        header = Text("Intercept Form Examples", font_size=24, color=YELLOW).to_edge(LEFT).shift(UP*1.5)
        self.add_fixed_in_frame_mobjects(header)
        self.play(Write(header))

        for ex in examples:
            eq_text = MathTex(ex["eq"], font_size=26).next_to(header, DOWN, aligned_edge=LEFT)
            vals = Tex(f"a={ex['a']}, b={ex['b']}, c={ex['c']}", font_size=24, color=GRAY).next_to(eq_text, DOWN, aligned_edge=LEFT)
            self.add_fixed_in_frame_mobjects(eq_text, vals)
            self.play(FadeIn(eq_text), FadeIn(vals))

            # Dots
            da = Dot3D(axes.c2p(ex['a'],0,0), color=RED)
            db = Dot3D(axes.c2p(0,ex['b'],0), color=GREEN)
            dc = Dot3D(axes.c2p(0,0,ex['c']), color=BLUE)
            
            self.play(GrowFromCenter(da), GrowFromCenter(db), GrowFromCenter(dc))

            # Polygon Plane
            poly = Polygon(
                axes.c2p(ex['a'],0,0), axes.c2p(0,ex['b'],0), axes.c2p(0,0,ex['c']),
                fill_color=BLUE_A, fill_opacity=0.4, stroke_color=WHITE
            )
            
            # Camera adjusting to octant
            if ex['a'] < 0 or ex['b'] < 0:
                self.move_camera(phi=70*DEGREES, theta=120*DEGREES, run_time=1)
            else:
                self.move_camera(phi=70*DEGREES, theta=30*DEGREES, run_time=1)

            self.play(Create(poly))
            self.wait(1.5)
            self.play(FadeOut(eq_text), FadeOut(vals), FadeOut(poly), FadeOut(da), FadeOut(db), FadeOut(dc))

        self.play(FadeOut(header))

    def run_point_normal_examples(self):
        # Re-setup scene specifically for this last batch since previous one cleared everything
        SCENE_OFFSET = RIGHT * 3.5
        axes = ThreeDAxes(
            x_range=[-4, 4, 1], y_range=[-4, 4, 1], z_range=[-4, 4, 1],
            x_length=5.5, y_length=5.5, z_length=5.5
        ).shift(SCENE_OFFSET)
        self.add(axes)
        self.set_camera_orientation(phi=70 * DEGREES, theta=30 * DEGREES)

        header = Text("Point-Normal Examples", font_size=24, color=YELLOW).to_edge(LEFT).shift(UP*1.5)
        self.add_fixed_in_frame_mobjects(header)
        self.play(Write(header))

        examples = [
            {"p": [1,1,1], "n": [1,1,1], "eq": "1(x-1)+1(y-1)+1(z-1)=0", "desc": "Centered at (1,1,1)"},
            {"p": [0,0,0], "n": [0,0,1], "eq": "0(x-0)+0(y-0)+1(z-0)=0", "desc": "XY Plane via Origin"},
            {"p": [2,0,0], "n": [1,0,0], "eq": "1(x-2)+0(y-0)+0(z-0)=0", "desc": "YZ Plane shifted to x=2"},
            {"p": [0,0,2], "n": [1,2,1], "eq": "1(x)+2(y)+1(z-2)=0", "desc": "Tilted via (0,0,2)"}
        ]

        for ex in examples:
            eq_text = MathTex(ex["eq"], font_size=24).next_to(header, DOWN, aligned_edge=LEFT)
            desc = Text(ex["desc"], font_size=20, color=ORANGE).next_to(eq_text, DOWN, aligned_edge=LEFT)
            self.add_fixed_in_frame_mobjects(eq_text, desc)
            self.play(FadeIn(eq_text), FadeIn(desc))

            # Point
            px, py, pz = ex['p']
            p_dot = Dot3D(axes.c2p(px, py, pz), color=ORANGE, radius=0.1)
            p_lbl = Text("P", color=ORANGE, font_size=24).next_to(p_dot, UP)
            
            # Normal
            nx, ny, nz = ex['n']
            n_end = np.array([px+nx, py+ny, pz+nz])
            arrow = Arrow3D(axes.c2p(px, py, pz), axes.c2p(*n_end), color=WHITE)
            
            self.play(GrowFromCenter(p_dot), Write(p_lbl), Create(arrow))

            # Surface
            # a(x-x0) + b(y-y0) + c(z-z0) = 0
            # ax + by + cz = ax0 + by0 + cz0
            rhs = nx*px + ny*py + nz*pz
            surf = self.create_plane_surface(
                axes, nx, ny, nz, rhs, 
                u_range=[px-2, px+2], v_range=[py-2, py+2], color=GREY_A
            )

            # Camera
            if nz == 1 and nx == 0: # Horizontal
                self.move_camera(phi=80*DEGREES, theta=30*DEGREES)
            elif nx == 1 and ny == 0 and nz == 0: # Vertical
                self.move_camera(phi=60*DEGREES, theta=-60*DEGREES)
            else:
                self.move_camera(phi=70*DEGREES, theta=45*DEGREES)

            self.play(Create(surf))
            self.wait(1.5)
            self.play(FadeOut(eq_text), FadeOut(desc), FadeOut(p_dot), FadeOut(p_lbl), FadeOut(arrow), FadeOut(surf))

%manim -qk -v warning PlaneEquations
```
