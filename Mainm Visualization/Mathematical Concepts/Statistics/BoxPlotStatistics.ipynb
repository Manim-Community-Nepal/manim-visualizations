# Statistical Measures & Box Plots: Geometric Interpretation

This document provides a geometric understanding of statistical measures using Box Plots, moving beyond simple formulas to visualize data density and spread.

## 1. The Geometry of the Box Plot
A box plot is a visual map of **data density** and **dispersion**.



[Image of box plot geometry]


* **The Box (The "Body"):** Represents the **Interquartile Range (IQR)**. It contains the middle **50%** of the data.
    * *Wide Box:* The "typical" crowd is spread out.
    * *Narrow Box:* High agreement/concentration among the middle group.
* **The Median Line (The "Cut"):** The solid line splitting the box. It divides the population exactly in half (**50% area left**, **50% area right**).
* **The Whiskers (The "Reach"):** Extend to the Minimum and Maximum values, showing the full **Range** of the data including extremes.

---

## 2. Measures of Central Tendency (Location)
These measures define the "center" of the dataset but behave differently geometrically.

* **Mean ($\bar{x}$):** The arithmetic balance point.
    * *Behavior:* It is **magnetic** to outliers.
* **Median ($Q_2$ / $P_{50}$):** The positional center.
    * *Behavior:* It is **robust**. An extreme outlier moves the Mean but often leaves the Median unchanged.
* **Mode:** The value with the highest frequency (the peak).

### Geometric Relationship (Skewness)
* **Symmetric:** $\text{Mean} \approx \text{Median}$
* **Right Skewed (Positive):** $\text{Mean} > \text{Median}$ (The tail pulls the mean to the right).
* **Left Skewed (Negative):** $\text{Mean} < \text{Median}$ (The tail pulls the mean to the left).

---

## 3. Measures of Dispersion (Spread)
These measures quantify how "scattered" the data is.

| Measure | Formula | Geometric Meaning | Robustness |
| :--- | :--- | :--- | :--- |
| **Range** | $Max - Min$ | The total length from whisker tip to whisker tip. | **Low** (Affected by extremes). |
| **IQR** | $Q_3 - Q_1$ | The width of the Box. | **High** (Describes the middle 50%). |
| **Quartile Deviation (Q.D.)** | $\frac{Q_3 - Q_1}{2}$ | **Half the width of the Box.** | **High** (Average distance from center for the middle group). |

---

## 4. Quartiles & Percentiles
Quartiles are specific checkpoints (**Percentiles**) that divide the sorted data into four equal groups.

* **$Q_1$ (First Quartile):** The $25^{th}$ Percentile ($P_{25}$).
* **$Q_2$ (Median):** The $50^{th}$ Percentile ($P_{50}$).
* **$Q_3$ (Third Quartile):** The $75^{th}$ Percentile ($P_{75}$).

$$
\text{Q.D.} = \frac{P_{75} - P_{25}}{2}
$$

---

## 5. Interpreting Shapes (Dataset Signatures)

* **Normal-ish:** Box is centered; whiskers are roughly equal length.
* **Skewed:** One whisker is much longer than the other. The box is shifted opposite to the tail.
* **Tight Spread:** The entire plot (Range) is short; the box (IQR) is very narrow.
* **Wide Spread:** The plot is long; the box is wide.
* **Central Spike:** Data is highly concentrated at one value. The box might collapse into a single line if $Q_1 = \text{Median} = Q_3$.

```python
from manim import *
import numpy as np

class StatisticalMeasuresBoxPlot(Scene):
    def construct(self):
        # --- CONFIGURATION & COLORS ---
        TheColors = {
            "mean": YELLOW,
            "median": RED,
            "mode": ORANGE,
            "q1": TEAL_D,
            "q3": TEAL_B,
            "box_fill": TEAL,
            "whiskers": LIGHT_GREY,
            "range": GREY,
            "iqr": BLUE,
            "qd": PURPLE,
            "data_points": WHITE,
        }

        # --- DATASETS ---
        datasets = [
            # 1. Normal-ish
            sorted([10, 11, 12, 13, 13, 14, 14, 14, 15, 15, 15, 15, 16, 16, 16, 17, 17, 18, 19, 20]),
            # 2. Right Skewed
            sorted([10, 10, 10, 11, 11, 11, 11, 12, 12, 12, 13, 13, 14, 14, 15, 16, 17, 18, 19, 20]),
            # 3. Left Skewed
            sorted([10, 11, 12, 13, 14, 15, 16, 16, 17, 17, 18, 18, 18, 19, 19, 19, 19, 20, 20, 20]),
            # 4. Wide Spread
            sorted([10, 10, 10, 11, 11, 12, 13, 14, 14, 15, 15, 16, 17, 18, 19, 19, 20, 20, 20]),
            # 5. Tight Spread
            sorted([13, 13, 14, 14, 14, 14, 15, 15, 15, 15, 15, 15, 16, 16, 16, 16, 16, 17, 17, 17]),
            # 6. Bimodal
            sorted([10, 10, 10, 11, 11, 11, 12, 13, 14, 15, 16, 17, 18, 19, 19, 19, 20, 20, 20, 20]),
            # 7. Uniform
            sorted([10, 10, 11, 11, 12, 12, 13, 13, 14, 14, 15, 15, 16, 16, 17, 17, 18, 18, 19, 19]),
            # 8. Central Spike
            sorted([10, 11, 15, 15, 15, 15, 15, 15, 15, 15, 15, 15, 15, 15, 15, 15, 16, 17, 18, 20]),
        ]

        # Descriptions corresponding to the datasets above
        descriptions = [
            "Normal-ish (Symmetric)",
            "Right Skewed (Positive)",
            "Left Skewed (Negative)",
            "Wide Spread (High Dispersion)",
            "Tight Spread (Low Dispersion)",
            "Bimodal (Two Peaks)",
            "Uniform (Flat)",
            "Central Spike w/ Outliers"
        ]

        # --- HELPER FUNCTIONS ---
        def calculate_stats(data):
            n = len(data)
            mean_val = np.mean(data)
            median_val = np.median(data)
            lower_half = data[:int(n/2)]
            upper_half = data[int(n/2):]
            q1_val = np.median(lower_half)
            q3_val = np.median(upper_half)
            counts = {}
            for x in data: counts[x] = counts.get(x, 0) + 1
            max_count = max(counts.values())
            modes = [k for k, v in counts.items() if v == max_count]
            
            return {
                "min": data[0], "max": data[-1],
                "q1": q1_val, "median": median_val, "q3": q3_val,
                "mean": mean_val, "mode": modes[0],
                "range": round(data[-1] - data[0], 2),
                "iqr": round(q3_val - q1_val, 2),
                "qd": round((q3_val - q1_val) / 2, 2)
            }

        def create_data_grid(data, center_point):
            # Creates a 2x10 grid of numbers
            group = VGroup()
            rows, cols = 2, 10
            for i in range(rows):
                row_group = VGroup()
                for j in range(cols):
                    idx = i * cols + j
                    if idx < len(data):
                        txt = Text(str(data[idx]), font_size=18, color=TheColors["data_points"])
                        row_group.add(txt)
                row_group.arrange(RIGHT, buff=0.35)
                group.add(row_group)
            group.arrange(DOWN, buff=0.2)
            group.move_to(center_point)
            label = Text("Dataset (20 Points)", font_size=16, color=GRAY).next_to(group, UP, buff=0.15)
            return VGroup(label, group)

        def create_box_plot(axes, stats, y_level, opacity=0.5):
            # Maps values to points relative to the axis + y_offset
            def p(val): return axes.n2p(val) + UP * y_level
            p_min, p_max = p(stats["min"]), p(stats["max"])
            p_q1, p_q3 = p(stats["q1"]), p(stats["q3"])
            p_med = p(stats["median"])
            box_h = 0.8 

            whisk_L = Line(p_min, p_q1, color=TheColors["whiskers"])
            whisk_R = Line(p_q3, p_max, color=TheColors["whiskers"])
            cap_L = Line(p_min + UP*box_h/4, p_min + DOWN*box_h/4, color=TheColors["whiskers"])
            cap_R = Line(p_max + UP*box_h/4, p_max + DOWN*box_h/4, color=TheColors["whiskers"])

            box = Rectangle(
                width=p_q3[0] - p_q1[0], height=box_h,
                stroke_color=TheColors["box_fill"], fill_color=[TheColors["q1"], TheColors["q3"]],
                fill_opacity=opacity
            ).move_to((p_q1 + p_q3)/2)
            
            med_line = Line(p_med + UP*box_h/2, p_med + DOWN*box_h/2, color=TheColors["median"], stroke_width=5)
            
            # Markers
            p_mean = p(stats["mean"])
            mean_mark = DashedLine(p_mean + UP*box_h/1.4, p_mean + DOWN*box_h/1.4, color=TheColors["mean"])
            
            p_mode = p(stats["mode"])
            mode_mark = Dot(p_mode + UP*box_h/1.1, color=TheColors["mode"], radius=0.08)

            return VGroup(whisk_L, whisk_R, cap_L, cap_R, box, med_line, mean_mark, mode_mark)

        def create_dimension_arrow(start_point, end_point, label_text, color, direction=UP, buff=0.1):
            # Custom Dotted Arrow
            line = DashedLine(start_point, end_point, color=color, dash_length=0.1)
            
            # Tips
            tip_scale = 0.15
            tip_start = ArrowTriangleFilledTip(start_angle=line.get_angle() + PI, color=color).scale(tip_scale).move_to(start_point)
            tip_end = ArrowTriangleFilledTip(start_angle=line.get_angle(), color=color).scale(tip_scale).move_to(end_point)
            
            # Label
            lbl = Text(str(label_text), font_size=16, color=color).next_to(line, direction, buff=buff)
            
            return VGroup(line, tip_start, tip_end, lbl)

        def set_explainer(header, body, color):
            # Standardized left-side text creator
            g = VGroup(
                Text(header, color=color, weight=BOLD, font_size=32),
                Text(body, color=WHITE, font_size=22, line_spacing=1.2).scale(0.9)
            ).arrange(DOWN, aligned_edge=LEFT).to_edge(LEFT, buff=1.0).shift(UP*1.0)
            return g

        # --- SCENE LAYOUT ---
        title = Title(r"Statistical Measures \& Box Plots").scale(0.8)
        self.add(title)

        # 1. ZONES
        RIGHT_ZONE_X = config.frame_width * 0.25 
        
        # Grid: Moved DOWN to 2.0
        GRID_POS = RIGHT * RIGHT_ZONE_X + UP * 2.0 
        
        # Axis: Moved DOWN to 2.3
        AXIS_POS = RIGHT * RIGHT_ZONE_X + DOWN * 2.3
        
        # Plot Y Offset: Distance from Axis center
        PLOT_Y_OFFSET = 1.2

        # Draw Axis
        axes = NumberLine(
            x_range=[9, 21, 1], length=6.5, include_numbers=True, font_size=18, color=GREY
        ).move_to(AXIS_POS)
        # Axis Label
        axis_lbl = Text("Value Scale", font_size=16, color=GREY).next_to(axes, DOWN, buff=0.8) 
        self.add(axes, axis_lbl)

        # --- PART 1: GEOMETRIC DEEP DIVE ---
        dummy_stats = {"min": 10, "q1": 12, "median": 15, "q3": 18, "max": 20, "mean": 15.5, "mode": 14}
        demo_plot = create_box_plot(axes, dummy_stats, PLOT_Y_OFFSET, opacity=0.5)

        # 1. Median
        txt_med = set_explainer("Median (The Middle)", "Divides the data in half.\n50% left, 50% right.", TheColors["median"])
        self.play(Write(txt_med))
        self.play(Create(demo_plot), run_time=1.5)
        self.play(
            demo_plot.animate.set_opacity(0.2),
            demo_plot[5].animate.set_opacity(1).set_color(TheColors["median"])
        )
        self.wait(1.5)

        # 2. Box / IQR
        txt_box = set_explainer("The Box (IQR)", "The middle 50% of data.\nFrom Q1 to Q3.", TheColors["box_fill"])
        self.play(Transform(txt_med, txt_box))
        self.play(
            demo_plot.animate.set_opacity(0.2),
            demo_plot[4].animate.set_opacity(0.8).set_stroke(opacity=1)
        )
        self.wait(1.5)

        # 3. Whiskers / Range
        txt_rng = set_explainer("Whiskers (Range)", "Extends to Min and Max.\nShows total spread.", TheColors["whiskers"])
        self.play(Transform(txt_med, txt_rng))
        self.play(
            demo_plot.animate.set_opacity(0.2),
            demo_plot[0:4].animate.set_opacity(1).set_color(WHITE)
        )
        self.wait(1.5)

        # 4. Dimension Concepts
        txt_dim = set_explainer("Dimensions", "IQR = Box Width\nQ.D. = Half Box Width", TheColors["qd"])
        self.play(Transform(txt_med, txt_dim))
        self.play(demo_plot.animate.set_opacity(1))
        
        # Conceptual Arrows
        p_q1 = axes.n2p(dummy_stats["q1"]) + UP * PLOT_Y_OFFSET
        p_q3 = axes.n2p(dummy_stats["q3"]) + UP * PLOT_Y_OFFSET
        p_mid = (p_q1 + p_q3)/2
        
        iqr_arrow = create_dimension_arrow(p_q1 + UP*0.5, p_q3 + UP*0.5, "IQR", TheColors["iqr"])
        qd_arrow = create_dimension_arrow(p_q1 + UP*1.0, p_mid + UP*1.0, "Q.D. (Half)", TheColors["qd"])

        self.play(Create(iqr_arrow))
        self.play(Create(qd_arrow))
        self.wait(3)

        # Cleanup Intro
        self.play(
            FadeOut(demo_plot), FadeOut(txt_med), FadeOut(iqr_arrow), FadeOut(qd_arrow)
        )

        # --- PART 2: REAL DATA ANIMATION LOOP ---
        current_grid = VGroup()
        current_plot_group = VGroup()
        current_info = VGroup()
        current_desc_label = VGroup()

        for i, ds in enumerate(datasets):
            stats = calculate_stats(ds)
            desc_text = descriptions[i]
            
            # 1. Grid
            new_grid = create_data_grid(ds, GRID_POS)
            
            # 2. Box Plot
            new_plot = create_box_plot(axes, stats, PLOT_Y_OFFSET)
            
            # 3. Dimension Arrows
            
            # Range Arrow: BELOW Axis
            r_start = axes.n2p(stats["min"])
            r_end = axes.n2p(stats["max"])
            arrow_rng = create_dimension_arrow(r_start + DOWN*0.6, r_end + DOWN*0.6, f"Range: {stats['range']}", TheColors["range"], direction=DOWN)

            # IQR Arrow: ABOVE Box
            q1_p = axes.n2p(stats["q1"]) + UP * PLOT_Y_OFFSET
            q3_p = axes.n2p(stats["q3"]) + UP * PLOT_Y_OFFSET
            arrow_iqr = create_dimension_arrow(q1_p + UP*0.5, q3_p + UP*0.5, f"IQR: {stats['iqr']}", TheColors["iqr"])

            # QD Arrow: ABOVE IQR
            med_p = axes.n2p(stats["median"]) + UP * PLOT_Y_OFFSET
            arrow_qd = create_dimension_arrow(q1_p + UP*1.0, med_p + UP*1.0, f"Q.D.: {stats['qd']}", TheColors["qd"])
            
            plot_combo = VGroup(new_plot, arrow_rng, arrow_iqr, arrow_qd)

            # 4. Text Info (Comprehensive List)
            t_header = Text(f"Dataset {i+1}", font_size=24, weight=BOLD)
            ul = Underline(t_header)
            
            stats_list = VGroup(
                Text(f"Mean: {stats['mean']:.2f}", color=TheColors["mean"], font_size=20),
                Text(f"Median: {stats['median']}", color=TheColors["median"], font_size=20),
                Text(f"Mode: {stats['mode']}", color=TheColors["mode"], font_size=20),
                Text(f"Q1: {stats['q1']} | Q3: {stats['q3']}", color=TheColors["q1"], font_size=20),
                Text(f"Range: {stats['range']}", color=TheColors["range"], font_size=20),
                Text(f"IQR: {stats['iqr']}", color=TheColors["iqr"], font_size=20),
                Text(f"Q.D.: {stats['qd']}", color=TheColors["qd"], font_size=20, weight=BOLD),
            ).arrange(DOWN, aligned_edge=LEFT, buff=0.15)
            
            new_info = VGroup(t_header, ul, stats_list).arrange(DOWN, aligned_edge=LEFT, buff=0.2).to_edge(LEFT, buff=0.5)

            # 5. Distribution Label (Middle Bottom) - FIXED ERROR HERE
            new_desc_label = Text(desc_text, font_size=24, color=YELLOW, weight=BOLD).to_edge(DOWN, buff=0.5)

            # 6. Animate
            if i == 0:
                self.play(FadeIn(new_grid))
                self.play(Write(new_info))
                self.play(Write(new_desc_label))
                self.play(TransformFromCopy(new_grid[1], new_plot), run_time=1.5)
                self.play(Create(arrow_rng), Create(arrow_iqr), Create(arrow_qd))
                
                current_grid = new_grid
                current_info = new_info
                current_plot_group = plot_combo
                current_desc_label = new_desc_label
            else:
                self.play(
                    Transform(current_grid, new_grid),
                    Transform(current_info, new_info),
                    Transform(current_desc_label, new_desc_label),
                    Transform(current_plot_group, plot_combo),
                    run_time=1.5
                )
            
            self.wait(2.5)



%manim -qk -v warning StatisticalMeasuresBoxPlot
```
