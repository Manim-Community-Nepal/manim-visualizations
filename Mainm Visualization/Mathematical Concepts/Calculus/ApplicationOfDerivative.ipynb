# Calculus: Applications of Derivatives (Deep Dive)

This notebook details the analytical steps to determine the behavior of a function curve. We use the specific cubic function from our visualization to demonstrate concepts like stationary points, concavity, and absolute extrema.

## 1. The Function and Derivatives

We analyze the function $f(x)$ by finding its first and second order derivatives using the Power Rule $\frac{d}{dx}(ax^n) = anx^{n-1}$.

**Original Function:**
$$f(x) = 0.5x^3 - 3x^2 + 4.5x + 1$$

**First Derivative (Slope):**
$$f'(x) = \frac{dy}{dx} = 1.5x^2 - 6x + 4.5$$

**Second Derivative (Concavity):**
$$f''(x) = \frac{d^2y}{dx^2} = 3x - 6$$

---

## 2. Finding Stationary Points

A **Stationary Point** occurs where the tangent line is horizontal, meaning the derivative (slope) is zero.

**Step:** Set $\frac{dy}{dx} = 0$ and solve for $x$.

$$1.5x^2 - 6x + 4.5 = 0$$

Divide the entire equation by $1.5$ to simplify:
$$x^2 - 4x + 3 = 0$$

Factor the quadratic:
$$(x - 1)(x - 3) = 0$$

**Result:**
We have two stationary points at **$x = 1$** and **$x = 3$**.

---

## 3. Monotonicity (Increasing & Decreasing)

We determine the intervals where the function is rising (increasing) or falling (decreasing) by testing the sign of the first derivative $\frac{dy}{dx}$ around our stationary points.

**Interval $x < 1$ (Test Point $x = 0$):**
$$f'(0) = 1.5(0)^2 - 6(0) + 4.5 = +4.5$$
Since $f'(0) > 0$, the function is **Increasing**.

**Interval $1 < x < 3$ (Test Point $x = 2$):**
$$f'(2) = 1.5(2)^2 - 6(2) + 4.5$$
$$f'(2) = 6 - 12 + 4.5 = -1.5$$
Since $f'(2) < 0$, the function is **Decreasing**.

**Interval $x > 3$ (Test Point $x = 4$):**
$$f'(4) = 1.5(4)^2 - 6(4) + 4.5$$
$$f'(4) = 24 - 24 + 4.5 = +4.5$$
Since $f'(4) > 0$, the function is **Increasing**.

---

## 4. Nature of Stationary Points (Maxima & Minima)

To classify the stationary points found in Step 2, we use the **Second Derivative Test**.

**At $x = 1$:**
Substitute into $f''(x) = 3x - 6$:
$$f''(1) = 3(1) - 6 = -3$$
Since $f''(1) < 0$, the graph is Concave Down (frowns).
$\therefore$ **$x = 1$ is a Local Maximum.**

**At $x = 3$:**
Substitute into $f''(x) = 3x - 6$:
$$f''(3) = 3(3) - 6 = +3$$
Since $f''(3) > 0$, the graph is Concave Up (smiles).
$\therefore$ **$x = 3$ is a Local Minimum.**

---

## 5. Concavity and Point of Inflection

An **Inflection Point** is where the concavity changes. This occurs where the second derivative is zero.

**Step:** Set $\frac{d^2y}{dx^2} = 0$.

$$3x - 6 = 0$$
$$3x = 6 \implies x = 2$$

**Verification of Sign Change:**
* **Left of 2 ($x=1$):** $f''(1) = -3$ (Concave Down)
* **Right of 2 ($x=3$):** $f''(3) = +3$ (Concave Up)

**Result:**
Since the concavity changes from negative to positive, **$x = 2$ is a Point of Inflection.**

---

## 6. Absolute Extrema (Global Max/Min)

To find the Absolute Maximum and Minimum on a closed interval, e.g., $x \in [0.5, 3.5]$, we must compare the $y$-values of the **Critical Points** against the **Endpoints**.

**Candidate Points:**
1.  **Start Endpoint:** $x = 0.5$
2.  **Stationary Point:** $x = 1$
3.  **Stationary Point:** $x = 3$
4.  **End Endpoint:** $x = 3.5$

**Evaluation:**
$$f(0.5) = 0.5(0.125) - 3(0.25) + 4.5(0.5) + 1 \approx 2.56$$
$$f(1) = 0.5 - 3 + 4.5 + 1 = 3.00$$
$$f(3) = 13.5 - 27 + 13.5 + 1 = 1.00$$
$$f(3.5) = 0.5(42.875) - 3(12.25) + 4.5(3.5) + 1 \approx 1.44$$

**Conclusion:**
* **Absolute Maximum:** $y = 3.00$ at $x = 1$.
* **Absolute Minimum:** $y = 1.00$ at $x = 3$.

---

## 7. The Saddle Point Case

A **Saddle Point** (or stationary point of inflection) occurs when the derivative is zero, but the point is neither a maximum nor a minimum.

**Example Function:** $g(x) = 0.5(x-2)^3 + 2$
**Derivative:** $g'(x) = 1.5(x-2)^2$

**Stationary Point:**
Setting $g'(x) = 0$ gives **$x = 2$**.

**Analysis:**
1.  **Second Derivative Test:**
    $$g''(x) = 3(x-2)$$
    $$g''(2) = 0$$
    The test is inconclusive.

2.  **First Derivative Sign Test:**
    * **Left of 2 ($x=1.9$):** $g'(1.9) = 1.5(-0.1)^2 = (+)$ Positive.
    * **Right of 2 ($x=2.1$):** $g'(2.1) = 1.5(0.1)^2 = (+)$ Positive.

**Result:**
Because the slope does *not* change sign (it rises, flattens, and rises again), **$x = 2$ is a Saddle Point**.
