```python
from manim import *

class TrigMnemonics(Scene):
    def construct(self):
        # ==========================================
        # 1. Title Setup
        # ==========================================
        title = Text(
            "Trig Derivatives: सजिलै सम्झने जादुयी तरिकाहरू (Mnemonics)", 
            font="Noto Sans Devanagari", 
            font_size=32, 
            color=YELLOW
        ).to_edge(UP)
        self.play(Write(title))
        self.wait(1)

        # ==========================================
        # 2. Rule 1: The 'C' Rule (Negative Sign)
        # ==========================================
        rule1_title = Text("नियम १: 'C' को अर्थ Negative (-)", font="Noto Sans Devanagari", font_size=28, color=RED).shift(UP * 1.5)
        self.play(FadeIn(rule1_title, shift=DOWN))
        self.wait(1)

        # टेक्स्टलाई अझ स्पष्ट र विस्तृत बनाइएको छ
        rule1_desc = Text(
            "यदि कुनै त्रिकोणमितीय फलन 'C' अक्षरबाट सुरु हुन्छ (जस्तै: cos, cot, csc) भने,\nतिनीहरूको Derivative को नतिजामा सधैँ Negative (-) चिन्ह लाग्छ!", 
            font="Noto Sans Devanagari", font_size=22, color=WHITE
        ).next_to(rule1_title, DOWN, buff=0.3)
        self.play(Write(rule1_desc))
        self.wait(2)

        # उदाहरण देखाउने
        c_examples = MathTex(
            r"\cos(x) \rightarrow -", r"\quad \cot(x) \rightarrow -", r"\quad \csc(x) \rightarrow -"
        , font_size=36).next_to(rule1_desc, DOWN, buff=0.5)
        
        c_examples[0][0].set_color(RED) # c in cos
        c_examples[1][0].set_color(RED) # c in cot
        c_examples[2][0].set_color(RED) # c in csc

        self.play(Write(c_examples))
        self.play(Indicate(c_examples[0][-1]), Indicate(c_examples[1][-1]), Indicate(c_examples[2][-1]), color=RED)
        self.wait(2)

        # स्क्रीन सफा गर्ने
        self.play(FadeOut(rule1_title), FadeOut(rule1_desc), FadeOut(c_examples))

        # ==========================================
        # 3. Rule 2: The Swap Rule (Sin & Cos)
        # ==========================================
        rule2_title = Text("नियम २: एक-अर्काको जोडी (Sin र Cos)", font="Noto Sans Devanagari", font_size=28, color=BLUE).shift(UP * 1.5)
        self.play(FadeIn(rule2_title, shift=DOWN))
        self.wait(1)

        eq_sin = MathTex(r"\frac{d}{dx} \sin(x) = \cos(x)", font_size=40).next_to(rule2_title, DOWN, buff=0.5)
        eq_cos = MathTex(r"\frac{d}{dx} \cos(x) = -\sin(x)", font_size=40).next_to(eq_sin, DOWN, buff=0.5)

        eq_sin[0][4:7].set_color(GREEN)
        eq_sin[0][11:14].set_color(BLUE)
        
        eq_cos[0][4:7].set_color(BLUE)
        eq_cos[0][11:15].set_color(RED) # Minus sign is red

        self.play(Write(eq_sin))
        self.wait(1)
        self.play(Write(eq_cos))
        self.wait(0.5)
        
        # यहाँ के भइरहेको छ भनेर अझ राम्रोसँग बुझाइएको छ
        rule2_desc = Text(
            "Sine र Cosine मिल्ने साथी हुन्, त्यसैले तिनीहरू एक-अर्कामा साटिन्छन् (sin को cos, cos को sin)।\n(तर नियम १ नबिर्सनुहोस्: Cos मा 'C' भएकोले यसको उत्तरमा माइनस (-) लाग्छ!)", 
            font="Noto Sans Devanagari", font_size=20, color=LIGHT_GREY
        ).next_to(eq_cos, DOWN, buff=0.5)
        self.play(Write(rule2_desc))
        self.wait(2.5)

        self.play(FadeOut(rule2_title), FadeOut(eq_sin), FadeOut(eq_cos), FadeOut(rule2_desc))

        # ==========================================
        # 4. Rule 3: The Magic Words (Sec-Sec-Tan & Csc-Csc-Cot)
        # ==========================================
        rule3_title = Text("नियम ३: ३-शब्दको जादु (Sec-Sec-Tan)", font="Noto Sans Devanagari", font_size=28, color=ORANGE).shift(UP * 2)
        self.play(FadeIn(rule3_title, shift=DOWN))
        self.wait(1)

        # 'Cover-up trick' कसरी प्रयोग गर्ने भनेर स्पष्ट पारिएको
        rule3_desc = Text(
            "पहिले 'sec - sec - tan' सम्झनुहोस्। जुन फलनको Derivative चाहिएको छ, त्यसलाई बट्टाले छोप्नुहोस्।\nनछोपिएका बाँकी दुईवटा पदहरूको गुणनफल नै तपाईंको सही उत्तर हो!", 
            font="Noto Sans Devanagari", font_size=20, color=WHITE
        ).next_to(rule3_title, DOWN, buff=0.3)
        self.play(Write(rule3_desc))
        self.wait(2)

        # --- Sec / Tan Mnemonic ---
        magic_words1 = MathTex(r"\sec(x)", r"\quad \sec(x)", r"\quad \tan(x)", font_size=42).next_to(rule3_desc, DOWN, buff=0.5)
        magic_words1[0].set_color(YELLOW)
        magic_words1[1].set_color(YELLOW)
        magic_words1[2].set_color(ORANGE)
        self.play(FadeIn(magic_words1, scale=1.2))
        self.wait(1.5)

        # Tan लाई छोप्दा -> Sec * Sec
        tan_box = SurroundingRectangle(magic_words1[2], color=RED, fill_color=BLACK, fill_opacity=0.8)
        self.play(Create(tan_box))
        eq_tan = MathTex(r"\frac{d}{dx}\tan(x) = \sec^2(x)", font_size=36).next_to(magic_words1, DOWN, buff=0.5)
        self.play(Write(eq_tan))
        self.wait(2)
        self.play(FadeOut(tan_box))

        # Sec लाई छोप्दा -> Sec * Tan
        sec_box = SurroundingRectangle(magic_words1[0], color=RED, fill_color=BLACK, fill_opacity=0.8)
        self.play(Create(sec_box))
        eq_sec = MathTex(r"\frac{d}{dx}\sec(x) = \sec(x)\cdot\tan(x)", font_size=36).next_to(eq_tan, DOWN, buff=0.3)
        self.play(Write(eq_sec))
        self.wait(2)
        self.play(FadeOut(sec_box), FadeOut(magic_words1), FadeOut(eq_tan), FadeOut(eq_sec), FadeOut(rule3_title), FadeOut(rule3_desc))

        # --- Csc / Cot Mnemonic ---
        rule4_title = Text("त्यसैगरी 'Co' फलनहरूको लागि अर्को जादु (Csc-Csc-Cot)", font="Noto Sans Devanagari", font_size=28, color=PURPLE).shift(UP * 2)
        self.play(FadeIn(rule4_title, shift=DOWN))
        self.wait(1)

        magic_words2 = MathTex(r"\csc(x)", r"\quad \csc(x)", r"\quad \cot(x)", font_size=42).next_to(rule4_title, DOWN, buff=0.5)
        magic_words2[0].set_color(PINK)
        magic_words2[1].set_color(PINK)
        magic_words2[2].set_color(PURPLE)
        self.play(FadeIn(magic_words2, scale=1.2))
        self.wait(1.5)

        # Cot लाई छोप्दा -> Csc * Csc (with minus)
        cot_box = SurroundingRectangle(magic_words2[2], color=RED, fill_color=BLACK, fill_opacity=0.8)
        self.play(Create(cot_box))
        eq_cot = MathTex(r"\frac{d}{dx}\cot(x) = -\csc^2(x)", font_size=36).next_to(magic_words2, DOWN, buff=0.5)
        self.play(Write(eq_cot))
        self.wait(2)
        self.play(FadeOut(cot_box))

        # Csc लाई छोप्दा -> Csc * Cot (with minus)
        csc_box = SurroundingRectangle(magic_words2[0], color=RED, fill_color=BLACK, fill_opacity=0.8)
        self.play(Create(csc_box))
        eq_csc = MathTex(r"\frac{d}{dx}\csc(x) = -\csc(x)\cdot\cot(x)", font_size=36).next_to(eq_cot, DOWN, buff=0.3)
        self.play(Write(eq_csc))
        self.wait(2)
        
        # 'C' रुलको रिमाइन्डरलाई अझ प्रभावकारी बनाइएको
        final_note = Text("('C' नियम सम्झनुहोस्: Csc र Cot दुवै 'C' बाट सुरु हुने भएकोले नतिजामा '-' लाग्छ!)", font="Noto Sans Devanagari", font_size=20, color=RED).next_to(eq_csc, DOWN, buff=0.5)
        self.play(Write(final_note))
        self.wait(3)

        # ==========================================
        # 5. Grand Finale 
        # ==========================================
        self.play(
            *[FadeOut(mob) for mob in self.mobjects if mob != title]
        )
        
        success_text = Text("यी जादुयी तरिकाहरू बुझेपछि, अब कुनै पनि सुत्र घोकेर रट्नु पर्दैन!", font="Noto Sans Devanagari", font_size=28, color=GREEN).move_to(ORIGIN)
        self.play(Write(success_text))
        self.play(Wiggle(success_text))
        self.wait(2)

%manim -qk -v warning TrigMnemonics
```
