# Continuity and Limits: Concepts & Examples

## 1. Core Concepts

### What is a Limit?
A limit determines the value a function $f(x)$ approaches as $x$ gets infinitely close to a specific point $c$. We check this from two directions (Left and Right).

### The "h" Method for Calculation
To calculate limits algebraically at a point $x = c$, we introduce a tiny positive number $h$ (where $h \to 0$).

* **Left Hand Limit (LHL):** We substitute $x = c - h$.
    $$\text{LHL} = \lim_{h \to 0} f(c - h)$$
* **Right Hand Limit (RHL):** We substitute $x = c + h$.
    $$\text{RHL} = \lim_{h \to 0} f(c + h)$$

### Condition for Continuity
A function is **Continuous** at $x = c$ if the limit exists and equals the function's value:
$$\text{LHL} = \text{RHL} = f(c)$$

---

## 2. Calculation Steps for Continuous Examples
*In these examples, the function is defined and smooth at $x=0$. We can often use **Direct Substitution**.*

**1. Quadratic ($f(x) = 0.5x^2$)**
* **LHL:** $0.5(0 - h)^2 = 0$
* **RHL:** $0.5(0 + h)^2 = 0$
* **Result:** Continuous ($0=0$).

**2. Absolute Value ($f(x) = |x|$)**
* **LHL:** $|0 - h| = |-h| = h \to 0$
* **RHL:** $|0 + h| = |h| = h \to 0$
* **Result:** Continuous ($0=0$).

**3. Sine Wave ($f(x) = \sin(x) + 1$)**
* **LHL:** $\sin(0 - h) + 1 = - \sin(h) + 1 \to 1$
* **RHL:** $\sin(0 + h) + 1 = \sin(h) + 1 \to 1$
* **Result:** Continuous ($1=1$).

**4. Exponential ($f(x) = e^x - 2$)**
* **LHL:** $e^{0-h} - 2 = e^0 - 2 = -1$
* **RHL:** $e^{0+h} - 2 = e^0 - 2 = -1$
* **Result:** Continuous ($-1=-1$).

**5. Cubic ($f(x) = 0.5(x^3 - 3x)$)**
* **LHL:** $0.5((0)^3 - 3(0)) = 0$
* **RHL:** $0.5((0)^3 - 3(0)) = 0$
* **Result:** Continuous ($0=0$).

---

## 3. Calculation Steps for Discontinuous Examples
*Here, we analyze the behavior as $h \to 0$ carefully because direct substitution often results in undefined values.*

**6. Signum / Jump ($f(x) = \frac{x}{|x|}$)**
* **LHL ($x = -h$):** $\frac{-h}{|-h|} = \frac{-h}{h} = -1$
* **RHL ($x = +h$):** $\frac{h}{|h|} = \frac{h}{h} = 1$
* **Result:** **Jump Discontinuity** ($LHL \neq RHL$).

**7. Infinite Reciprocal ($f(x) = \frac{1}{x}$)**
* **LHL ($x = -h$):** $\frac{1}{-h} \to -\infty$
* **RHL ($x = +h$):** $\frac{1}{h} \to +\infty$
* **Result:** **Infinite Discontinuity**.

**8. Removable / Hole ($f(x) = \frac{\sin x}{x}$)**
* **Concept:** Standard limit identity. As $x$ becomes tiny, $\sin(x) \approx x$.
* **LHL:** $\lim_{x \to 0^-} \frac{\sin x}{x} = 1$
* **RHL:** $\lim_{x \to 0^+} \frac{\sin x}{x} = 1$
* **Result:** **Removable Discontinuity**. The limit exists (1), but $f(0)$ is undefined (division by zero).

**9. Piecewise Shift**
* *Function:* $x+1$ if $x < 0$, and $x-1$ if $x \ge 0$.
* **LHL ($x < 0$):** $(0-h) + 1 \to 1$
* **RHL ($x \ge 0$):** $(0+h) - 1 \to -1$
* **Result:** **Jump Discontinuity** ($1 \neq -1$).

**10. Inverse Trig Jump ($f(x) = \arctan(1/x)$)**
* **LHL ($x = -h$):** $\arctan(\frac{1}{-h}) = \arctan(-\infty) = -\frac{\pi}{2}$
* **RHL ($x = +h$):** $\arctan(\frac{1}{h}) = \arctan(+\infty) = +\frac{\pi}{2}$
* **Result:** **Jump Discontinuity** (Finite Gap).

```python
from manim import *
import numpy as np

class ContinuityAndLimit(Scene):
    def construct(self):
        # --- COLORS & CONFIG ---
        self.lhl_color = TEAL
        self.rhl_color = ORANGE
        
        # --- SETUP ---
        self.setup_layout()
        self.intro_sequence()
        
        # --- DATASETS ---
        # Format: (Func, Func_Tex, C_val, Name, Type, LHL_Math, RHL_Math)
        
        # === PART 1: CONTINUOUS EXAMPLES ===
        continuous_examples = [
            (
                lambda x: 0.5 * x**2,
                r"0.5 x^2",
                0,
                "1. Quadratic (Polynomial)",
                "continuous",
                r"\lim_{x \to 0^-} 0.5x^2 \\ = 0.5(0)^2 = 0",
                r"\lim_{x \to 0^+} 0.5x^2 \\ = 0.5(0)^2 = 0"
            ),
            (
                lambda x: abs(x),
                r"|x|",
                0,
                "2. Absolute Value",
                "continuous",
                r"\lim_{x \to 0^-} (-x) \\ = 0",
                r"\lim_{x \to 0^+} (+x) \\ = 0"
            ),
            (
                lambda x: np.sin(x) + 1,
                r"\sin(x) + 1",
                0,
                "3. Sine Wave",
                "continuous",
                r"\lim_{x \to 0^-} (\sin(x)+1) \\ = \sin(0)+1 = 1",
                r"\lim_{x \to 0^+} (\sin(x)+1) \\ = \sin(0)+1 = 1"
            ),
            (
                lambda x: np.exp(x) - 2,
                r"e^x - 2",
                0,
                "4. Exponential",
                "continuous",
                r"\lim_{x \to 0^-} (e^x - 2) \\ = e^0 - 2 = -1",
                r"\lim_{x \to 0^+} (e^x - 2) \\ = e^0 - 2 = -1"
            ),
            (
                lambda x: 0.5 * (x**3 - 3*x),
                r"\frac{1}{2}(x^3 - 3x)",
                0,
                "5. Cubic Function",
                "continuous",
                r"\lim_{x \to 0^-} \frac{1}{2}(x^3-3x) \\ = 0",
                r"\lim_{x \to 0^+} \frac{1}{2}(x^3-3x) \\ = 0"
            ),
        ]

        # === PART 2: DISCONTINUOUS EXAMPLES ===
        discontinuous_examples = [
            (
                lambda x: -1 if x < 0 else 1,
                r"\frac{x}{|x|} \text{ (Signum)}",
                0,
                "6. Jump Discontinuity",
                "jump",
                r"\lim_{x \to 0^-} \frac{x}{|x|} \\ = -1",
                r"\lim_{x \to 0^+} \frac{x}{|x|} \\ = +1"
            ),
            (
                lambda x: 1/x if x != 0 else 0,
                r"\frac{1}{x}",
                0,
                "7. Infinite Discontinuity",
                "infinite",
                r"\lim_{x \to 0^-} \frac{1}{x} \\ = -\infty",
                r"\lim_{x \to 0^+} \frac{1}{x} \\ = +\infty"
            ),
            (
                lambda x: np.sin(x)/x if abs(x) > 0.01 else 1, # Value is 1, but hole logic will draw circle
                r"\frac{\sin(x)}{x}",
                0,
                "8. Removable (Hole)",
                "hole",
                r"\lim_{x \to 0} \frac{\sin(x)}{x} \\ = 1",
                r"\lim_{x \to 0} \frac{\sin(x)}{x} \\ = 1"
            ),
            (
                lambda x: x + 1 if x < 0 else x - 1,
                r"\begin{cases} x+1 & x < 0 \\ x-1 & x \ge 0 \end{cases}",
                0,
                "9. Piecewise Shift",
                "jump",
                r"\lim_{x \to 0^-} (x+1) \\ = 0 + 1 = 1",
                r"\lim_{x \to 0^+} (x-1) \\ = 0 - 1 = -1"
            ),
            (
                lambda x: np.arctan(1/x) if abs(x) > 0.001 else 0,
                r"\arctan(1/x)",
                0,
                "10. Inverse Trig Jump",
                "jump",
                r"\lim_{x \to 0^-} \arctan(1/x) \\ = -\frac{\pi}{2} \approx -1.57",
                r"\lim_{x \to 0^+} \arctan(1/x) \\ = +\frac{\pi}{2} \approx +1.57"
            )
        ]

        # Combine lists
        all_examples = continuous_examples + discontinuous_examples

        for data in all_examples:
            self.animate_example(*data)

    def setup_layout(self):
        self.separator = Line(UP * 4, DOWN * 4).shift(LEFT * 1)
        self.axes = Axes(
            x_range=[-3.5, 3.5, 1],
            y_range=[-2.5, 4.5, 1],
            x_length=6,
            y_length=6,
            axis_config={"include_numbers": True, "font_size": 20, "color": GREY},
            tips=False
        ).to_edge(RIGHT)
        self.axes_labels = self.axes.get_axis_labels(x_label="x", y_label="f(x)")

    def intro_sequence(self):
        title = Title(r"Continuity and Limit").scale(0.9)
        self.play(Write(title))
        self.play(Create(self.separator))
        self.wait(1)
        self.play(FadeOut(title))

    def animate_example(self, func, func_tex, c, name, d_type, lhl_math, rhl_math):
        # --- 1. LABELS ---
        name_lbl = Text(name, font_size=32, color=YELLOW).to_edge(LEFT).shift(UP*3.6)
        func_lbl = MathTex(f"f(x) = {func_tex}", font_size=36).next_to(name_lbl, DOWN, aligned_edge=LEFT, buff=0.2)
        
        self.add(name_lbl, func_lbl)
        self.play(Create(self.axes), Write(self.axes_labels))

        # --- 2. PLOT GRAPH ---
        graph_group = self.get_graph_visuals(func, c, d_type)
        self.play(Create(graph_group), run_time=1.0)

        # --- 3. LHL ---
        lhl_header = Tex(r"\textbf{Left Hand Limit} ($x \to c^-$):", color=self.lhl_color, font_size=42).next_to(func_lbl, DOWN, buff=0.6).align_to(name_lbl, LEFT)
        lhl_steps = MathTex(lhl_math, font_size=34, color=WHITE).next_to(lhl_header, DOWN, aligned_edge=LEFT, buff=0.15)
        
        self.play(Write(lhl_header), Write(lhl_steps))
        lhl_val, lhl_dot = self.animate_limit_approach(func, c, "left", d_type)

        # --- 4. RHL ---
        rhl_header = Tex(r"\textbf{Right Hand Limit} ($x \to c^+$):", color=self.rhl_color, font_size=42).next_to(lhl_steps, DOWN, buff=0.5).align_to(name_lbl, LEFT)
        rhl_steps = MathTex(rhl_math, font_size=34, color=WHITE).next_to(rhl_header, DOWN, aligned_edge=LEFT, buff=0.15)
        
        self.play(Write(rhl_header), Write(rhl_steps))
        rhl_val, rhl_dot = self.animate_limit_approach(func, c, "right", d_type)

        # --- 5. CONCLUSION ---
        if d_type == "continuous":
            res_text = Text("Result: Continuous", color=GREEN, font_size=36, weight=BOLD)
            self.play(Transform(lhl_dot, rhl_dot), FadeOut(rhl_dot))
            self.play(Flash(lhl_dot, color=GREEN, flash_radius=0.5))
        else:
            res_text = Text("Result: Discontinuous", color=RED, font_size=36, weight=BOLD)
            # Add visual cue for gap/infinity
            if d_type == "infinite":
                pass # Arrows already handle this
            elif d_type == "hole":
                # Flash the hole
                hole_circ = Circle(radius=0.15, color=RED).move_to(self.axes.coords_to_point(c, lhl_val))
                self.play(Create(hole_circ))
                self.play(FadeOut(hole_circ))
            else: # Jump
                brace = BraceBetweenPoints(lhl_dot.get_center(), rhl_dot.get_center(), direction=RIGHT)
                self.play(Create(brace))
                self.play(FadeOut(brace))

        res_text.next_to(rhl_steps, DOWN, buff=0.6).align_to(name_lbl, LEFT)
        self.play(Write(res_text))
        self.wait(1.5)

        # Cleanup
        self.play(
            *[FadeOut(m) for m in self.mobjects if m is not self.separator],
            run_time=0.5
        )

    def get_graph_visuals(self, func, c, d_type):
        g_group = VGroup()
        if d_type == "continuous": 
            p1 = self.axes.plot(func, x_range=[-3.5, 3.5, 0.05], color=WHITE, use_smoothing=True)
            g_group.add(p1)
        elif d_type == "jump":
            p1 = self.axes.plot(func, x_range=[-3.5, c - 0.01, 0.05], color=WHITE)
            p2 = self.axes.plot(func, x_range=[c + 0.01, 3.5, 0.05], color=WHITE)
            
            # Add endpoint circles (Left Open, Right Closed generally, or based on specific func)
            l_val = func(c-0.01)
            r_val = func(c+0.01)
            l_dot = Circle(radius=0.08, color=WHITE, stroke_width=2).move_to(self.axes.coords_to_point(c, l_val))
            r_dot = Dot(self.axes.coords_to_point(c, r_val), color=WHITE)
            g_group.add(p1, p2, l_dot, r_dot)
            
        elif d_type == "hole":
            # For sinc function or similar removable discontinuity
            p1 = self.axes.plot(func, x_range=[-3.5, 3.5, 0.05], color=WHITE)
            # Calculate limit value (approximate via small epsilon)
            limit_val = func(c + 0.0001)
            hole = Circle(radius=0.08, color=BLACK, fill_opacity=1, stroke_color=WHITE, stroke_width=2).move_to(self.axes.coords_to_point(c, limit_val))
            g_group.add(p1, hole)
            
        elif d_type == "infinite":
            p1 = self.axes.plot(func, x_range=[-3.5, -0.1], color=WHITE)
            p2 = self.axes.plot(func, x_range=[0.1, 3.5], color=WHITE)
            g_group.add(p1, p2)
            
        return g_group

    def animate_limit_approach(self, func, c, direction, d_type):
        if direction == "left":
            start_x, end_x = c - 2.0, c - 0.02
            color = self.lhl_color
        else:
            start_x, end_x = c + 2.0, c + 0.02
            color = self.rhl_color

        t = ValueTracker(start_x)
        d = Dot(color=color)
        
        def update_dot(mob):
            x = t.get_value()
            try:
                y = func(x)
                # Clamp for visuals
                if y > 5: y = 5 
                if y < -3: y = -3
                mob.move_to(self.axes.coords_to_point(x, y))
            except:
                pass
                
        d.add_updater(update_dot)
        
        # Label
        lbl = DecimalNumber(0, num_decimal_places=2, font_size=24, color=color)
        lbl.add_updater(lambda m: m.next_to(d, UP, buff=0.1).set_value(func(t.get_value())))

        h_line = always_redraw(lambda: self.axes.get_horizontal_line(d.get_center(), line_config={"dashed_ratio": 0.5, "color": color}))
        v_line = always_redraw(lambda: self.axes.get_vertical_line(d.get_center(), line_config={"dashed_ratio": 0.5, "color": color}))

        self.add(d, h_line, v_line, lbl)
        self.play(t.animate.set_value(end_x), run_time=1.5, rate_func=linear) # Sped up slightly for 10 examples
        self.remove(d, h_line, v_line, lbl)
        
        # Determine Final Dot Position
        final_y = func(end_x)
        
        # Visual handling for infinity
        if d_type == "infinite":
            if final_y > 4: 
                # Draw arrow up
                arrow = Arrow(start=self.axes.coords_to_point(c, 3), end=self.axes.coords_to_point(c, 4.5), color=color)
                self.add(arrow)
                return final_y, arrow
            elif final_y < -2:
                # Draw arrow down
                arrow = Arrow(start=self.axes.coords_to_point(c, -1.5), end=self.axes.coords_to_point(c, -3), color=color)
                self.add(arrow)
                return final_y, arrow
        
        # Visual handling for hole
        if d_type == "hole":
             final_y = func(c + 0.0001) # Approximate limit

        # Clamp for safety
        if final_y > 10: final_y = 4.5
        if final_y < -10: final_y = -2.5

        final_dot = Dot(self.axes.coords_to_point(c, final_y), color=color).scale(1.2)
        self.add(final_dot)
        return final_y, final_dot


%manim -ql -v warning ContinuityAndLimit
```
