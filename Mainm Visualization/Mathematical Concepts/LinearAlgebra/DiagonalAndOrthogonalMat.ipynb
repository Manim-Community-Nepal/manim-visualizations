```python
from manim import *
import numpy as np

# -------------------------
# Palette & small constants
# -------------------------
COLOR_PALETTE = {
    "e1": RED,
    "e2": GREEN,
    "e3": BLUE,
    "c1": YELLOW,
    "c2": PURPLE,
    "c3": TEAL,
    "panel_bg": "#1f2937",
}

LABEL_SCALE_SMALL = 0.7
LABEL_SCALE_MED  = 0.9

# -------------------------
# Scene
# -------------------------
class DiagonalOrthogonalExplainer(ThreeDScene):
    """
    Robust, refactored ManimCE scene:
      - uses ThreeDVector fallback (no Arrow3D)
      - avoids overlapping text by directional offsets and staggered fades
      - clears updaters safely before removing mobjects
      - keeps only one fixed-in-frame panel at a time
      - ends with interactive_embed() when available
    """

    def setup(self):
        # layout & camera
        self.left_shift = LEFT * 3.0
        self.set_camera_orientation(phi=30 * DEGREES, theta=-45 * DEGREES)
        # currently-displayed fixed panel (a single VGroup)
        self.current_panel = None

    # -------------------------
    # Utilities
    # -------------------------
    def _make_vec(self, endpoint, color=YELLOW, stroke_width=3):
        """
        Cross-version safe 3D arrow creation.
        Uses ThreeDVector (ManimCE standard). If Arrow3D exists, try it but fallback.
        """
        endpoint = np.array(endpoint, dtype=float)
        try:
            # prefer ThreeDVector (stable in ManimCE)
            arr = ThreeDVector(endpoint, stroke_width=stroke_width)
        except Exception:
            # second fallback: try Arrow (2D) placed in 3D (less ideal)
            arr = Arrow(start=ORIGIN, end=endpoint, buff=0)
        arr.set_color(color)
        arr.shift(self.left_shift)
        # set_shade_in_3d exists for vector-like objects; ignore errors gracefully
        try:
            arr.set_shade_in_3d(True)
        except Exception:
            pass
        return arr

    def _directional_label_offset(self, endpoint, radial=0.16, up_bias=0.06):
        """Small offset away from tip direction to reduce collisions."""
        e = np.array(endpoint, dtype=float)
        if np.linalg.norm(e) < 1e-8:
            return np.array([0.0, up_bias, 0.0])
        base = np.array([e[0], e[1], 0.0])
        if np.linalg.norm(base) < 1e-6:
            base = np.array([0.0, 0.0, 1.0])
        base = base / (np.linalg.norm(base) + 1e-9)
        bias = np.array([0.0, up_bias, 0.0])
        return base * radial + bias

    def _make_tip_label(self, endpoint, text, color=WHITE, scale=1.0, connector=True):
        """
        Make a MathTex label positioned slightly offset from the 3D endpoint.
        Returns (label, connector_or_None)
        """
        endpoint = np.array(endpoint, dtype=float)
        offset = self._directional_label_offset(endpoint, radial=0.18, up_bias=0.06)
        label_pos = self.left_shift + endpoint * 1.05 + offset
        lbl = MathTex(text, font_size=24 * scale)
        lbl.set_color(color)
        lbl.move_to(label_pos)
        conn = None
        if connector:
            start = label_pos
            end = self.left_shift + endpoint * 0.98
            conn = Line(start, end, stroke_width=1)
            try:
                conn.set_color(WHITE)
            except Exception:
                pass
        return lbl, conn

    def _make_panel(self, title_text, bullet_texts, title_size=32, bullet_size=20):
        """
        Build a single panel (background + content). Returns (panel_VGroup, bullets_VGroup).
        Panel is a VGroup([bg, content]).
        """
        def esc(s):
            return s.replace("&", "&amp;")   # fixes all future bullet lines
        title = MarkupText(esc(title_text), font_size=title_size)
        bullets = VGroup(*[
            MarkupText("• " + esc(t), font_size=bullet_size)
            for t in bullet_texts
        ])
        content = VGroup(title, bullets).arrange(DOWN, aligned_edge=LEFT, buff=0.22)
        # background rectangle sized to content
        padding = 0.25
        bg = RoundedRectangle(corner_radius=0.18,
                              height=content.height + padding,
                              width=content.width + 0.6)
        bg.set_fill(COLOR_PALETTE["panel_bg"], opacity=0.80)
        bg.set_stroke(width=0)
        panel = VGroup(bg, content).arrange(DOWN, aligned_edge=LEFT, buff=0.12)
        panel.to_edge(RIGHT).shift(UP * 0.6)
        # ensure content sits visually on top of bg
        content.move_to(panel.get_center(), aligned_edge=ORIGIN)
        return panel, bullets

    def _replace_panel(self, title_text, bullet_texts, run_time=0.5):
        """
        Replace the currently displayed panel with a new one.
        Ensures we only ever have one fixed-in-frame panel.
        Bullets fade in with a lag to prevent overlap.
        """
        new_panel, bullets = self._make_panel(title_text, bullet_texts)
        # remove old panel safely
        if self.current_panel is not None:
            try:
                self.remove_fixed_in_frame_mobjects(self.current_panel)
            except Exception:
                # fallback: attempt to remove children individually
                for sub in self.current_panel:
                    try:
                        self.remove_fixed_in_frame_mobjects(sub)
                    except Exception:
                        pass
        # add new panel and animate
        self.add_fixed_in_frame_mobjects(new_panel)
        if self.current_panel is not None:
            self.play(FadeOut(self.current_panel), FadeIn(new_panel), run_time=run_time)
        else:
            self.play(FadeIn(new_panel), run_time=run_time)
        # stagger bullet fade-ins (safer than single-block addition)
        try:
            self.play(LaggedStart(*[FadeIn(b, shift=UP * 0.12) for b in bullets], lag_ratio=0.16), run_time=0.5)
        except Exception:
            # fallback: simple FadeIn
            for b in bullets:
                self.play(FadeIn(b), run_time=0.12)
        self.current_panel = new_panel

    # -------------------------
    # Scene flow
    # -------------------------
    def construct(self):
        # initial panel
        self._replace_panel("<b>Diagonal &amp; Orthogonal Matrices</b>",
                            ["Diagonal: axis-aligned scaling",
                             "Orthogonal: rotations/reflections (length-preserving)"])
        self.wait(0.35)

        # Phase A
        mgrp_a = self.phase_a_diagonal()
        self.wait(0.35)

        # fade out phase A
        self.play(FadeOut(mgrp_a), run_time=0.6)
        self.wait(0.2)

        # Phase B
        mgrp_b, extra_panel = self.phase_b_orthogonal()
        self.wait(0.35)

        # remove extra fixed panel if present, then fade Phase B
        if extra_panel is not None:
            try:
                self.remove_fixed_in_frame_mobjects(extra_panel)
            except Exception:
                try:
                    self.remove(extra_panel)
                except Exception:
                    pass
        self.play(FadeOut(mgrp_b), run_time=0.6)
        self.wait(0.2)

        # summary
        self._replace_panel("<b>Summary</b>",
                            ["Diagonal: scales coordinates (axis-aligned).",
                             "Orthogonal: rotations/reflections (preserve lengths & angles)."])
        self.wait(0.5)

        # interactive end if available
        try:
            self.interactive_embed()
        except Exception:
            # not available: finish normally
            self.wait(0.8)

    # -------------------------
    # Phase A implementation
    # -------------------------
    def phase_a_diagonal(self):
        axes = ThreeDAxes(x_range=[-4, 4, 1], y_range=[-4, 4, 1], z_range=[-4, 4, 1],
                          x_length=5, y_length=5, z_length=5).shift(self.left_shift)
        x_lbl = axes.get_x_axis_label(MathTex("x")).shift(self.left_shift + 0.3 * RIGHT)
        y_lbl = axes.get_y_axis_label(MathTex("y")).shift(self.left_shift + 0.3 * UP)
        z_lbl = axes.get_z_axis_label(MathTex("z")).shift(self.left_shift + 0.3 * OUT)

        e1 = np.array([1.5, 0.0, 0.0])
        e2 = np.array([0.0, 1.2, 0.0])
        e3 = np.array([0.0, 0.0, 1.0])

        v1 = self._make_vec(e1, color=COLOR_PALETTE["e1"], stroke_width=3)
        v2 = self._make_vec(e2, color=COLOR_PALETTE["e2"], stroke_width=3)
        v3 = self._make_vec(e3, color=COLOR_PALETTE["e3"], stroke_width=3)
        v1_lbl, v1_conn = self._make_tip_label(e1, "e_1", COLOR_PALETTE["e1"], scale=LABEL_SCALE_MED)
        v2_lbl, v2_conn = self._make_tip_label(e2, "e_2", COLOR_PALETTE["e2"], scale=LABEL_SCALE_MED)
        v3_lbl, v3_conn = self._make_tip_label(e3, "e_3", COLOR_PALETTE["e3"], scale=LABEL_SCALE_MED)

        # show base geometry
        self.add(axes, x_lbl, y_lbl, z_lbl)
        # staggered fade-in for vectors and labels
        self.play(LaggedStart(FadeIn(v1), FadeIn(v2), FadeIn(v3),
                              FadeIn(v1_lbl), FadeIn(v2_lbl), FadeIn(v3_lbl),
                              lag_ratio=0.1), run_time=0.9)
        # connectors
        for c in (v1_conn, v2_conn, v3_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.18)
        self.wait(0.45)

        # Uniform scaling
        self._replace_panel("<b>Diagonal Matrix — Uniform scaling</b>", ["D = diag(2,2,2) → uniform scaling by 2"])
        D_uniform = np.diag([2.0, 2.0, 2.0])
        v1_u = self._make_vec(D_uniform @ e1, color=COLOR_PALETTE["e1"], stroke_width=2)
        v2_u = self._make_vec(D_uniform @ e2, color=COLOR_PALETTE["e2"], stroke_width=2)
        v3_u = self._make_vec(D_uniform @ e3, color=COLOR_PALETTE["e3"], stroke_width=2)
        v1u_lbl, v1u_conn = self._make_tip_label(D_uniform @ e1, "D e_1", COLOR_PALETTE["e1"], scale=LABEL_SCALE_SMALL)
        v2u_lbl, v2u_conn = self._make_tip_label(D_uniform @ e2, "D e_2", COLOR_PALETTE["e2"], scale=LABEL_SCALE_SMALL)
        v3u_lbl, v3u_conn = self._make_tip_label(D_uniform @ e3, "D e_3", COLOR_PALETTE["e3"], scale=LABEL_SCALE_SMALL)

        # transform vectors; fade original labels out first to avoid overlap
        self.play(Transform(v1, v1_u), Transform(v2, v2_u), Transform(v3, v3_u),
                  FadeOut(v1_lbl), FadeOut(v2_lbl), FadeOut(v3_lbl), run_time=1.1)
        for lbl in (v1u_lbl, v2u_lbl, v3u_lbl):
            self.play(FadeIn(lbl), run_time=0.22)
        for c in (v1u_conn, v2u_conn, v3u_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.16)
        self.wait(0.6)

        # short numeric demonstration (temporary)
        before_len = np.linalg.norm(e1)
        after_len = np.linalg.norm(D_uniform @ e1)
        before_lbl = DecimalNumber(before_len, num_decimal_places=2).next_to(self.left_shift + e1, DOWN + RIGHT * 0.15)
        after_lbl  = DecimalNumber(after_len, num_decimal_places=2).next_to(self.left_shift + D_uniform @ e1, DOWN + RIGHT * 0.15)
        temp_metrics = VGroup(before_lbl, after_lbl)
        # show and then remove safely
        self.play(FadeIn(temp_metrics), run_time=0.45)
        self.wait(0.6)
        for mob in temp_metrics:
            try:
                mob.clear_updaters()
            except Exception:
                pass
        self.play(FadeOut(temp_metrics), run_time=0.35)
        try:
            self.remove(*temp_metrics)
        except Exception:
            pass

        # Anisotropic scaling
        self._replace_panel("<b>Diagonal Matrix — Anisotropic scaling</b>",
                            ["D = diag(1.2, 0.5, 2.0) → axis-dependent scaling"])
        D_aniso = np.diag([1.2, 0.5, 2.0])
        v1_a = self._make_vec(D_aniso @ e1, color=COLOR_PALETTE["e1"], stroke_width=2)
        v2_a = self._make_vec(D_aniso @ e2, color=COLOR_PALETTE["e2"], stroke_width=2)
        v3_a = self._make_vec(D_aniso @ e3, color=COLOR_PALETTE["e3"], stroke_width=2)
        v1a_lbl, v1a_conn = self._make_tip_label(D_aniso @ e1, "D e_1", COLOR_PALETTE["e1"], scale=LABEL_SCALE_SMALL)
        v2a_lbl, v2a_conn = self._make_tip_label(D_aniso @ e2, "D e_2", COLOR_PALETTE["e2"], scale=LABEL_SCALE_SMALL)
        v3a_lbl, v3a_conn = self._make_tip_label(D_aniso @ e3, "D e_3", COLOR_PALETTE["e3"], scale=LABEL_SCALE_SMALL)

        self.play(Transform(v1, v1_a), Transform(v2, v2_a), Transform(v3, v3_a),
                  FadeOut(v1u_lbl), FadeOut(v2u_lbl), FadeOut(v3u_lbl), run_time=1.0)
        for lbl in (v1a_lbl, v2a_lbl, v3a_lbl):
            self.play(FadeIn(lbl), run_time=0.22)
        for c in (v1a_conn, v2a_conn, v3a_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.14)

        # numeric lengths (temporary)
        l1 = np.linalg.norm(D_aniso @ e1)
        l2 = np.linalg.norm(D_aniso @ e2)
        l3 = np.linalg.norm(D_aniso @ e3)
        l1_lbl = DecimalNumber(l1, num_decimal_places=2).next_to(self.left_shift + D_aniso @ e1, DOWN + RIGHT * 0.08)
        l2_lbl = DecimalNumber(l2, num_decimal_places=2).next_to(self.left_shift + D_aniso @ e2, DOWN + RIGHT * 0.08)
        l3_lbl = DecimalNumber(l3, num_decimal_places=2).next_to(self.left_shift + D_aniso @ e3, DOWN + RIGHT * 0.08)
        metrics = VGroup(l1_lbl, l2_lbl, l3_lbl)
        self.play(FadeIn(metrics), run_time=0.5)
        self.wait(0.9)
        for mob in metrics:
            try:
                mob.clear_updaters()
            except Exception:
                pass
        self.play(FadeOut(metrics), run_time=0.35)
        try:
            self.remove(*metrics)
        except Exception:
            pass

        # small illustrative arrows (positioned away from labels)
        arrow_shrink = Arrow(self.left_shift + np.array([0.5, 1.0, 0]),
                             self.left_shift + np.array([0.5, 0.7, 0]), buff=0)
        shrink_lbl = Text("squash", font_size=18).next_to(arrow_shrink, RIGHT)
        arrow_stretch = Arrow(self.left_shift + np.array([0.15, 0.15, 1.0]),
                              self.left_shift + np.array([0.15, 0.15, 1.9]), buff=0)
        stretch_lbl = Text("stretch", font_size=18).next_to(arrow_stretch, RIGHT)
        self.play(GrowArrow(arrow_shrink), FadeIn(shrink_lbl), GrowArrow(arrow_stretch), FadeIn(stretch_lbl))
        self.wait(0.7)

        # return group of all 3D objects created in this phase
        items = VGroup(axes, x_lbl, y_lbl, z_lbl,
                       v1, v2, v3,
                       v1_a, v2_a, v3_a,
                       v1a_lbl, v2a_lbl, v3a_lbl,
                       arrow_shrink, shrink_lbl, arrow_stretch, stretch_lbl)
        return items

    # -------------------------
    # Phase B implementation
    # -------------------------
    def phase_b_orthogonal(self):
        self._replace_panel("<b>Orthogonal Matrix — Rotation / Reflection</b>",
                            ["Preserves lengths and inner products", "Columns are orthonormal"])

        axes_b = ThreeDAxes(x_range=[-3.5, 3.5, 1], y_range=[-3.5, 3.5, 1], z_range=[-3.5, 3.5, 1],
                            x_length=4.5, y_length=4.5, z_length=4.5).shift(self.left_shift)
        x_b = axes_b.get_x_axis_label(MathTex("x")).shift(self.left_shift + 0.25 * RIGHT)
        y_b = axes_b.get_y_axis_label(MathTex("y")).shift(self.left_shift + 0.25 * UP)
        z_b = axes_b.get_z_axis_label(MathTex("z")).shift(self.left_shift + 0.25 * OUT)
        self.play(FadeIn(axes_b), FadeIn(x_b), FadeIn(y_b), FadeIn(z_b), run_time=0.6)

        e1b = np.array([1.5, 0.0, 0.0])
        e2b = np.array([0.0, 1.2, 0.0])
        e3b = np.array([0.0, 0.0, 1.0])

        vb1 = self._make_vec(e1b, color=COLOR_PALETTE["e1"], stroke_width=3)
        vb2 = self._make_vec(e2b, color=COLOR_PALETTE["e2"], stroke_width=3)
        vb3 = self._make_vec(e3b, color=COLOR_PALETTE["e3"], stroke_width=3)
        vb1_lbl, vb1_conn = self._make_tip_label(e1b, "v_1", COLOR_PALETTE["e1"], scale=LABEL_SCALE_MED)
        vb2_lbl, vb2_conn = self._make_tip_label(e2b, "v_2", COLOR_PALETTE["e2"], scale=LABEL_SCALE_MED)
        vb3_lbl, vb3_conn = self._make_tip_label(e3b, "v_3", COLOR_PALETTE["e3"], scale=LABEL_SCALE_MED)

        self.play(LaggedStart(FadeIn(vb1), FadeIn(vb2), FadeIn(vb3),
                              FadeIn(vb1_lbl), FadeIn(vb2_lbl), FadeIn(vb3_lbl),
                              lag_ratio=0.12), run_time=0.9)
        for c in (vb1_conn, vb2_conn, vb3_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.15)
        self.wait(0.45)

        # Rotation about z
        theta = 50 * DEGREES
        Rz = np.array([[np.cos(theta), -np.sin(theta), 0.0],
                       [np.sin(theta),  np.cos(theta), 0.0],
                       [0.0,            0.0,           1.0]])
        vb1_end = Rz @ e1b
        vb2_end = Rz @ e2b
        vb3_end = Rz @ e3b

        vb1_rot = self._make_vec(vb1_end, color=COLOR_PALETTE["e1"], stroke_width=2)
        vb2_rot = self._make_vec(vb2_end, color=COLOR_PALETTE["e2"], stroke_width=2)
        vb3_rot = self._make_vec(vb3_end, color=COLOR_PALETTE["e3"], stroke_width=2)
        vb1r_lbl, vb1r_conn = self._make_tip_label(vb1_end, "R v_1", COLOR_PALETTE["e1"], scale=LABEL_SCALE_SMALL)
        vb2r_lbl, vb2r_conn = self._make_tip_label(vb2_end, "R v_2", COLOR_PALETTE["e2"], scale=LABEL_SCALE_SMALL)
        vb3r_lbl, vb3r_conn = self._make_tip_label(vb3_end, "R v_3", COLOR_PALETTE["e3"], scale=LABEL_SCALE_SMALL)

        self.play(Transform(vb1, vb1_rot), Transform(vb2, vb2_rot), Transform(vb3, vb3_rot),
                  FadeOut(vb1_lbl), FadeOut(vb2_lbl), FadeOut(vb3_lbl), run_time=1.3)
        for lbl in (vb1r_lbl, vb2r_lbl, vb3r_lbl):
            self.play(FadeIn(lbl), run_time=0.22)
        for c in (vb1r_conn, vb2r_conn, vb3r_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.12)
        self.wait(0.5)

        # lengths (temporary)
        before_len = np.linalg.norm(e1b)
        after_len = np.linalg.norm(vb1_end)
        b_lbl = DecimalNumber(before_len, num_decimal_places=3).next_to(self.left_shift + e1b, DOWN + RIGHT * 0.12)
        a_lbl = DecimalNumber(after_len, num_decimal_places=3).next_to(self.left_shift + vb1_end, DOWN + RIGHT * 0.12)
        self.play(FadeIn(b_lbl), FadeIn(a_lbl))
        eq_line = Line(b_lbl.get_center(), a_lbl.get_center())
        eq_text = Text("equal", font_size=18).next_to(eq_line, UP)
        self.play(Create(eq_line), FadeIn(eq_text))
        self.wait(0.7)
        for m in (b_lbl, a_lbl, eq_line, eq_text):
            try:
                m.clear_updaters()
            except Exception:
                pass
        self.play(FadeOut(b_lbl), FadeOut(a_lbl), FadeOut(eq_line), FadeOut(eq_text), run_time=0.35)
        try:
            self.remove(b_lbl, a_lbl, eq_line, eq_text)
        except Exception:
            pass

        # columns of R (orthonormal)
        c1 = Rz[:, 0]
        c2 = Rz[:, 1]
        c3 = Rz[:, 2]
        col1 = self._make_vec(c1 * 1.5, color=COLOR_PALETTE["c1"], stroke_width=2)
        col2 = self._make_vec(c2 * 1.5, color=COLOR_PALETTE["c2"], stroke_width=2)
        col3 = self._make_vec(c3 * 1.5, color=COLOR_PALETTE["c3"], stroke_width=2)
        c1_lbl, c1_conn = self._make_tip_label(c1 * 1.5, "c_1", COLOR_PALETTE["c1"], scale=LABEL_SCALE_SMALL)
        c2_lbl, c2_conn = self._make_tip_label(c2 * 1.5, "c_2", COLOR_PALETTE["c2"], scale=LABEL_SCALE_SMALL)
        c3_lbl, c3_conn = self._make_tip_label(c3 * 1.5, "c_3", COLOR_PALETTE["c3"], scale=LABEL_SCALE_SMALL)

        self.play(FadeIn(col1), FadeIn(col2), FadeIn(col3), run_time=0.9)
        for lbl in (c1_lbl, c2_lbl, c3_lbl):
            self.play(FadeIn(lbl), run_time=0.22)
        for c in (c1_conn, c2_conn, c3_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.14)

        # column lengths (temporary)
        c1_len = DecimalNumber(np.linalg.norm(c1), num_decimal_places=3).next_to(self.left_shift + c1 * 1.5, RIGHT)
        c2_len = DecimalNumber(np.linalg.norm(c2), num_decimal_places=3).next_to(self.left_shift + c2 * 1.5, RIGHT)
        c3_len = DecimalNumber(np.linalg.norm(c3), num_decimal_places=3).next_to(self.left_shift + c3 * 1.5, RIGHT)
        self.play(FadeIn(c1_len), FadeIn(c2_len), FadeIn(c3_len))
        self.wait(0.5)
        for m in (c1_len, c2_len, c3_len):
            try:
                m.clear_updaters()
            except Exception:
                pass
        self.play(FadeOut(c1_len), FadeOut(c2_len), FadeOut(c3_len), run_time=0.3)
        try:
            self.remove(c1_len, c2_len, c3_len)
        except Exception:
            pass

        # angle arc between c1 and c2
        c1n = c1 / np.linalg.norm(c1)
        c2n = c2 / np.linalg.norm(c2)
        dotp = float(np.dot(c1, c2))
        angle_between = float(np.arccos(np.clip(np.dot(c1n, c2n), -1, 1)))
        u, w = c1n, c2n
        R = 0.5
        def arc_func(t):
            return self.left_shift + R * (np.cos(t) * u + np.sin(t) * w)
        arc = ParametricFunction(arc_func, t_range=[0, angle_between], stroke_width=3)
        arc_lbl = MathTex(f"{(angle_between/DEGREES):.0f}^\\circ", font_size=20).move_to(
            self.left_shift + 0.9 * (np.cos(angle_between/2) * u + np.sin(angle_between/2) * w)
        )
        self.play(Create(arc), FadeIn(arc_lbl))
        self.wait(0.6)

        # small extra fixed panel with dot product info (explicitly removable by caller)
        dot_text = MarkupText(f"<b>⟨c₁, c₂⟩ = {dotp:.3f}</b>", font_size=20)
        dot_text.to_edge(RIGHT).shift(DOWN * 1.4)
        self.add_fixed_in_frame_mobjects(dot_text)
        self.play(FadeIn(dot_text))
        self.wait(0.9)
        extra_panel = dot_text  # caller will remove it

        # reflection demo on c1
        reflect = np.diag([1.0, -1.0, 1.0])
        ref_end = reflect @ (c1 * 1.5)
        ref_vec = self._make_vec(ref_end, color=COLOR_PALETTE["c1"], stroke_width=2)
        ref_lbl, ref_conn = self._make_tip_label(ref_end, "Ref(c_1)", COLOR_PALETTE["c1"], scale=LABEL_SCALE_SMALL)
        self.play(Transform(col1, ref_vec), FadeOut(c1_lbl), run_time=0.8)
        if ref_conn is not None:
            self.play(GrowFromCenter(ref_conn), run_time=0.15)
        self.play(FadeIn(ref_lbl), run_time=0.18)
        self.wait(0.5)

        all_objs = VGroup(axes_b, x_b, y_b, z_b,
                          vb1, vb2, vb3,
                          vb1r_lbl, vb2r_lbl, vb3r_lbl,
                          col1, col2, col3,
                          c1_lbl, c2_lbl, c3_lbl,
                          arc, arc_lbl,
                          ref_vec, ref_lbl)
        return all_objs, extra_panel


%manim -ql -v warning DiagonalOrthogonalExplainer
```

```python
from manim import *
import numpy as np

# ---------- Config / Colors ----------
PALETTE = {
    "e1": RED,
    "e2": GREEN,
    "e3": BLUE,
    "col1": YELLOW,
    "col2": PURPLE,
    "col3": TEAL,
    "panel_bg": "#1f2937",
}
LABEL_SMALL = 0.75
LABEL_MED = 0.95

# ---------- Helpers ----------
def safe_three_d_vector(vec, color=YELLOW, stroke_width=3):
    """
    Create a 3D vector in a cross-version-safe way.
    Prefer ThreeDVector for ManimCE; fallback to Arrow if missing.
    """
    try:
        mv = ThreeDVector(np.array(vec, dtype=float), stroke_width=stroke_width)
    except Exception:
        # last-resort fallback (2D Arrow placed in 3D coordinates)
        mv = Arrow(ORIGIN, np.array(vec, dtype=float), buff=0)
    mv.set_color(color)
    try:
        mv.set_shade_in_3d(True)
    except Exception:
        pass
    return mv

def directional_label_offset(endpoint, radial=0.18, up_bias=0.06):
    """
    Compute an offset so labels move outward away from the tip, reducing overlap.
    Works in the XY plane plus a small upward bias.
    """
    e = np.array(endpoint, dtype=float)
    if np.linalg.norm(e[:2]) < 1e-6:
        base = np.array([0.0, 0.0, 1.0])
    else:
        base = np.array([e[0], e[1], 0.0])
        base = base / (np.linalg.norm(base) + 1e-9)
    bias = np.array([0.0, up_bias, 0.0])
    return base * radial + bias

def make_tip_label(endpoint, text, color=WHITE, scale=1.0, connector=True, origin_shift=LEFT * 3.0):
    """
    Create a MathTex label placed slightly offset from endpoint in 3D (so it doesn't overlap arrows).
    Returns (label, connector_line_or_None). The label is NOT fixed-in-frame; it's placed in world coords.
    """
    endpoint = np.array(endpoint, dtype=float)
    offset = directional_label_offset(endpoint, radial=0.18, up_bias=0.06)
    pos = origin_shift + endpoint * 1.05 + offset
    lbl = MathTex(text, font_size=24 * scale)
    lbl.set_color(color)
    lbl.move_to(pos)
    conn = None
    if connector:
        start = pos
        end = origin_shift + endpoint * 0.98
        conn = Line(start, end, stroke_width=1)
        try:
            conn.set_color(WHITE)
        except Exception:
            pass
    return lbl, conn

def make_panel(title_text, items, width_pad=0.6, height_pad=0.25, title_size=30, bullet_size=20):
    """
    Create a right-side fixed-in-frame panel (bg + stacked content).
    Return (panel_group, bullets_group)
    """
    title = Text(title_text, font_size=title_size)
    bullets = VGroup(*[Text("• " + s, font_size=bullet_size) for s in items])
    content = VGroup(title, bullets).arrange(DOWN, aligned_edge=LEFT, buff=0.22)
    bg = RoundedRectangle(corner_radius=0.18, height=content.height + height_pad, width=content.width + width_pad)
    bg.set_fill(PALETTE["panel_bg"], opacity=0.80)
    bg.set_stroke(width=0)
    panel = VGroup(bg, content).arrange(DOWN, aligned_edge=LEFT, buff=0.12)
    panel.to_edge(RIGHT).shift(UP * 0.6)
    # ensure text sits on the background
    content.move_to(panel.get_center(), aligned_edge=ORIGIN)
    return panel, bullets

# ---------- Scene ----------
class DiagonalOrthogonalVisualization(ThreeDScene):
    """
    Intuitive demo showing:
      - diagonal matrices as axis-aligned scaling (uniform & anisotropic)
      - orthogonal matrices as rotations/reflections (length-preserving, orthonormal columns)
    Design goals:
      - no overlapping text (offset labels, staggered fades)
      - safe cross-version 3D vector creation
      - single fixed-in-frame panel at a time
      - interactive camera available at the end
    """
    def setup(self):
        # left area where the 3D axes + vectors appear
        self.origin_shift = LEFT * 3.0
        self.set_camera_orientation(phi=30 * DEGREES, theta=-45 * DEGREES)
        self.current_panel = None  # stores the VGroup added with add_fixed_in_frame_mobjects

    def replace_panel(self, title, bullets, run_time=0.5):
        """
        Replace the fixed-in-frame right panel safely.
        Bullets fade in staggered to avoid text stacking.
        """
        new_panel, bullets_group = make_panel(title, bullets)
        # remove old fixed-in-frame panel if any
        if self.current_panel is not None:
            try:
                self.remove_fixed_in_frame_mobjects(self.current_panel)
            except Exception:
                # try to remove members if top-level removal fails
                for sub in self.current_panel:
                    try:
                        self.remove_fixed_in_frame_mobjects(sub)
                    except Exception:
                        pass
        # add new panel
        self.add_fixed_in_frame_mobjects(new_panel)
        # animate replacement
        if self.current_panel is not None:
            self.play(FadeOut(self.current_panel), FadeIn(new_panel), run_time=run_time)
        else:
            self.play(FadeIn(new_panel), run_time=run_time)
        # stagger bullets fade-in for clarity
        try:
            self.play(LaggedStart(*[FadeIn(b, shift=UP * 0.12) for b in bullets_group], lag_ratio=0.16), run_time=0.5)
        except Exception:
            for b in bullets_group:
                self.play(FadeIn(b), run_time=0.12)
        self.current_panel = new_panel

    def construct(self):
        # --- Intro panel ---
        self.replace_panel("Diagonal & Orthogonal", ["Diagonal → axis scaling", "Orthogonal → rotations/reflections"])
        self.wait(0.5)

        # --- Phase A: Diagonal matrices ---
        group_a = self.phase_diagonal()
        self.wait(0.4)
        self.play(FadeOut(group_a), run_time=0.6)
        self.wait(0.25)

        # --- Phase B: Orthogonal matrices ---
        group_b, extra_dot_panel = self.phase_orthogonal()
        self.wait(0.3)

        # remove the small extra panel (dot product) before fading out group_b
        if extra_dot_panel is not None:
            try:
                self.remove_fixed_in_frame_mobjects(extra_dot_panel)
            except Exception:
                try:
                    self.remove(extra_dot_panel)
                except Exception:
                    pass
        self.play(FadeOut(group_b), run_time=0.6)
        self.wait(0.25)

        # --- Summary panel ---
        # escape &, use plain Text bullets so no markup parsing errors
        self.replace_panel("Summary", ["Diagonal: axis-aligned scaling", "Orthogonal: preserves lengths & angles"])
        self.wait(0.5)

        # --- Interactive finale ---
        try:
            # allows mouse rotation/zoom if supported
            self.interactive_embed()
        except Exception:
            # if not available, end normally
            self.wait(0.8)

    # ---------------- Phase A ----------------
    def phase_diagonal(self):
        # axes and labels
        axes = ThreeDAxes(x_range=[-4, 4, 1], y_range=[-4, 4, 1], z_range=[-3, 3, 1],
                          x_length=5, y_length=5, z_length=5).shift(self.origin_shift)
        x_lbl = axes.get_x_axis_label(MathTex("x")).shift(self.origin_shift + 0.3 * RIGHT)
        y_lbl = axes.get_y_axis_label(MathTex("y")).shift(self.origin_shift + 0.3 * UP)
        z_lbl = axes.get_z_axis_label(MathTex("z")).shift(self.origin_shift + 0.3 * OUT)

        # basis vectors (different lengths for visual variety)
        e1 = np.array([1.5, 0.0, 0.0])
        e2 = np.array([0.0, 1.2, 0.0])
        e3 = np.array([0.0, 0.0, 1.0])

        v1 = safe_three_d_vector(e1, color=PALETTE["e1"], stroke_width=3)
        v2 = safe_three_d_vector(e2, color=PALETTE["e2"], stroke_width=3)
        v3 = safe_three_d_vector(e3, color=PALETTE["e3"], stroke_width=3)

        v1_lbl, v1_conn = make_tip_label(e1, "e_1", color=PALETTE["e1"], scale=LABEL_MED, origin_shift=self.origin_shift)
        v2_lbl, v2_conn = make_tip_label(e2, "e_2", color=PALETTE["e2"], scale=LABEL_MED, origin_shift=self.origin_shift)
        v3_lbl, v3_conn = make_tip_label(e3, "e_3", color=PALETTE["e3"], scale=LABEL_MED, origin_shift=self.origin_shift)

        # show base axes + vectors (staggered)
        self.add(axes, x_lbl, y_lbl, z_lbl)
        self.play(LaggedStart(FadeIn(v1), FadeIn(v2), FadeIn(v3),
                              FadeIn(v1_lbl), FadeIn(v2_lbl), FadeIn(v3_lbl),
                              lag_ratio=0.12),
                  run_time=0.9)
        for c in (v1_conn, v2_conn, v3_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.18)
        self.wait(0.45)

        # Uniform scaling example
        self.replace_panel("Diagonal — Uniform scaling", ["D = diag(2,2,2) → uniform scale ×2"])
        D = np.diag([2.0, 2.0, 2.0])
        v1_u = safe_three_d_vector(D @ e1, color=PALETTE["e1"], stroke_width=2)
        v2_u = safe_three_d_vector(D @ e2, color=PALETTE["e2"], stroke_width=2)
        v3_u = safe_three_d_vector(D @ e3, color=PALETTE["e3"], stroke_width=2)
        v1u_lbl, v1u_conn = make_tip_label(D @ e1, "D e_1", color=PALETTE["e1"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        v2u_lbl, v2u_conn = make_tip_label(D @ e2, "D e_2", color=PALETTE["e2"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        v3u_lbl, v3u_conn = make_tip_label(D @ e3, "D e_3", color=PALETTE["e3"], scale=LABEL_SMALL, origin_shift=self.origin_shift)

        # transform vectors; fade out original labels first (prevents overlap)
        self.play(Transform(v1, v1_u), Transform(v2, v2_u), Transform(v3, v3_u),
                  FadeOut(v1_lbl), FadeOut(v2_lbl), FadeOut(v3_lbl), run_time=1.1)
        for lbl in (v1u_lbl, v2u_lbl, v3u_lbl):
            self.play(FadeIn(lbl), run_time=0.22)
        for c in (v1u_conn, v2u_conn, v3u_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.14)
        self.wait(0.6)

        # show numeric demonstration briefly then clear
        before = np.linalg.norm(e1)
        after = np.linalg.norm(D @ e1)
        bnum = DecimalNumber(before, num_decimal_places=2).next_to(self.origin_shift + e1, DOWN + RIGHT * 0.15)
        anum = DecimalNumber(after, num_decimal_places=2).next_to(self.origin_shift + D @ e1, DOWN + RIGHT * 0.15)
        tempnums = VGroup(bnum, anum)
        self.play(FadeIn(tempnums), run_time=0.45)
        self.wait(0.6)
        for m in tempnums:
            try:
                m.clear_updaters()
            except Exception:
                pass
        self.play(FadeOut(tempnums), run_time=0.35)
        try:
            self.remove(*tempnums)
        except Exception:
            pass

        # Anisotropic scaling example
        self.replace_panel("Diagonal — Anisotropic scaling", ["D = diag(1.2, 0.5, 2.0) → axis-specific"])
        D2 = np.diag([1.2, 0.5, 2.0])
        v1_a = safe_three_d_vector(D2 @ e1, color=PALETTE["e1"], stroke_width=2)
        v2_a = safe_three_d_vector(D2 @ e2, color=PALETTE["e2"], stroke_width=2)
        v3_a = safe_three_d_vector(D2 @ e3, color=PALETTE["e3"], stroke_width=2)
        v1a_lbl, v1a_conn = make_tip_label(D2 @ e1, "D e_1", color=PALETTE["e1"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        v2a_lbl, v2a_conn = make_tip_label(D2 @ e2, "D e_2", color=PALETTE["e2"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        v3a_lbl, v3a_conn = make_tip_label(D2 @ e3, "D e_3", color=PALETTE["e3"], scale=LABEL_SMALL, origin_shift=self.origin_shift)

        self.play(Transform(v1, v1_a), Transform(v2, v2_a), Transform(v3, v3_a),
                  FadeOut(v1u_lbl), FadeOut(v2u_lbl), FadeOut(v3u_lbl), run_time=1.0)
        for lbl in (v1a_lbl, v2a_lbl, v3a_lbl):
            self.play(FadeIn(lbl), run_time=0.22)
        for c in (v1a_conn, v2a_conn, v3a_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.14)

        # show 3 small annotations (stretch/squash) placed away from other labels
        arrow_shrink = Arrow(self.origin_shift + np.array([0.5, 1.0, 0]), self.origin_shift + np.array([0.5, 0.7, 0]), buff=0)
        shrink_lbl = Text("squash", font_size=18).next_to(arrow_shrink, RIGHT)
        arrow_stretch = Arrow(self.origin_shift + np.array([0.15, 0.15, 1.0]), self.origin_shift + np.array([0.15, 0.15, 1.9]), buff=0)
        stretch_lbl = Text("stretch", font_size=18).next_to(arrow_stretch, RIGHT)
        self.play(GrowArrow(arrow_shrink), FadeIn(shrink_lbl), GrowArrow(arrow_stretch), FadeIn(stretch_lbl))
        self.wait(0.7)

        # Return the group for later fade-out
        group = VGroup(axes, x_lbl, y_lbl, z_lbl,
                       v1, v2, v3,
                       v1_a, v2_a, v3_a,
                       v1a_lbl, v2a_lbl, v3a_lbl,
                       arrow_shrink, shrink_lbl, arrow_stretch, stretch_lbl)
        return group

    # ---------------- Phase B ----------------
    def phase_orthogonal(self):
        # update panel
        self.replace_panel("Orthogonal — Rotation / Reflection", ["Preserves lengths & inner products", "Columns are orthonormal"])

        axes = ThreeDAxes(x_range=[-3.5, 3.5, 1], y_range=[-3.5, 3.5, 1], z_range=[-3.5, 3.5, 1],
                          x_length=4.5, y_length=4.5, z_length=4.5).shift(self.origin_shift)
        x_lbl = axes.get_x_axis_label(MathTex("x")).shift(self.origin_shift + 0.25 * RIGHT)
        y_lbl = axes.get_y_axis_label(MathTex("y")).shift(self.origin_shift + 0.25 * UP)
        z_lbl = axes.get_z_axis_label(MathTex("z")).shift(self.origin_shift + 0.25 * OUT)

        # basis vectors
        e1 = np.array([1.5, 0.0, 0.0])
        e2 = np.array([0.0, 1.2, 0.0])
        e3 = np.array([0.0, 0.0, 1.0])

        vb1 = safe_three_d_vector(e1, color=PALETTE["e1"], stroke_width=3)
        vb2 = safe_three_d_vector(e2, color=PALETTE["e2"], stroke_width=3)
        vb3 = safe_three_d_vector(e3, color=PALETTE["e3"], stroke_width=3)
        vb1_lbl, vb1_conn = make_tip_label(e1, "v_1", color=PALETTE["e1"], scale=LABEL_MED, origin_shift=self.origin_shift)
        vb2_lbl, vb2_conn = make_tip_label(e2, "v_2", color=PALETTE["e2"], scale=LABEL_MED, origin_shift=self.origin_shift)
        vb3_lbl, vb3_conn = make_tip_label(e3, "v_3", color=PALETTE["e3"], scale=LABEL_MED, origin_shift=self.origin_shift)

        self.play(FadeIn(axes), FadeIn(x_lbl), FadeIn(y_lbl), FadeIn(z_lbl), run_time=0.6)
        self.play(LaggedStart(FadeIn(vb1), FadeIn(vb2), FadeIn(vb3),
                              FadeIn(vb1_lbl), FadeIn(vb2_lbl), FadeIn(vb3_lbl),
                              lag_ratio=0.12), run_time=0.9)
        for c in (vb1_conn, vb2_conn, vb3_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.15)
        self.wait(0.45)

        # Rotation about Z
        theta = 50 * DEGREES
        Rz = np.array([[np.cos(theta), -np.sin(theta), 0.0],
                       [np.sin(theta),  np.cos(theta), 0.0],
                       [0.0,            0.0,           1.0]])
        vb1_end = Rz @ e1
        vb2_end = Rz @ e2
        vb3_end = Rz @ e3

        vb1_rot = safe_three_d_vector(vb1_end, color=PALETTE["e1"], stroke_width=2)
        vb2_rot = safe_three_d_vector(vb2_end, color=PALETTE["e2"], stroke_width=2)
        vb3_rot = safe_three_d_vector(vb3_end, color=PALETTE["e3"], stroke_width=2)
        vb1r_lbl, vb1r_conn = make_tip_label(vb1_end, "R v_1", color=PALETTE["e1"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        vb2r_lbl, vb2r_conn = make_tip_label(vb2_end, "R v_2", color=PALETTE["e2"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        vb3r_lbl, vb3r_conn = make_tip_label(vb3_end, "R v_3", color=PALETTE["e3"], scale=LABEL_SMALL, origin_shift=self.origin_shift)

        # animate rotation; fade original labels to prevent overlap
        self.play(Transform(vb1, vb1_rot), Transform(vb2, vb2_rot), Transform(vb3, vb3_rot),
                  FadeOut(vb1_lbl), FadeOut(vb2_lbl), FadeOut(vb3_lbl), run_time=1.2)
        for lbl in (vb1r_lbl, vb2r_lbl, vb3r_lbl):
            self.play(FadeIn(lbl), run_time=0.22)
        for c in (vb1r_conn, vb2r_conn, vb3r_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.12)
        self.wait(0.45)

        # Show length preservation (brief)
        bnum = DecimalNumber(np.linalg.norm(e1), num_decimal_places=3).next_to(self.origin_shift + e1, DOWN + RIGHT * 0.12)
        anum = DecimalNumber(np.linalg.norm(vb1_end), num_decimal_places=3).next_to(self.origin_shift + vb1_end, DOWN + RIGHT * 0.12)
        self.play(FadeIn(bnum), FadeIn(anum))
        eq_line = Line(bnum.get_center(), anum.get_center())
        eq_text = Text("equal", font_size=18).next_to(eq_line, UP)
        self.play(Create(eq_line), FadeIn(eq_text))
        self.wait(0.7)
        for m in (bnum, anum, eq_line, eq_text):
            try:
                m.clear_updaters()
            except Exception:
                pass
        self.play(FadeOut(bnum), FadeOut(anum), FadeOut(eq_line), FadeOut(eq_text), run_time=0.35)
        try:
            self.remove(bnum, anum, eq_line, eq_text)
        except Exception:
            pass

        # Columns of rotation matrix = orthonormal vectors
        c1 = Rz[:, 0]
        c2 = Rz[:, 1]
        c3 = Rz[:, 2]
        col1 = safe_three_d_vector(c1 * 1.5, color=PALETTE["col1"], stroke_width=2)
        col2 = safe_three_d_vector(c2 * 1.5, color=PALETTE["col2"], stroke_width=2)
        col3 = safe_three_d_vector(c3 * 1.5, color=PALETTE["col3"], stroke_width=2)
        c1_lbl, c1_conn = make_tip_label(c1 * 1.5, "c_1", color=PALETTE["col1"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        c2_lbl, c2_conn = make_tip_label(c2 * 1.5, "c_2", color=PALETTE["col2"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        c3_lbl, c3_conn = make_tip_label(c3 * 1.5, "c_3", color=PALETTE["col3"], scale=LABEL_SMALL, origin_shift=self.origin_shift)

        self.play(FadeIn(col1), FadeIn(col2), FadeIn(col3), run_time=0.9)
        for lbl in (c1_lbl, c2_lbl, c3_lbl):
            self.play(FadeIn(lbl), run_time=0.22)
        for c in (c1_conn, c2_conn, c3_conn):
            if c is not None:
                self.play(GrowFromCenter(c), run_time=0.14)
        self.wait(0.45)

        # Show dot product in a small fixed panel (so it never overlaps world labels)
        dot_val = float(np.dot(c1, c2))
        dot_panel = Text(f"⟨c₁, c₂⟩ = {dot_val:.3f}", font_size=20)
        dot_panel.to_edge(RIGHT).shift(DOWN * 1.4)
        self.add_fixed_in_frame_mobjects(dot_panel)
        self.play(FadeIn(dot_panel))
        self.wait(0.8)

        # Angle arc between c1 and c2 (visual)
        c1n = c1 / np.linalg.norm(c1)
        c2n = c2 / np.linalg.norm(c2)
        angle = float(np.arccos(np.clip(np.dot(c1n, c2n), -1, 1)))
        R = 0.5
        def arc_fn(t):
            return self.origin_shift + R * (np.cos(t) * c1n + np.sin(t) * c2n)
        arc = ParametricFunction(arc_fn, t_range=[0, angle], stroke_width=3)
        ang_label = MathTex(f"{(angle/DEGREES):.0f}^\\circ", font_size=20).move_to(
            self.origin_shift + 0.9 * (np.cos(angle/2) * c1n + np.sin(angle/2) * c2n)
        )
        self.play(Create(arc), FadeIn(ang_label))
        self.wait(0.6)

        # Reflection demo (flip y coordinate of c1)
        reflect = np.diag([1.0, -1.0, 1.0])
        ref_end = reflect @ (c1 * 1.5)
        ref_vec = safe_three_d_vector(ref_end, color=PALETTE["col1"], stroke_width=2)
        ref_lbl, ref_conn = make_tip_label(ref_end, "Ref(c_1)", color=PALETTE["col1"], scale=LABEL_SMALL, origin_shift=self.origin_shift)
        self.play(Transform(col1, ref_vec), FadeOut(c1_lbl), run_time=0.8)
        if ref_conn is not None:
            self.play(GrowFromCenter(ref_conn), run_time=0.12)
        self.play(FadeIn(ref_lbl))
        self.wait(0.5)

        # return world objects and the extra panel object so caller can remove it
        world_objs = VGroup(axes, x_lbl, y_lbl, z_lbl,
                            vb1, vb2, vb3,
                            vb1r_lbl, vb2r_lbl, vb3r_lbl,
                            col1, col2, col3,
                            c1_lbl, c2_lbl, c3_lbl,
                            arc, ang_label,
                            ref_vec, ref_lbl)
        return world_objs, dot_panel

# end of file


%manim -ql -v warning DiagonalOrthogonalVisualization
```

```python
from manim import *
import numpy as np

# ================================================================
# 3D TRANSFORMATION EXPLAINER (Diagonal + Orthogonal) - Fixed
# Compatible with Manim Community v0.19.0
# ================================================================
class DiagonalOrthogonalExplainer(ThreeDScene):

    def setup(self):
        # Keep persistent holders for fixed-in-frame panels so we can safely
        # remove/re-add them without errors.
        self.explanation_panel = VGroup()
        self.matrix_display = VGroup()

    # Helper: create a right-side fixed explanation panel (safe width)
    def _update_explanation_panel(self, title, bullets):
        # Remove previous so we don't stack
        try:
            self.remove_fixed_in_frame_mobjects(self.explanation_panel)
        except Exception:
            pass

        title_text = Text(title, font_size=30, color=YELLOW)
        title_bg = SurroundingRectangle(title_text, buff=0.18, color=YELLOW)
        header = VGroup(title_text, title_bg)

        # Build bullets ensuring they wrap visually by limiting font size/width
        bullet_lines = VGroup(*[
            Text(f"• {b}", font_size=20, color=WHITE) for b in bullets
        ])
        bullet_lines.arrange(DOWN, aligned_edge=LEFT, buff=0.25)

        panel = VGroup(header, bullet_lines)
        panel.arrange(DOWN, aligned_edge=LEFT, buff=0.35)

        # Anchor panel to the top-right of the screen and set explicit width
        panel.to_corner(UR)
        panel.shift(LEFT * 0.8 + DOWN * 0.5)
        panel.set_width(5.2)

        self.explanation_panel = panel
        self.add_fixed_in_frame_mobjects(self.explanation_panel)

    # Helper: show matrix title + matrix below explanation, fixed in frame
    def _display_matrix(self, matrix_np, title="Transformation Matrix"):
        try:
            self.remove_fixed_in_frame_mobjects(self.matrix_display)
        except Exception:
            pass

        matrix_m = Matrix(matrix_np, h_buff=1.0, v_buff=0.6)
        matrix_m.scale(0.65)
        matrix_title = Text(title, font_size=22, color=WHITE)

        self.matrix_display = VGroup(matrix_title, matrix_m)
        self.matrix_display.arrange(DOWN, buff=0.35)

        # Place below the explanation panel and align to the right side
        self.matrix_display.to_corner(UR)
        # Nudges to position it below the explanation and inside the screen
        self.matrix_display.shift(LEFT * 0.8 + DOWN * 1.6)
        self.matrix_display.set_width(4.6)

        self.add_fixed_in_frame_mobjects(self.matrix_display)

    # Utility: create 3D arrow basis with labels that stay near arrow tips in 3D
    def _make_basis(self, shift_vec=ORIGIN):
        e1 = Arrow3D(start=ORIGIN, end=np.array([1, 0, 0]), color=RED, thickness=0.05, height=0.18, base_radius=0.07)
        e2 = Arrow3D(start=ORIGIN, end=np.array([0, 1, 0]), color=GREEN, thickness=0.05, height=0.18, base_radius=0.07)
        e3 = Arrow3D(start=ORIGIN, end=np.array([0, 0, 1]), color=BLUE, thickness=0.05, height=0.18, base_radius=0.07)

        # Text labels for endpoints — keep as 3D Text so they transform with the scene.
        x_label = Text("X", font_size=20, color=RED).next_to(e1.get_end(), RIGHT*0.6 + UP*0.15)
        y_label = Text("Y", font_size=20, color=GREEN).next_to(e2.get_end(), UP*0.6 + RIGHT*0.15)
        z_label = Text("Z", font_size=20, color=BLUE).next_to(e3.get_end(), OUT*0.6 + UP*0.05)

        basis = VGroup(e1, e2, e3, x_label, y_label, z_label)
        basis.shift(shift_vec)
        return basis

    def construct(self):
        # --- Layout constants ---
        axes_shift = LEFT * 4.0  # move 3D scene to the left to make room

        # --- Camera orientation chosen to avoid panel overlap ---
        # phi controls elevation; theta controls azimuthal rotation.
        # These values give a pleasant isometric-ish view and avoid projecting
        # the right-side fixed panel onto axes.
        self.set_camera_orientation(phi=65 * DEGREES, theta=45 * DEGREES, zoom=1)

        # ---------- PHASE 1: Diagonal matrix ----------
        self._update_explanation_panel(
            "Diagonal Matrix",
            [
                "Equal diagonal entries → uniform scaling",
                "Different diagonal entries → anisotropic scaling",
                "Each axis scales independently",
                "Lengths change but angles preserved (no shear)"
            ]
        )
        self._display_matrix(np.identity(3), "Identity Matrix")

        axes = ThreeDAxes(
            x_range=[-3, 3, 1], y_range=[-3, 3, 1], z_range=[-3, 3, 1],
            x_length=5, y_length=5, z_length=5,
            axis_config={"include_numbers": True, "font_size": 18}
        )
        axes.shift(axes_shift)

        basis = self._make_basis(axes_shift)

        # Cube for scaling demo — placed at the origin of the 3D axes
        cube = Cube(side_length=1.4, color=YELLOW, fill_opacity=0.6, stroke_opacity=1)
        cube.shift(axes_shift)

        # Entrance
        self.play(FadeIn(axes), FadeIn(basis), Create(cube), run_time=1.6)
        self.wait(0.4)

        # Uniform scaling (equal diagonal entries)
        D_equal = 1.8 * np.identity(3)
        self._display_matrix(D_equal, "Uniform Scaling Matrix")

        def animate_uniform(mob, alpha):
            M = interpolate(np.identity(3), D_equal, alpha)
            new_cube = Cube(side_length=1.4).apply_matrix(M)
            new_cube.shift(axes_shift)
            mob.become(new_cube)

        self.play(UpdateFromAlphaFunc(cube, animate_uniform), run_time=2)
        self.wait(0.6)

        # Unequal diagonal scaling (anisotropic)
        D_unequal = np.diag([2.4, 1.0, 0.5])
        self._display_matrix(D_unequal, "Anisotropic Scaling Matrix")

        def animate_unequal(mob, alpha):
            M = interpolate(D_equal, D_unequal, alpha)
            new_cube = Cube(side_length=1.4).apply_matrix(M)
            new_cube.shift(axes_shift)
            mob.become(new_cube)

        self.play(UpdateFromAlphaFunc(cube, animate_unequal), run_time=2.2)
        self.wait(0.6)

        # Show scaled axes (small translucent axes attached to cube to emphasize axis-wise scaling)
        scaled_e1 = Arrow3D(ORIGIN, D_unequal @ np.array([1, 0, 0]), color=RED, thickness=0.04, height=0.14, base_radius=0.06)
        scaled_e2 = Arrow3D(ORIGIN, D_unequal @ np.array([0, 1, 0]), color=GREEN, thickness=0.04, height=0.14, base_radius=0.06)
        scaled_e3 = Arrow3D(ORIGIN, D_unequal @ np.array([0, 0, 1]), color=BLUE, thickness=0.04, height=0.14, base_radius=0.06)
        scaled_basis = VGroup(scaled_e1, scaled_e2, scaled_e3)
        scaled_basis.shift(axes_shift)

        self.play(FadeIn(scaled_basis), run_time=0.9)
        self.wait(0.6)

        # Fade out diagonal phase (panels are fixed-in-frame so remove them separately)
        self.play(FadeOut(scaled_basis), FadeOut(cube), FadeOut(basis), FadeOut(axes))
        # Remove fixed-in-frame panels for next phase
        self.play(FadeOut(self.matrix_display), FadeOut(self.explanation_panel))
        self.wait(0.4)

        # ---------- PHASE 2: Orthogonal matrix (rotation + length preservation) ----------
        self._update_explanation_panel(
            "Orthogonal Matrix",
            [
                "Represents pure rotation or reflection",
                "Preserves lengths and angles",
                "Orthonormal columns → 90° between columns",
                "Each column has length 1"
            ]
        )
        self._display_matrix(np.identity(3), "Initial Orthogonal Matrix")

        axes2 = ThreeDAxes(
            x_range=[-3, 3, 1], y_range=[-3, 3, 1], z_range=[-3, 3, 1],
            x_length=5, y_length=5, z_length=5,
            axis_config={"include_numbers": True, "font_size": 18}
        )
        axes2.shift(axes_shift)

        basis2 = self._make_basis(axes_shift)

        self.play(FadeIn(axes2), FadeIn(basis2), run_time=1.2)
        self.wait(0.4)

        # Demonstrate rotations around Z axis and show matrix updates
        angles = [45, 60]
        for deg in angles:
            rad = deg * DEGREES
            R = np.array([
                [np.cos(rad), -np.sin(rad), 0],
                [np.sin(rad),  np.cos(rad), 0],
                [0, 0, 1]
            ])
            self._display_matrix(R, f"Rotation Matrix ({deg}°)")

            # Rotated basis (in world coords) — apply rotation to unit vectors
            rotated_e1 = Arrow3D(ORIGIN, R @ np.array([1, 0, 0]), color=RED, thickness=0.05, height=0.18, base_radius=0.07)
            rotated_e2 = Arrow3D(ORIGIN, R @ np.array([0, 1, 0]), color=GREEN, thickness=0.05, height=0.18, base_radius=0.07)
            rotated_e3 = Arrow3D(ORIGIN, R @ np.array([0, 0, 1]), color=BLUE, thickness=0.05, height=0.18, base_radius=0.07)

            rotated_x_label = Text("X", font_size=20, color=RED).next_to(rotated_e1.get_end(), RIGHT*0.6 + UP*0.15)
            rotated_y_label = Text("Y", font_size=20, color=GREEN).next_to(rotated_e2.get_end(), UP*0.6 + RIGHT*0.15)
            rotated_z_label = Text("Z", font_size=20, color=BLUE).next_to(rotated_e3.get_end(), OUT*0.6 + UP*0.05)

            rotated_basis = VGroup(rotated_e1, rotated_e2, rotated_e3, rotated_x_label, rotated_y_label, rotated_z_label)
            rotated_basis.shift(axes_shift)

            # Transform the basis2 group into rotated_basis
            self.play(Transform(basis2, rotated_basis), run_time=1.8)
            self.wait(0.4)

            # Draw angle between first two rotated axes — use projection onto XY plane
            # Project endpoints to the XY plane to compute the angle cleanly
            p1 = rotated_e1.get_end()[:2]
            p2 = rotated_e2.get_end()[:2]
            # Build 2D lines from origin to endpoints in plane then lift into 3D for Angle
            line1 = Line(ORIGIN, np.array([p1[0], p1[1], 0]), stroke_width=2)
            line2 = Line(ORIGIN, np.array([p2[0], p2[1], 0]), stroke_width=2)
            # Position these temporary lines in the scene near the axes_shift so the Angle will compute visually
            line1.shift(axes_shift)
            line2.shift(axes_shift)

            angle_arc = Angle(line1, line2, radius=0.8, color=YELLOW)
            angle_label = MathTex(f"{deg}^\\circ", color=YELLOW).scale(0.7)
            angle_label.next_to(angle_arc, UP + RIGHT, buff=0.05)

            self.play(Create(angle_arc), FadeIn(angle_label))
            self.wait(0.6)

            # Remove angle and its helpers
            self.play(FadeOut(angle_arc), FadeOut(angle_label), FadeOut(line1), FadeOut(line2))
            self.wait(0.25)

        # Show length preservation: measure the length of one basis vector before and after
        # We'll create two short lines: one at original (1) and one transformed (should also be 1)
        original_end = np.array([1, 0, 0]) + axes_shift
        rotated_end = basis2[0].get_end()

        original_line = Line(ORIGIN + axes_shift, original_end, stroke_width=2)
        rotated_line = Line(ORIGIN + axes_shift, rotated_end, stroke_width=2)

        len_text_1 = Text("Length = 1", font_size=18).next_to(original_end, UP + LEFT, buff=0.1)
        len_text_2 = Text("Length = 1", font_size=18).next_to(rotated_end, UP + LEFT, buff=0.1)

        self.play(Create(original_line), Create(rotated_line), FadeIn(len_text_1), FadeIn(len_text_2))
        self.wait(1.0)

        # Clean up final scene
        self.play(FadeOut(basis2), FadeOut(axes2), FadeOut(original_line), FadeOut(rotated_line), FadeOut(len_text_1), FadeOut(len_text_2))
        # Remove fixed panels
        self.play(FadeOut(self.matrix_display), FadeOut(self.explanation_panel))
        self.wait(0.6)

# Run command (example):
%manim -ql -v warning DiagonalOrthogonalExplainer
```

```python
from manim import *

class OrthogonalDiagonalExplanation(Scene):

    def construct(self):

        # --------------------------------------------------------
        # 1) BASE PLANE + AXES
        # --------------------------------------------------------
        plane = NumberPlane(
            x_range=[-4, 4],
            y_range=[-4, 4],
            background_line_style={"stroke_width": 1, "stroke_opacity": 0.8}
        ).scale(1.1)

        x_label = MathTex("x").next_to(plane.x_axis.get_end(), RIGHT)
        y_label = MathTex("y").next_to(plane.y_axis.get_end(), UP)
        axes_labels = VGroup(x_label, y_label)

        self.play(Create(plane), Write(axes_labels))
        self.wait(1)

        # Base vectors
        v1 = Arrow(start=ORIGIN, end=[2, 1, 0], buff=0, color=YELLOW)
        v2 = Arrow(start=ORIGIN, end=[1, 2, 0], buff=0, color=GREEN)
        v1_label = MathTex("v_1", color=YELLOW).next_to(v1.get_end(), RIGHT)
        v2_label = MathTex("v_2", color=GREEN).next_to(v2.get_end(), RIGHT)

        self.play(GrowArrow(v1), Write(v1_label))
        self.play(GrowArrow(v2), Write(v2_label))
        self.wait(1)

        # --------------------------------------------------------
        # 2) ROTATION MATRIX (Orthogonal)
        # --------------------------------------------------------
        theta = 45 * DEGREES
        rot_matrix_tex = MathTex(
            R"R(\theta)=\begin{bmatrix}\cos\theta & -\sin\theta \\ \sin\theta & \cos\theta\end{bmatrix}",
            R"\quad \theta = 45^\circ"
        ).to_edge(UP).set_color(BLUE)

        ortho_note = MathTex(
            R"\text{Orthogonal Matrix: } R^TR = I",
            R"\text{Preserves lengths and angles}"
        ).scale(0.8).next_to(rot_matrix_tex, DOWN).set_color(BLUE_E)

        self.play(Write(rot_matrix_tex))
        self.play(FadeIn(ortho_note))
        self.wait(1)

        # Rotation animation
        self.play(
            Rotate(v1, angle=theta, about_point=ORIGIN),
            Rotate(v2, angle=theta, about_point=ORIGIN),
            run_time=2
        )
        self.wait(1)

        # Labels follow rotation
        v1_label_rot = MathTex("R v_1", color=YELLOW).next_to(v1.get_end(), RIGHT)
        v2_label_rot = MathTex("R v_2", color=GREEN).next_to(v2.get_end(), RIGHT)

        self.play(Transform(v1_label, v1_label_rot))
        self.play(Transform(v2_label, v2_label_rot))
        self.wait(1)

        # --------------------------------------------------------
        # 3) CLEAR ROTATION TEXT
        # --------------------------------------------------------
        self.play(FadeOut(rot_matrix_tex), FadeOut(ortho_note))
        self.wait(0.5)

        # --------------------------------------------------------
        # 4) DIAGONAL MATRIX (Scaling)
        # --------------------------------------------------------
        scale_matrix_tex = MathTex(
            R"D=\begin{bmatrix}2 & 0 \\ 0 & 0.5\end{bmatrix}"
        ).to_edge(UP).set_color(RED)

        diag_note = MathTex(
            R"\text{Diagonal Matrix: Independent scaling on axes}",
            R"\text{Changes lengths and angles}"
        ).scale(0.8).next_to(scale_matrix_tex, DOWN).set_color(RED_E)

        self.play(Write(scale_matrix_tex))
        self.play(FadeIn(diag_note))
        self.wait(1)

        # Apply scaling
        scale_transform = [[2, 0], [0, 0.5]]

        v1_scaled_end = [2*2, 1*0.5, 0]     # (x*2, y*0.5)
        v2_scaled_end = [1*2, 2*0.5, 0]

        v1_scaled = Arrow(ORIGIN, v1_scaled_end, buff=0, color=YELLOW)
        v2_scaled = Arrow(ORIGIN, v2_scaled_end, buff=0, color=GREEN)

        self.play(
            Transform(v1, v1_scaled),
            Transform(v2, v2_scaled),
            run_time=2
        )

        v1_label_scaled = MathTex("D v_1", color=YELLOW).next_to(v1_scaled_end, RIGHT)
        v2_label_scaled = MathTex("D v_2", color=GREEN).next_to(v2_scaled_end, RIGHT)

        self.play(Transform(v1_label, v1_label_scaled))
        self.play(Transform(v2_label, v2_label_scaled))
        self.wait(1)

        # --------------------------------------------------------
        # Outro
        # --------------------------------------------------------
        final_msg = MathTex(
            R"\text{Orthogonal: preserves geometry}", 
            R"\text{Diagonal: scales geometry}"
        ).scale(0.9).set_color(WHITE).to_edge(DOWN)

        self.play(FadeIn(final_msg))
        self.wait(2)



%manim -ql -v warning OrthogonalVsDiagonal
```
