# Orthogonal Matrices: The Geometry of Rigid Motion

## 1. What is an Orthogonal Matrix?
An **orthogonal matrix** ($Q$) is a square matrix whose columns and rows are **orthonormal vectors** (unit length and perpendicular to each other).

In linear algebra terms, a real square matrix $Q$ is orthogonal if its transpose is equal to its inverse:

$$Q^T = Q^{-1}$$

This implies the fundamental identity:

$$Q^T Q = I$$

### The Geometric Intuition
Geometrically, an orthogonal matrix represents a **Rigid Transformation** (Isometry) centered at the origin.
* It rotates or reflects space.
* It **never** stretches, squishes, or shears the grid.
* The origin $(0,0)$ remains fixed.

---

## 2. Key Properties
The Manim animation demonstrates three specific properties that every orthogonal matrix must satisfy:

### A. Preservation of Length (Isometry)
If you take a vector $\vec{v}$ and multiply it by an orthogonal matrix $Q$, the length (norm) of the resulting vector remains unchanged.

$$||Q\vec{v}|| = ||\vec{v}||$$

> **Visual Proof:** In the animation, the yellow brace measuring $\hat{i}$ always stays at length $1$.

### B. Preservation of Angles (Orthogonality)
If two vectors are perpendicular (at $90^\circ$) before the transformation, they remain perpendicular after. The dot product is preserved.

$$(\vec{u} \cdot \vec{v} = 0) \implies (Q\vec{u} \cdot Q\vec{v} = 0)$$

> **Visual Proof:** In the animation, the vectors $\hat{i}$ and $\hat{j}$ always maintain a right angle.

### C. Determinant is $\pm 1$
Because the matrix does not scale the area of the unit square, the magnitude of the determinant is always 1.

* **$\det(Q) = +1$:** A pure **Rotation**. Orientation is preserved.
* **$\det(Q) = -1$:** A **Reflection** (improper rotation). The space is "flipped" like a mirror.

## 3. Numerical Examples (From the Animation)

Below are the 5 specific matrices visualized in the animation.

### Case 1: Standard Rotation ($60^\circ$)
Rotates the grid counter-clockwise.
$$Q = \begin{bmatrix} 0.5 & -0.866 \\ 0.866 & 0.5 \end{bmatrix}$$
* **Type:** Rotation
* **Determinant:** $+1$

### Case 2: Reflection (Across X-Axis)
Keeps $x$ constant, flips $y$.
$$Q = \begin{bmatrix} 1 & 0 \\ 0 & -1 \end{bmatrix}$$
* **Type:** Reflection
* **Determinant:** $-1$

### Case 3: Permutation Matrix
Swaps the $x$ and $y$ axes. Equivalent to reflecting across the line $y=x$.
$$Q = \begin{bmatrix} 0 & 1 \\ 1 & 0 \end{bmatrix}$$
* **Type:** Axis Swap
* **Determinant:** $-1$

### Case 4: Pythagorean Rotation
A rotation based on the 3-4-5 right triangle ($\frac{3}{5}=0.6, \frac{4}{5}=0.8$). This provides a clean rotation without irrational numbers.
$$Q = \begin{bmatrix} 0.6 & -0.8 \\ 0.8 & 0.6 \end{bmatrix}$$
* **Type:** Rotation ($\approx 53.1^\circ$)
* **Determinant:** $+1$

### Case 5: $180^\circ$ Inversion
Rotates halfway around the circle. Negates both axes.
$$Q = \begin{bmatrix} -1 & 0 \\ 0 & -1 \end{bmatrix}$$
* **Type:** Point Reflection
* **Determinant:** $+1$ (Double reflection cancels out the negative sign)

## 4. The Inverse Transformation

Usually, calculating a matrix inverse is computationally expensive. However, for orthogonal matrices, the inverse is simply the transpose.

$$\text{Inverse} = \text{Transpose}$$

To undo the transformation, you swap the rows and columns. For example, to undo **Case 1**:

$$Q^{-1} = Q^T = \begin{bmatrix} 0.5 & 0.866 \\ -0.866 & 0.5 \end{bmatrix}$$
