# Summary Notes: 2D Rotational Matrices

### 1. Definition
A **Rotation Matrix** is a linear transformation that turns a vector (or grid) counter-clockwise by an angle $\theta$ around the origin $(0,0)$. 
* **Positive $\theta$:** Counter-Clockwise
* **Negative $\theta$:** Clockwise

---

### 2. The "Secret" Rule (Basis Vectors)
To construct any 2D rotation matrix, you only need to track the destination of the basis vectors:
* **Column 1:** The new coordinates of $\hat{i}$ (originally $[1, 0]$).
* **Column 2:** The new coordinates of $\hat{j}$ (originally $[0, 1]$).



[Image of rotating basis vectors unit circle]


---

### 3. The General Formula
For a counter-clockwise rotation $\theta$:

$$
R_\theta = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix}
$$

---

### 4. Cheat Sheet: Common Angles

| Angle ($\theta$) | Rotation Visual | Resulting Matrix | Memory Hook |
| :--- | :--- | :--- | :--- |
| **90°** | Quarter Turn Left | $\begin{bmatrix} 0 & -1 \\ 1 & 0 \end{bmatrix}$ | Swap $x,y$; make $x$ negative. |
| **180°** | Half Turn | $\begin{bmatrix} -1 & 0 \\ 0 & -1 \end{bmatrix}$ | Negate both $x$ and $y$. |
| **270°** | Three-Quarter Turn | $\begin{bmatrix} 0 & 1 \\ -1 & 0 \end{bmatrix}$ | Swap $x,y$; make $y$ negative. |
| **360°** | Full Circle | $\begin{bmatrix} 1 & 0 \\ 0 & 1 \end{bmatrix}$ | Identity Matrix (No change). |

---

### 5. Calculation: Transforming a Point
To find the new position $(x', y')$ of a point $(x, y)$:

$$
\begin{bmatrix} x' \\ y' \end{bmatrix} = \begin{bmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{bmatrix} \begin{bmatrix} x \\ y \end{bmatrix}
$$

**Algebraic Expansion:**
1. $x' = x\cos\theta - y\sin\theta$
2. $y' = x\sin\theta + y\cos\theta$
