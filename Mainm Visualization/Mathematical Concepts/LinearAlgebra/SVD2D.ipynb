# Singular Value Decomposition (SVD): Geometric Interpretation

## Introduction
Singular Value Decomposition (SVD) is a fundamental technique in linear algebra that breaks down **any** matrix (square or rectangular) into three simpler, distinct transformations. 

If we view a matrix $A$ as a machine that transforms space, SVD tells us that no matter how complex the transformation (shearing, skewing, flipping), it can always be decomposed into a sequence of **Rotation $\rightarrow$ Scaling $\rightarrow$ Rotation**.

### The Core Equation
$$A = U \Sigma V^T$$

Where:
* $A$: The original transformation matrix ($m \times n$).
* $U$: Left Singular Vectors ($m \times m$) $\rightarrow$ **Output Rotation**.
* $\Sigma$: Singular Values ($m \times n$) $\rightarrow$ **Scaling**.
* $V^T$: Right Singular Vectors Transposed ($n \times n$) $\rightarrow$ **Input Rotation**.

---

## 1. The Three Stages of Transformation
The transformation occurs in a specific **Right-to-Left** order (as the vector $x$ hits $V^T$ first in the multiplication $Ax = U\Sigma V^T x$).

1.  **Step 1: Rotate ($V^T$)**
    * **The "Analyzer"**: This is an orthogonal matrix.
    * **Action**: Rotates the input space (grid/basis vectors) to align with the axes along which stretching will occur.
    * **Geometric Meaning**: It finds the "natural" axes of the data.

2.  **Step 2: Scale ($\Sigma$)**
    * **The "Amplifier"**: This is a diagonal matrix containing singular values $\sigma_i$.
    * **Action**: Stretches or shrinks the space strictly along the coordinate axes ($x$ and $y$).
    * **Geometric Meaning**: The unit circle becomes an ellipse. The widths of this ellipse are determined by $\sigma_1, \sigma_2, \dots$.

3.  **Step 3: Rotate ($U$)**
    * **The "Builder"**: This is an orthogonal matrix.
    * **Action**: Rotates the stretched ellipse to its final orientation in the output space.

---

## 2. Deep Dive: Animation Examples

### Example 1: The Symmetric Stretch
**Matrix:**
$$A = \begin{bmatrix} 2 & 1 \\ 1 & 2 \end{bmatrix}$$

This matrix is **symmetric** ($A = A^T$), which implies a special relationship where the input and output rotations are effectively the same (or flipped).

* **Step 1 ($V^T$): Input Rotation**
    $$V^T \approx \begin{bmatrix} -0.71 & -0.71 \\ -0.71 & 0.71 \end{bmatrix}$$
    * **Angle**: $45^\circ$.
    * **Observation**: The matrix stretches space most along the diagonal $y=x$. $V^T$ rotates the standard grid by $45^\circ$ so the x-axis lines up with this diagonal.

* **Step 2 ($\Sigma$): Scaling**
    $$\Sigma = \begin{bmatrix} 3 & 0 \\ 0 & 1 \end{bmatrix}$$
    * **$\sigma_1 = 3$**: The first axis is stretched by a factor of 3.
    * **$\sigma_2 = 1$**: The second axis is unchanged.

* **Step 3 ($U$): Output Rotation**
    $$U \approx \begin{bmatrix} -0.71 & 0.71 \\ -0.71 & -0.71 \end{bmatrix}$$
    * **Angle**: $45^\circ$ (aligns back to diagonal).
    * **Conclusion**: For symmetric matrices, the input rotation aligns the grid, and the output rotation puts it back. $U$ and $V$ are identical (up to sign flips).

---

### Example 2: The Shear (Slanting)
**Matrix:**
$$A = \begin{bmatrix} 1.5 & 1 \\ 0 & 1 \end{bmatrix}$$

A "Shear" looks like pushing the top of a square sideways. SVD reveals this is actually a complex combination of rotations and scaling.

* **Step 1 ($V^T$): Input Rotation**
    $$V^T \approx \begin{bmatrix} -0.85 & -0.53 \\ -0.53 & 0.85 \end{bmatrix}$$
    * **Angle**: $\approx -32^\circ$.
    * **Observation**: To create a shear, we first rotate the square **backwards** (clockwise).

* **Step 2 ($\Sigma$): Scaling**
    $$\Sigma \approx \begin{bmatrix} 2.07 & 0 \\ 0 & 0.72 \end{bmatrix}$$
    * **$\sigma_1 \approx 2.07$**: Significant stretch along X.
    * **$\sigma_2 \approx 0.72$**: **Shrink** along Y.
    * **Observation**: The shape is now a long, thin rectangle.

* **Step 3 ($U$): Output Rotation**
    $$U \approx \begin{bmatrix} -0.97 & -0.26 \\ -0.26 & 0.97 \end{bmatrix}$$
    * **Angle**: $\approx -15^\circ$.
    * **Observation**: We rotate the thin rectangle forward, but only by $15^\circ$.
    * **Conclusion**: The mismatch between the input rotation ($32^\circ$) and output rotation ($15^\circ$) is what creates the visual effect of "slanting" or shearing.

---

## 3. Summary Comparison

| Component | Example 1 (Symmetric) | Example 2 (Shear) | Interpretation |
| :--- | :--- | :--- | :--- |
| **$V^T$** | $45^\circ$ | $\approx -32^\circ$ | **Aligns** inputs to the stretching axes. |
| **$\Sigma$** | $[3, 1]$ | $[2.07, 0.72]$ | **Scales** (Stretches/Squashes). |
| **$U$** | $45^\circ$ | $\approx -15^\circ$ | **Rotates** result to final orientation. |
| **Result** | Input/Output axes align. | Input/Output axes differ. | Difference between $U$ and $V^T$ creates shear/skew. |

```python
from manim import *
import numpy as np

class SVDAnimation(Scene):
    def construct(self):
        # --- PART 1: CONCEPTUAL EXPLANATION ---
        # We show this once at the start to explain the theory
        self.show_equation_and_dimensions()
        
        # Clear the scene to prepare for geometric examples
        self.wait(1)
        self.clear()

        # --- PART 2: GEOMETRIC TRANSFORMATION EXAMPLES ---
        # We will loop through 2 distinct examples to show how SVD handles different matrices
        
        examples = [
            # Example 1: The Symmetric Matrix (Original)
            # Good for showing orthogonal stretching
            (np.array([[2.0, 1.0], [1.0, 2.0]]), "Ex 1: Symmetric Matrix"),
            
            # Example 2: A Shear Matrix
            # Good for showing how 'slanting' is composed of rotations and scaling
            (np.array([[1.5, 1.0], [0.0, 1.0]]), "Ex 2: Shear Matrix")
        ]

        for i, (matrix, title_text) in enumerate(examples):
            # Set the current matrix properties
            self.matrix_A = matrix
            self.u, self.s, self.vh = np.linalg.svd(self.matrix_A)
            self.sigma = np.diag(self.s)
            self.geo_title_text = title_text # Store title for the method to use

            # Run the visualizer
            self.show_geometry_transformation()
            
            # Wait and clear for the next example (unless it's the last one)
            self.wait(2)
            if i < len(examples) - 1:
                self.clear()

    def show_equation_and_dimensions(self):
        # 1. Title
        title = Title("Singular Value Decomposition (SVD)")
        self.add(title)

        # 2. The Main Equation A = U Sigma V^T
        equation = MathTex(
            "A", "=", "U", "\\Sigma", "V^T"
        ).scale(1.5).shift(UP * 2)
        
        equation.set_color_by_tex("A", BLUE)
        equation.set_color_by_tex("U", YELLOW)
        equation.set_color_by_tex("\\Sigma", GREEN)
        equation.set_color_by_tex("V^T", RED)
        
        self.play(Write(equation))
        self.wait(1)

        # 3. Matrix Boxes (Visualizing Dimensions)
        # Let's visualize for an m x n matrix where m > n
        m, n = 2.5, 1.5 
        
        # Helper to create labeled boxes
        def create_matrix_box(label, h, w, color, dim_text):
            box = Rectangle(height=h, width=w, color=color, fill_opacity=0.2, stroke_width=2)
            lbl = MathTex(label).move_to(box.get_center())
            dims = MathTex(dim_text).scale(0.6).next_to(box, DOWN, buff=0.1)
            return VGroup(box, lbl, dims)

        # Create the boxes
        box_A = create_matrix_box("A", m, n, BLUE, "m \\times n")
        eq_sign = MathTex("=").scale(1.5)
        box_U = create_matrix_box("U", m, m, YELLOW, "m \\times m")
        box_S = create_matrix_box("\\Sigma", m, n, GREEN, "m \\times n")
        box_Vt = create_matrix_box("V^T", n, n, RED, "n \\times n")

        # Layout
        visual_group = VGroup(box_A, eq_sign, box_U, box_S, box_Vt).arrange(RIGHT, buff=0.4)
        visual_group.next_to(equation, DOWN, buff=1)

        self.play(FadeIn(visual_group, shift=UP))
        self.wait(1)

        # 4. Explain Roles (Right to Left logic)
        
        # Explanation for V^T
        label_vt = Tex("1. Rotation", color=RED, font_size=34).next_to(box_Vt, UP)
        self.play(Indicate(box_Vt, color=RED), FadeIn(label_vt))
        self.wait(0.5)

        # Explanation for Sigma
        label_sig = Tex("2. Scaling", color=GREEN, font_size=34).next_to(box_S, UP)
        self.play(Indicate(box_S, color=GREEN), FadeIn(label_sig))
        
        # Show specific component info for Sigma
        sigma_detail = Tex("Contains singular values $\sigma_i$", color=GREEN, font_size=24)
        sigma_detail.next_to(box_S, DOWN, buff=0.6)
        self.play(Write(sigma_detail))
        self.wait(1)

        # Explanation for U
        label_u = Tex("3. Rotation", color=YELLOW, font_size=34).next_to(box_U, UP)
        self.play(Indicate(box_U, color=YELLOW), FadeIn(label_u))
        self.wait(2)

        # Clear Part 1
        self.play(
            FadeOut(visual_group), 
            FadeOut(equation), 
            FadeOut(label_vt), FadeOut(label_sig), FadeOut(label_u), 
            FadeOut(sigma_detail),
            FadeOut(title)
        )

    def show_geometry_transformation(self):
        # Title for Part 2
        # Use the title text passed from construct
        if hasattr(self, 'geo_title_text'):
            title_str = self.geo_title_text
        else:
            title_str = "Geometric Interpretation"
            
        geo_title = Title(title_str).to_edge(UP).add_background_rectangle()
        self.add_foreground_mobject(geo_title)

        # 1. Setup Grid and Shapes
        
        # Create the static "ghost" background grid
        # This ensures the original grid is visible with opacity
        background_plane = NumberPlane(
            x_range=[-7, 7, 1],
            y_range=[-5, 5, 1],
            background_line_style={
                "stroke_color": GREY,
                "stroke_width": 1.5,
                "stroke_opacity": 0.3 # Low opacity for the ghost effect
            }
        )
        # Hide the axes of the ghost grid so they don't clutter
        background_plane.get_axes().set_opacity(0)
        self.add(background_plane)

        # Create the main, transformable grid
        main_plane = NumberPlane(
            x_range=[-7, 7, 1],
            y_range=[-5, 5, 1],
            background_line_style={
                "stroke_color": TEAL,
                "stroke_width": 2,
                "stroke_opacity": 0.8
            }
        )
        
        # Add basis vectors i-hat and j-hat
        v_x = Vector([1, 0], color=GREEN)
        v_y = Vector([0, 1], color=RED)
        
        # Add a unit circle (shows deformation best)
        circle = Circle(radius=1, color=WHITE, stroke_opacity=0.8)
        
        # Add a square (shows rotation/shear best)
        square = Square(side_length=2, color=BLUE, fill_opacity=0.2).move_to(ORIGIN)

        # Group everything that will be transformed
        # Note: background_plane is NOT in this group
        transformable_group = VGroup(main_plane, square, circle, v_x, v_y)
        
        self.play(
            Create(main_plane), 
            FadeIn(square), 
            Create(circle), 
            GrowArrow(v_x), 
            GrowArrow(v_y)
        )
        self.wait(0.5)

        # Helper to display current matrix
        def update_matrix_label(name, matrix_val, color):
            # Format matrix numbers
            mat_str = "\\begin{bmatrix} " + f"{matrix_val[0][0]:.2f} & {matrix_val[0][1]:.2f} \\\\ {matrix_val[1][0]:.2f} & {matrix_val[1][1]:.2f}" + " \\end{bmatrix}"
            
            label = MathTex(f"{name} =", mat_str, color=color).add_background_rectangle()
            # Place in top-left, but shifted down to clear the title
            label.to_corner(UL).shift(DOWN * 1.5)
            return label

        # --- STEP 1: V Transpose (Rotation) ---
        # The input space is rotated to align with the axes of stretching
        
        lbl_step1 = update_matrix_label("V^T", self.vh, RED)
        step1_text = Text("Step 1: Align input basis (Rotation)", font_size=24, color=RED).next_to(lbl_step1, DOWN)
        
        self.play(FadeIn(lbl_step1), FadeIn(step1_text))
        
        # Apply matrix V^T
        self.play(
            ApplyMatrix(self.vh, transformable_group),
            run_time=1.5
        )
        self.wait(0.5)
        self.play(FadeOut(lbl_step1), FadeOut(step1_text))

        # --- STEP 2: Sigma (Scaling) ---
        # Stretches space along the coordinate axes
        
        lbl_step2 = update_matrix_label("\\Sigma", self.sigma, GREEN)
        step2_text = Text(f"Step 2: Scale by {self.s[0]:.2f} and {self.s[1]:.2f}", font_size=24, color=GREEN).next_to(lbl_step2, DOWN)
        
        self.play(FadeIn(lbl_step2), FadeIn(step2_text))
        
        # Apply matrix Sigma
        self.play(
            ApplyMatrix(self.sigma, transformable_group),
            run_time=1.5
        )
        self.wait(0.5)
        self.play(FadeOut(lbl_step2), FadeOut(step2_text))

        # --- STEP 3: U (Rotation) ---
        # Rotates the stretched shape to the final output orientation
        
        lbl_step3 = update_matrix_label("U", self.u, YELLOW)
        step3_text = Text("Step 3: Rotate to output (Rotation)", font_size=24, color=YELLOW).next_to(lbl_step3, DOWN)
        
        self.play(FadeIn(lbl_step3), FadeIn(step3_text))
        
        # Apply matrix U
        self.play(
            ApplyMatrix(self.u, transformable_group),
            run_time=1.5
        )
        self.wait(1)
        
        # --- CONCLUSION ---
        # Verify result
        final_lbl = Text("Total Transformation A", font_size=32, color=BLUE).add_background_rectangle()
        # Place it where the last label was
        final_lbl.move_to(lbl_step3.get_center())
        
        self.play(ReplacementTransform(lbl_step3, final_lbl), FadeOut(step3_text))



%manim -qk -v warning SVDAnimation
```
