# 📘 Deep Dive Note: Linear Discriminant Analysis (LDA)

This note provides a rigorous mathematical foundation for **Linear Discriminant Analysis (LDA)**. It parallels visual intuitions often seen in geometric animations, moving from core prerequisites to a step-by-step manual calculation.

---

## 🛠️ Part 1: Prerequisites

To deeply understand LDA, you must be comfortable with the following concepts from Linear Algebra and Statistics.

### 1.1 Linear Algebra
* **Dot Product ($\mathbf{w}^T \mathbf{x}$)**: Geometrically, this represents the **projection** of a data vector $\mathbf{x}$ onto a line defined by vector $\mathbf{w}$. In LDA, we project high-dimensional data onto a 1D line.
* **Eigenvalues & Eigenvectors ($A\mathbf{v} = \lambda\mathbf{v}$)**: An eigenvector $\mathbf{v}$ is a direction that does not change orientation when a linear transformation $A$ is applied; it only scales by $\lambda$. In LDA, the optimal projection direction is an eigenvector.
* **Matrix Inversion ($A^{-1}$)**: Used to "undo" a transformation. In LDA, multiplying by $S_W^{-1}$ acts as a "normalization" step that removes the shape of the data's internal spread (whitening).

### 1.2 Statistics
* **Mean Vector ($\boldsymbol{\mu}$)**: The center of mass of a dataset.
$$\boldsymbol{\mu} = \frac{1}{N} \sum_{i=1}^{N} \mathbf{x}_i$$

* **Covariance Matrix ($\Sigma$)**: Describes how variables change together. The diagonal elements are variances; off-diagonal elements are covariances.
$$\Sigma = \frac{1}{N-1} \sum_{i=1}^{N} (\mathbf{x}_i - \boldsymbol{\mu})(\mathbf{x}_i - \boldsymbol{\mu})^T$$

---

## 🧠 Part 2: The Core Mathematics



LDA seeks a projection vector $\mathbf{w}$ that achieves two competing goals simultaneously:
1.  **Maximize** the distance between the means of the projected classes (Signal).
2.  **Minimize** the spread (variance) within each class after projection (Noise).

This goal is formalized by **Fisher's Criterion**:

$$J(\mathbf{w}) = \frac{\mathbf{w}^T S_B \mathbf{w}}{\mathbf{w}^T S_W \mathbf{w}}$$

Where $S_B$ is the **Between-Class Scatter Matrix** and $S_W$ is the **Within-Class Scatter Matrix**.

### 📉 2.1 The Within-Class Scatter Matrix ($S_W$)
This matrix represents the "fuzziness" or noise inside the classes. We want to **shrink** this. It is calculated as the sum of the scatter matrices for each class:

$$S_W = \sum_{c=1}^{C} S_c$$

Where $S_c$ is the scatter matrix for class $c$:
$$S_c = \sum_{\mathbf{x} \in D_c} (\mathbf{x} - \boldsymbol{\mu}_c)(\mathbf{x} - \boldsymbol{\mu}_c)^T$$

> **Note:** $S_c$ is essentially the unscaled covariance matrix (i.e., we don't divide by $N-1$).

### ↔️ 2.2 The Between-Class Scatter Matrix ($S_B$)
This matrix represents the separation or useful signal. We want to **maximize** this. For a simple 2-class problem, it simplifies to the outer product of the difference between means:

$$S_B = (\boldsymbol{\mu}_1 - \boldsymbol{\mu}_2)(\boldsymbol{\mu}_1 - \boldsymbol{\mu}_2)^T$$

### 💡 2.3 The Solution
To maximize $J(\mathbf{w})$, we take the derivative w.r.t $\mathbf{w}$ and set it to zero. This yields the **Generalized Eigenvalue Problem**:

$$S_B \mathbf{w} = \lambda S_W \mathbf{w}$$

Assuming $S_W$ is invertible (it usually is with enough data), we can rearrange this:

$$(S_W^{-1} S_B) \mathbf{w} = \lambda \mathbf{w}$$

The optimal $\mathbf{w}$ is the eigenvector corresponding to the largest eigenvalue.

**🚀 The Shortcut (For 2 Classes):**
Since $S_B$ is determined entirely by the vector difference of means, the solution simplifies elegantly to:

$$\mathbf{w}^* \propto S_W^{-1}(\boldsymbol{\mu}_1 - \boldsymbol{\mu}_2)$$

## ✍️ Part 3: Worked Calculation Example

Let's calculate the LDA projection step-by-step using a simplified 2D dataset.

**The Dataset:**
* 🟢 **Class 1 ($C_1$):** $\mathbf{x}_1 = (1, 2), \mathbf{x}_2 = (2, 3)$
* 🔴 **Class 2 ($C_2$):** $\mathbf{x}_3 = (3, 1), \mathbf{x}_4 = (4, 2)$

### Step 1: Calculate Mean Vectors
Find the center point for each class.

$$\boldsymbol{\mu}_1 = \frac{1}{2} \begin{bmatrix} 1+2 \\ 2+3 \end{bmatrix} = \begin{bmatrix} 1.5 \\ 2.5 \end{bmatrix}$$

$$\boldsymbol{\mu}_2 = \frac{1}{2} \begin{bmatrix} 3+4 \\ 1+2 \end{bmatrix} = \begin{bmatrix} 3.5 \\ 1.5 \end{bmatrix}$$

### Step 2: Calculate Scatter Matrices ($S_1$ and $S_2$)
For **Class 1**:
$$\mathbf{x}_1 - \boldsymbol{\mu}_1 = \begin{bmatrix} 1 \\ 2 \end{bmatrix} - \begin{bmatrix} 1.5 \\ 2.5 \end{bmatrix} = \begin{bmatrix} -0.5 \\ -0.5 \end{bmatrix}$$

$$\mathbf{x}_2 - \boldsymbol{\mu}_1 = \begin{bmatrix} 2 \\ 3 \end{bmatrix} - \begin{bmatrix} 1.5 \\ 2.5 \end{bmatrix} = \begin{bmatrix} 0.5 \\ 0.5 \end{bmatrix}$$

Now, calculate the outer product sum:

$$S_1 = \begin{bmatrix} -0.5 \\ -0.5 \end{bmatrix}[-0.5 \ -0.5] + \begin{bmatrix} 0.5 \\ 0.5 \end{bmatrix}[0.5 \ 0.5]$$

$$S_1 = \begin{bmatrix} 0.25 & 0.25 \\ 0.25 & 0.25 \end{bmatrix} + \begin{bmatrix} 0.25 & 0.25 \\ 0.25 & 0.25 \end{bmatrix} = \begin{bmatrix} 0.5 & 0.5 \\ 0.5 & 0.5 \end{bmatrix}$$

For **Class 2** (following the same logic):
$$S_2 = \begin{bmatrix} 0.5 & 0.5 \\ 0.5 & 0.5 \end{bmatrix}$$

### Step 3: Within-Class Scatter ($S_W$)
Add the matrices together:

$$S_W = S_1 + S_2 = \begin{bmatrix} 1.0 & 1.0 \\ 1.0 & 1.0 \end{bmatrix}$$

> ⚠️ **Real World Check:** In this simplified example, our points lie perfectly on a line, making $S_W$ **singular** (determinant is 0). In real code, we add a tiny "regularization" term to the diagonal to make it invertible.
>
> Let's approximate: $S_W \approx \begin{bmatrix} 1.1 & 1.0 \\ 1.0 & 1.1 \end{bmatrix}$

### Step 4: Difference of Means
Calculate the vector separating the class centers:

$$\mathbf{d} = \boldsymbol{\mu}_1 - \boldsymbol{\mu}_2 = \begin{bmatrix} 1.5 \\ 2.5 \end{bmatrix} - \begin{bmatrix} 3.5 \\ 1.5 \end{bmatrix} = \begin{bmatrix} -2.0 \\ 1.0 \end{bmatrix}$$

### Step 5: Solve for $\mathbf{w}$
We use the shortcut formula: $\mathbf{w} = S_W^{-1} \mathbf{d}$.

1.  **Find Inverse $S_W^{-1}$:**
    $$\det(S_W) = (1.1)(1.1) - (1)(1) = 0.21$$
    $$S_W^{-1} = \frac{1}{0.21} \begin{bmatrix} 1.1 & -1.0 \\ -1.0 & 1.1 \end{bmatrix} \approx \begin{bmatrix} 5.2 & -4.7 \\ -4.7 & 5.2 \end{bmatrix}$$

2.  **Multiply Matrix by Difference Vector:**
    $$\mathbf{w} = \begin{bmatrix} 5.2 & -4.7 \\ -4.7 & 5.2 \end{bmatrix} \begin{bmatrix} -2.0 \\ 1.0 \end{bmatrix}$$

    $$w_x = (5.2)(-2) + (-4.7)(1) = -10.4 - 4.7 = -15.1$$
    $$w_y = (-4.7)(-2) + (5.2)(1) = 9.4 + 5.2 = 14.6$$

### 🎯 Final Result
$$\mathbf{w} \approx \begin{bmatrix} -15.1 \\ 14.6 \end{bmatrix}$$

This vector points roughly towards $(-1, 1)$. This makes sense! Our data was spread diagonally along $(1, 1)$, so to separate the classes best, LDA rotates the projection axis to be **perpendicular** to the variance.

```python
from manim import *
import numpy as np

class LDAExplained(Scene):
    def construct(self):
        # --- CONFIGURATION ---
        np.random.seed(42)
        
        # Colors
        C1_COLOR = TEAL
        C2_COLOR = MAROON
        PROJ_COLOR = YELLOW
        MEAN_COLOR = WHITE
        GOOD_COLOR = GREEN
        BAD_COLOR = RED
        MATH_COLOR = BLUE_B
        
        # Layout Constants
        TEXT_SCALE = 0.6
        
        # --- TITLE ---
        title = Title(r"LDA Step-by-Step: Scatter Matrices \& Eigenvalues", include_underline=True)
        self.add(title)
        
        # --- LAYOUT SETUP ---
        plane = NumberPlane(
            x_range=[-4, 8, 1],
            y_range=[-4, 6, 1],
            background_line_style={"stroke_opacity": 0.3}
        ).scale(0.8).shift(RIGHT * 4.5) 
        
        # Left side: Text Container
        info_text_group = VGroup()
        
        # Helper to handle colored lines
        def update_text(*lines):
            nonlocal info_text_group
            new_group = VGroup()
            
            # Keywords to colorize automatically
            kw_map = {
                "S_W": C1_COLOR,
                "S_B": GOOD_COLOR,
                r"\lambda": PROJ_COLOR,
                r"\mathbf{w}": PROJ_COLOR
            }
            
            for line_data in lines:
                if isinstance(line_data, tuple):
                    content, col = line_data
                else:
                    content, col = line_data, WHITE

                if len(content.strip()) == 0:
                    new_group.add(Rectangle(height=0.15, width=0.1, fill_opacity=0, stroke_opacity=0))
                    continue
                
                # Check if it's a pure math line (starts/ends with $)
                if content.strip().startswith("$") and content.strip().endswith("$"):
                    tex_string = content.strip()[1:-1]
                    
                    # FIX: Do NOT apply color map if the string contains complex structures like \frac
                    # This prevents the "Context of error: ) = \frac{}" crash
                    if "\\frac" in tex_string:
                        new_group.add(MathTex(tex_string, color=col))
                    else:
                        new_group.add(MathTex(tex_string, color=col, tex_to_color_map=kw_map))
                else:
                    new_group.add(Tex(content, color=col))
            
            new_group.arrange(DOWN, aligned_edge=LEFT, buff=0.15)
            new_group.scale(TEXT_SCALE).to_edge(LEFT, buff=0.5).shift(UP*0.5)
            
            anims = [FadeIn(new_group, shift=RIGHT*0.5)]
            if len(info_text_group) > 0:
                anims.append(FadeOut(info_text_group, shift=RIGHT*0.1))
            
            self.play(*anims)
            info_text_group = new_group

        self.play(Create(plane), run_time=1)

        # --- DATA GENERATION ---
        cov = [[0.8, 0.6], [0.6, 0.8]] 
        mean1, mean2 = [1, 2], [4, 0]
        N = 40
        data1 = np.random.multivariate_normal(mean1, cov, N)
        data2 = np.random.multivariate_normal(mean2, cov, N)
        
        dots1 = VGroup(*[Dot(plane.c2p(x, y), color=C1_COLOR, radius=0.07) for x, y in data1])
        dots2 = VGroup(*[Dot(plane.c2p(x, y), color=C2_COLOR, radius=0.07) for x, y in data2])
        mu1_dot = Dot(plane.c2p(*mean1), color=MEAN_COLOR, radius=0.1)
        mu2_dot = Dot(plane.c2p(*mean2), color=MEAN_COLOR, radius=0.1)

        # Pre-compute Matrices
        s1 = np.cov(data1.T) * (N-1)
        s2 = np.cov(data2.T) * (N-1)
        sw = s1 + s2
        
        diff_means = np.array(mean1) - np.array(mean2)
        sb = np.outer(diff_means, diff_means)
        
        sw_inv = np.linalg.inv(sw)
        matrix_A = sw_inv @ sb
        
        eig_vals_A, eig_vecs_A = np.linalg.eig(matrix_A)
        sorted_indices = np.argsort(eig_vals_A.real)[::-1]
        lda_eigval = eig_vals_A[sorted_indices[0]].real
        lda_eigvec = eig_vecs_A[:, sorted_indices[0]].real
        lda_eigvec = lda_eigvec / np.linalg.norm(lda_eigvec)

        # =========================================
        # --- PART 1: SETUP & GOAL ---
        # =========================================
        update_text(
            r"\textbf{1. The Goal}",
            r"Find a projection direction $\mathbf{w}$",
            r"that best separates the classes.",
            " ",
            ("Class 1 (Teal)", C1_COLOR),
            ("Class 2 (Maroon)", C2_COLOR),
        )
        self.play(LaggedStart(FadeIn(dots1), FadeIn(dots2), lag_ratio=0.01))
        self.play(Create(mu1_dot), Create(mu2_dot))
        self.wait(2)

        # =========================================
        # --- PART 2: WITHIN-CLASS SCATTER ($S_W$) ---
        # =========================================
        update_text(
            r"\textbf{2. Within-Class Scatter ($S_W$)}",
            r"Measure the 'spread' \textit{inside} each class.",
            r"We want to MINIMIZE this.",
            " ",
            r"$S_W = S_1 + S_2$",
            r"where $S_i = \sum (x - \mu_i)(x - \mu_i)^T$"
        )
        
        def get_cov_ellipse(cov_matrix, mean, color):
            vals, vecs = np.linalg.eigh(cov_matrix)
            angle = np.arctan2(vecs[1, 1], vecs[1, 0])
            width = 2 * np.sqrt(vals[1]) * 0.5 
            height = 2 * np.sqrt(vals[0]) * 0.5
            ell = Ellipse(width=width, height=height, color=color, stroke_opacity=0.8)
            ell.rotate(angle).move_to(plane.c2p(*mean))
            return ell

        ell1 = get_cov_ellipse(s1, mean1, C1_COLOR)
        ell2 = get_cov_ellipse(s2, mean2, C2_COLOR)
        
        self.play(Create(ell1), Create(ell2))
        self.wait(3)

        # =========================================
        # --- PART 3: BETWEEN-CLASS SCATTER ($S_B$) ---
        # =========================================
        update_text(
            r"\textbf{3. Between-Class Scatter ($S_B$)}",
            r"Measure distance between class means.",
            r"We want to MAXIMIZE this.",
            " ",
            r"$S_B = (\mu_1 - \mu_2)(\mu_1 - \mu_2)^T$",
            ("It's spread along the line connecting means.", GOOD_COLOR)
        )

        mean_center = (np.array(mean1) + np.array(mean2)) / 2
        vec_between = Arrow(
            plane.c2p(*mean2), 
            plane.c2p(*mean1), 
            color=GOOD_COLOR, buff=0, stroke_width=6
        )
        lbl_between = Tex(r"$\mu_1 - \mu_2$", color=GOOD_COLOR).next_to(vec_between, UP).scale(0.7)

        self.play(FadeOut(ell1), FadeOut(ell2))
        self.play(GrowFromCenter(vec_between), Write(lbl_between))
        self.wait(3)

        # =========================================
        # --- PART 4: FISHER'S CRITERION ---
        # =========================================
        # The equation containing \frac will now be rendered safely in white
        update_text(
            r"\textbf{4. The Optimization Problem}",
            r"Maximize Fisher's Ratio $J(\mathbf{w})$:",
            r"$J(\mathbf{w}) = \frac{\mathbf{w}^T S_B \mathbf{w}}{\mathbf{w}^T S_W \mathbf{w}}$",
            " ",
            r"Maximizing this leads to a",
            ("Generalized Eigenvalue Problem:", MATH_COLOR),
            r"$S_B \mathbf{w} = \lambda S_W \mathbf{w}$"
        )
        self.wait(4)

        update_text(
            r"\textbf{5. Solving for $\mathbf{w}$}",
            r"Assuming $S_W$ is invertible,",
            r"we rearrange to standard form:",
            " ",
            r"$(S_W^{-1} S_B) \mathbf{w} = \lambda \mathbf{w}$",
            " ",
            r"The optimal direction $\mathbf{w}$ is the",
            ("eigenvector with the largest eigenvalue.", PROJ_COLOR)
        )
        self.play(FadeOut(vec_between), FadeOut(lbl_between))
        self.wait(3)

        # =========================================
        # --- PART 5: VISUALIZING THE SOLUTION ---
        # =========================================
        center_pt = plane.c2p(*mean_center)
        viz_scale = 3.0
        eig_arrow = Arrow(
            center_pt,
            center_pt + np.array([lda_eigvec[0], lda_eigvec[1], 0]) * viz_scale,
            color=PROJ_COLOR, buff=0, stroke_width=8
        )
        eig_line_full = Line(
            center_pt - np.array([lda_eigvec[0], lda_eigvec[1], 0]) * 6,
            center_pt + np.array([lda_eigvec[0], lda_eigvec[1], 0]) * 6,
            color=PROJ_COLOR, stroke_opacity=0.5
        )
        
        eig_label = Tex(r"Optimal $\mathbf{w}$", color=PROJ_COLOR).next_to(eig_arrow.get_end(), UP).scale(0.7)

        self.play(GrowFromPoint(eig_arrow, center_pt), Write(eig_label))
        self.wait()
        self.play(Create(eig_line_full), FadeOut(eig_arrow))

        update_text(
            r"\textbf{6. The Result}",
            r"This eigenvector direction balances",
            r"maximizing separation vs minimizing internal spread.",
            " ",
            ("Projecting data onto this line...", PROJ_COLOR)
        )
        self.wait(2)

        # =========================================
        # --- PART 6: FINAL PROJECTION ---
        # =========================================
        anims = []
        for dot in dots1:
            anims.append(dot.animate.move_to(eig_line_full.get_projection(dot.get_center())))
        for dot in dots2:
            anims.append(dot.animate.move_to(eig_line_full.get_projection(dot.get_center())))
        
        anims.append(mu1_dot.animate.move_to(eig_line_full.get_projection(mu1_dot.get_center())))
        anims.append(mu2_dot.animate.move_to(eig_line_full.get_projection(mu2_dot.get_center())))
        
        self.play(*anims, FadeOut(eig_label), run_time=2.5, rate_func=smooth)
        
        p1 = eig_line_full.get_projection(plane.c2p(*mean1))
        p2 = eig_line_full.get_projection(plane.c2p(*mean2))
        brace = BraceBetweenPoints(p1, p2, UP, color=PROJ_COLOR)
        b_text = brace.get_text("Maximized Fisher Ratio").scale(0.6).set_color(PROJ_COLOR)
        
        self.play(GrowFromCenter(brace), Write(b_text))
        self.wait(5)


%manim -qk -v warning LDAExplained
```
