# Vector Calculus Visualization Notes

## 1. General Formulas

Given a vector field $\vec{F}(x,y,z) = P\mathbf{\hat{i}} + Q\mathbf{\hat{j}} + R\mathbf{\hat{k}} = \langle P, Q, R \rangle$:

### **Divergence (Scalar)**
Measures the net rate of flow **out** of a point (Expansion/Compression).

$$
\nabla \cdot \vec{F} = \frac{\partial P}{\partial x} + \frac{\partial Q}{\partial y} + \frac{\partial R}{\partial z}
$$

### **Curl (Vector)**
Measures the local **rotation** or "swirl" at a point.

$$
\nabla \times \vec{F} = \begin{vmatrix} \mathbf{\hat{i}} & \mathbf{\hat{j}} & \mathbf{\hat{k}} \\ \frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\ P & Q & R \end{vmatrix}
$$

---

## 2. Divergence Examples (Derivations)

### **Example 1: Positive Divergence (3D Source)**
* **Field:** $\vec{F} = \langle x, y, z \rangle$
* **Visual:** An explosion. Particles accelerate outward in all directions.
* **Derivation:**
  $$
  \frac{\partial}{\partial x}(x) + \frac{\partial}{\partial y}(y) + \frac{\partial}{\partial z}(z) = 1 + 1 + 1 = \mathbf{3}
  $$
* **Result:** Positive scalar ($+3$). Net **outflow** everywhere.

### **Example 2: Negative Divergence (3D Sink)**
* **Field:** $\vec{F} = \langle -x, -y, -z \rangle$
* **Visual:** An implosion. Particles collapse inward toward the origin.
* **Derivation:**
  $$
  \frac{\partial}{\partial x}(-x) + \frac{\partial}{\partial y}(-y) + \frac{\partial}{\partial z}(-z) = -1 + (-1) + (-1) = \mathbf{-3}
  $$
* **Result:** Negative scalar ($-3$). Net **inflow** everywhere.

### **Example 3: Zero Divergence (Pure Rotation)**
* **Field:** $\vec{F} = \langle -y, x, 0 \rangle$
* **Visual:** Particles circulate around the Z-axis (Tornado). No fluid is created or destroyed.
* **Derivation:**
  $$
  \frac{\partial}{\partial x}(-y) + \frac{\partial}{\partial y}(x) + \frac{\partial}{\partial z}(0) = 0 + 0 + 0 = \mathbf{0}
  $$
* **Result:** Zero. The fluid is **incompressible**.

### **Example 4: 1D Expansion**
* **Field:** $\vec{F} = \langle x, 0, 0 \rangle$
* **Visual:** Stretching along the X-axis only (like a rubber band).
* **Derivation:**
  $$
  \frac{\partial}{\partial x}(x) + \frac{\partial}{\partial y}(0) + \frac{\partial}{\partial z}(0) = 1 + 0 + 0 = \mathbf{1}
  $$
* **Result:** Positive ($+1$). Expansion occurs only in one direction.

### **Example 5: 1D Compression**
* **Field:** $\vec{F} = \langle -x, 0, 0 \rangle$
* **Visual:** Squishing along the X-axis only.
* **Derivation:**
  $$
  \frac{\partial}{\partial x}(-x) + \frac{\partial}{\partial y}(0) + \frac{\partial}{\partial z}(0) = -1 + 0 + 0 = \mathbf{-1}
  $$
* **Result:** Negative ($-1$). Compression occurs in one direction.

---

## 3. Curl Examples (Derivations)

### **Example 1: Zero Curl (Radial Expansion)**
* **Field:** $\vec{F} = \langle x, y, z \rangle$
* **Visual:** Particles move straight out. There is no spinning.
* **Derivation:**
  $$
  \begin{vmatrix} \mathbf{\hat{i}} & \mathbf{\hat{j}} & \mathbf{\hat{k}} \\ \partial_x & \partial_y & \partial_z \\ x & y & z \end{vmatrix}
  $$
  * $\mathbf{\hat{i}}$ component: $\frac{\partial z}{\partial y} - \frac{\partial y}{\partial z} = 0 - 0 = 0$
  * $\mathbf{\hat{j}}$ component: $\frac{\partial z}{\partial x} - \frac{\partial x}{\partial z} = 0 - 0 = 0$
  * $\mathbf{\hat{k}}$ component: $\frac{\partial y}{\partial x} - \frac{\partial x}{\partial y} = 0 - 0 = 0$
* **Result:** $\vec{0}$. The field is **irrotational**.

### **Example 2: Positive Z-Curl (Standard Spin)**
* **Field:** $\vec{F} = \langle -y, x, 0 \rangle$
* **Visual:** Counter-Clockwise (CCW) rotation around the Z-axis.
* **Derivation:**
  Focus on the $\mathbf{\hat{k}}$ component (rotation in xy-plane):
  $$
  (\nabla \times \vec{F})_z = \frac{\partial Q}{\partial x} - \frac{\partial P}{\partial y}
  $$
  $$
  = \frac{\partial}{\partial x}(x) - \frac{\partial}{\partial y}(-y) = 1 - (-1) = \mathbf{2}
  $$
* **Result:** $\langle 0, 0, 2 \rangle$. Positive Z-component indicates **CCW** rotation.

### **Example 3: Negative Z-Curl (Reverse Spin)**
* **Field:** $\vec{F} = \langle y, -x, 0 \rangle$
* **Visual:** Clockwise (CW) rotation around the Z-axis.
* **Derivation:**
  $$
  (\nabla \times \vec{F})_z = \frac{\partial}{\partial x}(-x) - \frac{\partial}{\partial y}(y) = -1 - (1) = \mathbf{-2}
  $$
* **Result:** $\langle 0, 0, -2 \rangle$. Negative Z-component indicates **CW** rotation.

### **Example 4: Y-Axis Rotation**
* **Field:** $\vec{F} = \langle z, 0, -x \rangle$
* **Visual:** Particles spin around the vertical Y-axis.
* **Derivation:**
  $$
  \begin{vmatrix} \mathbf{\hat{i}} & \mathbf{\hat{j}} & \mathbf{\hat{k}} \\ \partial_x & \partial_y & \partial_z \\ z & 0 & -x \end{vmatrix}
  $$
  * $\mathbf{\hat{i}}$: $\partial_y(-x) - \partial_z(0) = 0$
  * $\mathbf{\hat{j}}$: $-[\partial_x(-x) - \partial_z(z)] = -[-1 - 1] = -[-2] = \mathbf{2}$
  * $\mathbf{\hat{k}}$: $\partial_x(0) - \partial_y(z) = 0$
* **Result:** $\langle 0, 2, 0 \rangle$. Rotation is purely around the **Y-axis**.

### **Example 5: Shear Flow**
* **Field:** $\vec{F} = \langle y, 0, 0 \rangle$
* **Visual:** Horizontal flow that gets faster as $y$ increases. A paddle wheel placed here would spin because the top pushes harder than the bottom.
* **Derivation:**
  $$
  (\nabla \times \vec{F})_z = \frac{\partial}{\partial x}(0) - \frac{\partial}{\partial y}(y) = 0 - 1 = \mathbf{-1}
  $$
* **Result:** $\langle 0, 0, -1 \rangle$. Even though paths are straight lines, the **speed difference** causes a Clockwise rotation (Curl).

```python
from manim import *
import numpy as np

class VectorCalculusStudy(ThreeDScene):
    def construct(self):
        # 1. SHOW FORMULAS (Stacked Layout)
        self.show_formulas_stacked()
        
        # 2. SETUP 3D SCENE
        self.set_camera_orientation(phi=75 * DEGREES, theta=30 * DEGREES)
        self.axes = ThreeDAxes(x_range=[-3,3], y_range=[-3,3], z_range=[-3,3])
        self.add(self.axes)
        self.begin_ambient_camera_rotation(rate=0.15) 

        # ==========================================
        # PART 1: 5 DIVERGENCE EXAMPLES
        # ==========================================
        div_scenarios = [
            {
                "name": "Div 1: Positive (3D Source)",
                "func": lambda p: np.array([p[0], p[1], p[2]]), 
                "eq": r"\vec{F} = \langle x, y, z \rangle",
                "calc": r"\nabla \cdot \vec{F} = 3",
                "desc": "Explosion outward in all directions",
                "color": YELLOW
            },
            {
                "name": "Div 2: Negative (3D Sink)",
                "func": lambda p: np.array([-p[0], -p[1], -p[2]]),
                "eq": r"\vec{F} = \langle -x, -y, -z \rangle",
                "calc": r"\nabla \cdot \vec{F} = -3",
                "desc": "Implosion inward to origin",
                "color": RED
            },
            {
                "name": "Div 3: Zero (Incompressible)",
                "func": lambda p: np.array([-p[1], p[0], 0]),
                "eq": r"\vec{F} = \langle -y, x, 0 \rangle",
                "calc": r"\nabla \cdot \vec{F} = 0",
                "desc": "Fluid circulates but does not expand",
                "color": BLUE
            },
            {
                "name": "Div 4: 1D Expansion (Linear)",
                "func": lambda p: np.array([p[0], 0, 0]),
                "eq": r"\vec{F} = \langle x, 0, 0 \rangle",
                "calc": r"\nabla \cdot \vec{F} = 1",
                "desc": "Stretching along X-axis only",
                "color": GREEN
            },
            {
                "name": "Div 5: 1D Compression",
                "func": lambda p: np.array([-p[0], 0, 0]),
                "eq": r"\vec{F} = \langle -x, 0, 0 \rangle",
                "calc": r"\nabla \cdot \vec{F} = -1",
                "desc": "Squishing along X-axis only",
                "color": ORANGE
            },
        ]

        self.next_section("Divergence Examples")
        for data in div_scenarios:
            self.visualize_scenario(data)

        # ==========================================
        # PART 2: 5 CURL EXAMPLES
        # ==========================================
        curl_scenarios = [
            {
                "name": "Curl 1: Zero (Radial Source)",
                "func": lambda p: np.array([p[0], p[1], p[2]]), 
                "eq": r"\vec{F} = \langle x, y, z \rangle",
                "calc": r"\nabla \times \vec{F} = \vec{0}",
                "desc": "Expands but has NO rotation",
                "color": GREY
            },
            {
                "name": "Curl 2: Positive Z-Curl (CCW)",
                "func": lambda p: np.array([-p[1], p[0], 0]),
                "eq": r"\vec{F} = \langle -y, x, 0 \rangle",
                "calc": r"\nabla \times \vec{F} = \langle 0, 0, 2 \rangle",
                "desc": "Standard Counter-Clockwise Spin",
                "color": TEAL
            },
            {
                "name": "Curl 3: Negative Z-Curl (CW)",
                "func": lambda p: np.array([p[1], -p[0], 0]),
                "eq": r"\vec{F} = \langle y, -x, 0 \rangle",
                "calc": r"\nabla \times \vec{F} = \langle 0, 0, -2 \rangle",
                "desc": "Clockwise Spin",
                "color": MAROON
            },
            {
                "name": "Curl 4: Y-Axis Rotation",
                "func": lambda p: np.array([p[2], 0, -p[0]]), # <z, 0, -x>
                "eq": r"\vec{F} = \langle z, 0, -x \rangle",
                "calc": r"\nabla \times \vec{F} = \langle 0, 2, 0 \rangle",
                "desc": "Spinning around the Y-axis",
                "color": PURPLE
            },
            {
                "name": "Curl 5: Shear Flow",
                "func": lambda p: np.array([p[1], 0, 0]), # <y, 0, 0>
                "eq": r"\vec{F} = \langle y, 0, 0 \rangle",
                "calc": r"\nabla \times \vec{F} = \langle 0, 0, -1 \rangle",
                "desc": "Straight lines, but paddle wheel spins!",
                "color": PINK
            },
        ]

        self.next_section("Curl Examples")
        for data in curl_scenarios:
            self.visualize_scenario(data)

    # ----------------------------------------------------
    # VISUALIZATION FUNCTIONS
    # ----------------------------------------------------

    def show_formulas_stacked(self):
        """Displays Divergence on top and Curl matrix on bottom"""
        
        # 1. Divergence (Top)
        div_label = Text("Divergence (Scalar)", font_size=36, color=YELLOW)
        div_eq = MathTex(
            r"\nabla \cdot \vec{F} = \frac{\partial P}{\partial x} + \frac{\partial Q}{\partial y} + \frac{\partial R}{\partial z}"
        , font_size=40)
        div_group = VGroup(div_label, div_eq).arrange(DOWN, buff=0.3)

        # 2. Curl (Bottom - Matrix Form)
        curl_label = Text("Curl (Vector)", font_size=36, color=BLUE)
        curl_eq = MathTex(
            r"\nabla \times \vec{F} = \begin{vmatrix} \mathbf{\hat{i}} & \mathbf{\hat{j}} & \mathbf{\hat{k}} \\ \frac{\partial}{\partial x} & \frac{\partial}{\partial y} & \frac{\partial}{\partial z} \\ P & Q & R \end{vmatrix}"
        , font_size=40)
        curl_group = VGroup(curl_label, curl_eq).arrange(DOWN, buff=0.3)

        # Stack Groups Vertically
        full_layout = VGroup(div_group, curl_group).arrange(DOWN, buff=1.0)
        
        self.add_fixed_in_frame_mobjects(full_layout)
        self.play(Write(full_layout), run_time=2)
        self.wait(4)
        self.play(FadeOut(full_layout))
        self.remove(full_layout) 

    def visualize_scenario(self, data):
        # 1. HUD (Heads-Up Display)
        header = Text(data["name"], font_size=26, color=data["color"]).to_corner(UL)
        eq_tex = MathTex(data["eq"], font_size=32).next_to(header, DOWN, aligned_edge=LEFT)
        calc_tex = MathTex(data["calc"], font_size=32).next_to(eq_tex, DOWN, aligned_edge=LEFT)
        desc_tex = Text(data["desc"], font_size=22, color=GRAY).next_to(calc_tex, DOWN, aligned_edge=LEFT)
        
        # Background for text
        hud_content = VGroup(header, eq_tex, calc_tex, desc_tex)
        bg = BackgroundRectangle(hud_content, fill_opacity=0.85, buff=0.2)
        
        self.add_fixed_in_frame_mobjects(bg, header, eq_tex, calc_tex, desc_tex)
        self.play(FadeIn(bg), Write(hud_content), run_time=1)

        # 2. 3D Vector Field Arrows
        # Using sigmoid length_func to keep arrows from getting too huge at edges
        vector_field = ArrowVectorField(
            data["func"],
            x_range=[-2, 2, 1],
            y_range=[-2, 2, 1],
            z_range=[-2, 2, 1],
            colors=[data["color"], WHITE],
            length_func=lambda norm: 0.5 * sigmoid(norm)
        )
        self.play(Create(vector_field))

        # 3. Flow Particles (Dots)
        particles = VGroup()
        for _ in range(40):
            pos = np.random.uniform(-2, 2, size=3)
            dot = Dot3D(point=pos, radius=0.06, color=data["color"])
            particles.add(dot)
            
        def update_particles(mob, dt):
            for dot in mob:
                p = dot.get_center()
                v = data["func"](p)
                dot.shift(v * dt * 0.8) 
                
                # Recycle particles if they leave view
                if np.linalg.norm(p) > 3.5:
                    dot.move_to(np.random.uniform(-1.5, 1.5, size=3))

        particles.add_updater(update_particles)
        self.add(particles)

        # 4. Wait for viewer to absorb concept
        self.wait(4)

        # 5. Cleanup
        particles.remove_updater(update_particles)
        self.remove(particles)
        self.play(FadeOut(vector_field), FadeOut(hud_content), FadeOut(bg))


%manim -ql -v warning VectorCalculusStudy
```
